## Cel

Znaleźć i naprawić błąd priorytetu działań w gotowym wzorze.

## Typowy błąd

Ten wzór miał policzyć średnią trzech ocen, ale zapomniał nawiasów — wynik jest błędny.

```python
ocenka1, ocenka2, ocenka3 = 4, 5, 3
srednyaya_s_oshibkoj = ocenka1 + ocenka2 + ocenka3 / 3
print(srednyaya_s_oshibkoj)   # 10.0 — неверно! это НЕ среднее трёх оценок
```

## Podstawowa praktyka zadań ★

Popraw wzór tak, aby poprawnie obliczał średnią trzech punktów (dodaj nawiasy wokół sumy). Oczekiwane jest 4,0.

```python
ocenka1, ocenka2, ocenka3 = 4, 5, 3
srednyaya = (ocenka1 + ocenka2 + ocenka3) / 3
print(srednyaya)
```
