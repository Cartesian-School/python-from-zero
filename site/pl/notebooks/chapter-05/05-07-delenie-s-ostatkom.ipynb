## Cel

Opanować //, % i divmod() na rzeczywistych zadaniach grupowania.

## Sprawa robocza

```python
candies = 17
group_size = 5
print(candies // group_size)
print(candies % group_size)
```

## Eksperyment 1

Check divmod() – powinno zwracać obie liczby jednocześnie, jak krotka.

```python
print(divmod(17, 5))
```

## Podstawowa praktyka zadań ★

22 uczestników zostało podzielonych na zespoły po 4 osoby. Ile będzie kompletnych drużyn, a ile osób zostanie bez własnego zespołu? Zwróć obie liczby oddzielone spacją w jednym print().

```python
uchastniki = 22
razmer_komandy = 4
komand, ostatok = divmod(uchastniki, razmer_komandy)
print(komand, ostatok)
```
