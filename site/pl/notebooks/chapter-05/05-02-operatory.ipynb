## Cel

Master //, % ** oraz skrócone operatory przydziałów.

## Sprawa robocza

```python
print(17 // 5)
print(17 % 5)
print(2 ** 10)
```

## Eksperyment 1

Sprawdź parzystość kilku liczb przez `% 2`.

```python
for n in [4, 7, 10, 13]:
    print(n, n % 2 == 0)
```

## Eksperyment 2

Zastosowanie `+=` do gromadzenia wyniku w grze przez trzy ruchy.

```python
score = 0
score += 10
score += 25
score += 7
print(score)
```

## Typowy błąd

Łatwo pomylić `//` (dzielenie całkowite) i `/` (zwyczajne) – dają różne rodzaje wyników.

```python
print(10 // 3, type(10 // 3))
print(10 / 3, type(10 / 3))
```

## Podstawowa praktyka zadań ★

Sprawdź kolejność obliczeń: liczenie `2 + 3 * 4 ** 2` i porównaj to ze swoją prognozą na papierze.

```python
print(2 + 3 * 4 ** 2)
```

## Sprawdzenie wyniku

```python
assert 2 + 3 * 4 ** 2 == 50
print("Верно: 4 ** 2 = 16, 3 * 16 = 48, 48 + 2 = 50")
```
