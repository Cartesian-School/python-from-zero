## Cel

Opanuj moduł `math`: floor, ceil, sqrt, factorial, trygonometrię.

## Sprawa robocza

```python
import math

print(math.floor(4.7))
print(math.ceil(4.2))
print(math.sqrt(64))
```

## Eksperyment 1

Oblicz czynniki kilku liczb z rzędu.

```python
import math

for n in range(1, 6):
    print(n, "! =", math.factorial(n))
```

## Eksperyment 2

`round()` — funkcja wbudowana (nie z math). Porównaj ją z floor/ceil na tej samej liczbie.

```python
import math

x = 4.5
print(round(x), math.floor(x), math.ceil(x))
```

## Podstawowa praktyka zadań ★

Znajdź długość przeciwprostokątnej trójkąta prostokątnego o przyprostokątnych 3 i 4, używając `math.sqrt`.

```python
import math

a, b = 3, 4
hypotenuse = math.sqrt(a ** 2 + b ** 2)
print(hypotenuse)
```

## Sprawdzenie wyniku

```python
import math

assert math.sqrt(3 ** 2 + 4 ** 2) == 5.0
print("Верно: классический треугольник 3-4-5.")
```
