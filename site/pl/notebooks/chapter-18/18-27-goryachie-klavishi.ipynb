## Cel

Przypisz Ctrl+Z/Ctrl+Y do cofania i ponownego robienia na poziomie okna.

## Pro mainloop() w tym laptopie

Zamiast `root.mainloop()` używany tutaj `root.update()` + `root.destroy()` jest takie, że aplikacja jest zbudowana dokładnie tak samo, tylko bez niekończącego się oczekiwania na końcu (zobacz wyjaśnienie w rozdziale 16).

## Sprawa robocza

```python
import tkinter as tk

root = tk.Tk()
log = []

def undo(_event=None):
    log.append("undo")

def redo(_event=None):
    log.append("redo")

root.bind("<Control-z>", undo)
root.bind("<Control-y>", redo)
root.update()

# UWAGA: root.bind(„<Control-z>”
# nieistniejącym wiązaniem jest pusty łańcuch znaków „”
# bool(...), a nie przez `is not None`: pusta łańcuch znaków też „nie None”!
print("Ctrl+Z привязан:", bool(root.bind("<Control-z>")))
print("Ctrl+Q не привязан:", bool(root.bind("<Control-q>")))

# Same obsługujące funkcje wywołujemy bezpośrednio — jakby klawisze zostały już naciśnięte
undo()
redo()
undo()
print("Журнал вызовов:", log)
root.destroy()
```

## Sprawdzenie wyniku

```python
assert log == ["undo", "redo", "undo"], "порядок вызовов должен совпадать с порядком нажатий"
print("Верно: обработчики отработали в том же порядке, в каком их вызвали.")
```

## Zadanie niezależne od zadania ★★
