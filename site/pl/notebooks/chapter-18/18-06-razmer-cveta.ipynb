## O mainloop() i wydarzeniach myszy na tym laptopie

Zamiast `root.mainloop()` używany tutaj `root.update()` + `root.destroy()`. A zamiast prawdziwych kliknięć myszką wywołujemy funkcje-obsługujące bezpośrednio z małym obiektem `FakeEvent(x, y)` zamiast rzeczywistego zdarzenia, Tkinter w rzeczywistej aplikacji przekazuje obiekt o tych samych polach `.x`/`.y`, więc kod obsługi sprawdzany jest naprawdę.

## Przykład roboczy – Scale

```python
import tkinter as tk

root = tk.Tk()
tolschina = 3

def vybrat_tolschinu(znachenie):
    global tolschina
    tolschina = int(znachenie)

scale = tk.Scale(root, from_=1, to=10, orient="horizontal", command=vybrat_tolschinu)
scale.pack()   # ВАЖНО: пока виджет не размещён на экране, Tk не вызывает command
scale.set(7)
root.update()
print("Толщина:", tolschina)   # 7 — значение ползунка дошло до переменной
root.destroy()
```

## Sprawdzenie wyniku

Jeśli usuniesz `scale.pack()`, Handler `vybrat_tolschinu` w ogóle nie zostanie wezwana, i `tolschina` pozostaje 3 – spróbuj, to dobra ilustracja tego, co `command` jest związany z żywym widżetem na ekranie, a nie z obiektem w pamięci.

```python
assert tolschina == 7, "command у Scale должен был обновить tolschina до 7"
print("Верно: ползунок передал значение 7 в переменную через command.")
```

## Eksperyment 1 – Wybieracz kolorów w pętli

```python
root = tk.Tk()
toolbar = tk.Frame(root)
toolbar.pack()

tekuschij_cvet = "black"

def vybrat_cvet(cvet):
    global tekuschij_cvet
    tekuschij_cvet = cvet

cveta = ["black", "red", "blue", "green", "orange", "purple"]
knopki = []
for cvet in cveta:
    b = tk.Button(toolbar, bg=cvet, width=2, command=lambda c=cvet: vybrat_cvet(c))
    b.pack(side="left")
    knopki.append(b)

root.update()
knopki[3].invoke()  # "зелёный" — четвёртая кнопка
print("Выбранный цвет:", tekuschij_cvet)
root.destroy()
```

## Sprawdzenie wyniku

```python
assert tekuschij_cvet == "green"
print("Верно: четвёртая кнопка палитры выбрала зелёный.")
```
