## Cel

Scale grubość linii związanej ze stanem rysunku.

## Pro mainloop() w tym laptopie

Zamiast `root.mainloop()` używany tutaj `root.update()` + `root.destroy()` jest takie, że aplikacja jest zbudowana dokładnie tak samo, tylko bez niekończącego się oczekiwania na końcu (zobacz wyjaśnienie w rozdziale 16).

## Sprawa robocza

```python
import tkinter as tk

root = tk.Tk()
width = {"value": 1}   # НЕ 12 — иначе не будет видно, сработал ли command

def set_width(value):
    width["value"] = int(value)

scale = tk.Scale(root, from_=1, to=20, orient="horizontal", command=set_width)
scale.pack()
scale.set(12)
root.update()
print("Текущая толщина:", width["value"])   # 12, а не 1 — значит command сработал
root.destroy()
```

## Sprawdzenie wyniku

```python
assert width["value"] == 12, "command у Scale должен был записать 12 в width"
print("Верно: ползунок изменил состояние, а не только своё внутреннее значение.")
```
