## Cel

Zdefiniuj i ponownie wykorzystaj funkcję do obliczeń.

## Sprawa robocza

```python
def summa_kvadratov(a, b):
    return a ** 2 + b ** 2

def kvadrat_summy(a, b):
    return (a + b) ** 2

print(summa_kvadratov(3, 4))
print(kvadrat_summy(3, 4))
```

## Podstawowa praktyka zadań ★

Użyj obu funkcji, aby znaleźć różnicę między kwadratem sumy a sumą kwadratów dla a=3, b=4.

```python
raznost = kvadrat_summy(3, 4) - summa_kvadratov(3, 4)
print(raznost)
```
