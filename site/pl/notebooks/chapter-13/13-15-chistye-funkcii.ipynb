## Podstawowa praktyka zadań ★

```python
def rectangle_area(width, height):
    return width * height

log = []

def draw_square_effect(size):
    log.append(f"Нарисован квадрат {size}")

area = rectangle_area(4, 5)
draw_square_effect(10)

print(area, log)
```
