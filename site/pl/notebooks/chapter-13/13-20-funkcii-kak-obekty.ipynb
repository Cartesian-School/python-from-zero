# 13-20 · Funkcje jako obiekty

Praktyka do sekcji [„Funkcje jako obiekty”

## Cel

Funkcje zapisu w słowniku i pod inną nazwą wywołaj przez oba linki.

## Podstawowa praktyka zadań ★

```python
def double(x):
    return x * 2

def square(x):
    return x ** 2

operations = {"double": double, "square": square}
result_double = operations["double"](5)
result_square = operations["square"](5)

action = double
result_action = action(7)

print(result_double, result_square, result_action)
```
