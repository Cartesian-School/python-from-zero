## Krótkie przypomnienie

Komenda `print(...)` pokazuje, co jej przekazałeś. Zawsze pisz w Python
polega na cudzysłowie — pojedynczym `'...'` lub podwójnie `"..."`nie ma różnicy.

## Sprawa robocza

```python
print("Привет, мир!")
```

## Eksperyment 1

Zamień tekst w cudzysłowie na własne pozdrowienia i ponownie przejdź komórkę.

```python
print("Привет, Cartesian School!")
```

## Eksperyment 2

`print()` może wyświetlać nie tylko tekst. Spróbuj podać mu liczbę lub przykład
arytmetyce Python sam wszystko obliczy.

```python
print(2024 - 1991)
print(7 * 6)
```

## Typowy błąd

Jeśli usuniesz cudzysłów, Python pomyśli, że mówisz o czymś po imieniu, a nie o zwykłym
tekstu. **Nazwa** — to zakodowana etykieta, po której Python szuka wcześniej z nią powiązanego
wartość lub funkcję. Nazwy i zmienne są szczegółowo omówione w poniższych teoretycznych materiałach
sekcję. Wystarczy znać: nazwę `Привет` jeszcze nie określono, więc Python zgłosi
błąd `NameError`.

```python
# błąd: tekst bez cudzysłowu
print(Привет)
```

## Poprawka

Wystarczy zwrócić cudzysłów – wtedy Python rozumie, że to zwykły tekst, a nie nazwa
zmiennej.

```python
print("Привет")
```

## Podstawowa praktyka zadań ★

Na ekranie wyświetl trzy linie: swoje imię, ulubiony kolor i liczbę, którą lubisz.
Każdy łańcuch znaków to osobne wyzwanie `print()`.

```python
print("Cartesian School")
print("Электрик-синий")
print(14)
```

## Praktyka samodzielne

`print()` można przywołać dowolną liczbę razy z rzędu, co oznacza, że można używać symboli na ekranie
złożyć prosty obraz. Spróbuj narysować coś własnego tekstem: dom, choinkę,
emoji to wszystko od znaków klawiaturowych.

```python
print("  /\\  ")
print(" /  \\ ")
print("/____\\")
```

## Zadanie dodatkowe ★★★

`print()` może przyjmować kilka wartości oddzielonych przecinkami jednocześnie, w takim przypadku wyświetlane są w jednym
linię oddzieloną spację. Wyjście za pomocą jednego polecenia `print()` trzy różne znaczenia: tekst, liczba oraz
wynik wyrażenia arytmetycznego.

```python
print("Результат:", 6, 7 * 6)
```

## Zerknijmy do środka: type()

Każda wartość w Python ma swój typ — tekst, liczbę, liczbę dziesiętną i tak dalej. Funkcja `type()` pokazuje typ bezpośrednio w interpreterze, bez żadnych dodatkowych narzędzi.

```python
print(type("Привет"))
print(type(42))
print(type(3.14))
```

## Zapytajmy Python: help()

Python potrafi się wyjaśnić. Funkcja `help()` pokazuje wbudowaną pomoc dla dowolnej funkcji — tę samą, która jest w oficjalnej dokumentacji na docs.python.org, ale bezpośrednio w interpreterze, bez przeglądarki.

```python
help(print)
```
