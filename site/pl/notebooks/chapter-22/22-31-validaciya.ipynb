## Cel

testować tę samą funkcję clean_title() używaną w tym projekcie (projects/flask/todo-app/app.py) na kilku przypadkach brzegowych.

## Sprawa robocza

```python
MAX_TITLE_LENGTH = 200


def clean_title(raw_title):
    title = raw_title.strip()
    if not title or len(title) > MAX_TITLE_LENGTH:
        return None
    return title


print(clean_title("Купить хлеб"))
print(clean_title("   "))
print(clean_title("а" * 201))
```

## Sprawdzenie wyniku

```python
assert clean_title("Купить хлеб") == "Купить хлеб"
assert clean_title("  Пробелы по краям  ") == "Пробелы по краям"
assert clean_title("") is None
assert clean_title("   ") is None
assert clean_title("а" * 200) == "а" * 200
assert clean_title("а" * 201) is None
print("Верно: пустой, состоящий из пробелов и слишком длинный ввод отклоняются.")
```
