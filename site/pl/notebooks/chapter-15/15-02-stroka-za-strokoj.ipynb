## Cel

Opanować liniowe czytanie plików.

## Przygotowanie – Stwórz plik tutoriala

```python
spisok = ["яблоки", "хлеб", "молоко", "сыр"]
with open("spisok.txt", "w", encoding="utf-8") as f:
    f.writelines(item + "\n" for item in spisok)
print("Учебный файл списка покупок создан.")
```

## Sprawa robocza

```python
with open("spisok.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())
```

## Eksperyment 1 – readlines()

```python
with open("spisok.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(len(lines), "строк")
print(lines[0])
```

## Podstawowa praktyka zadań ★

Policz, ile jest linii w pliku, nie używając len(readlines()) — tylko licznik w pętli.

```python
count = 0
with open("spisok.txt", "r", encoding="utf-8") as file:
    for line in file:
        count += 1

print("Строк в файле:", count)
```
