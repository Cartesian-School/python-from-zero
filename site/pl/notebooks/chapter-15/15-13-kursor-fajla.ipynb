#15-13 · Kursor pliku: tell() i seek()

Praktyka do sekcji [„Kursor pliku”](/pl/chapters/rozdzial-15/15-13-kursor-fajla.html).

## Przygotowania

```python
with open("alfavit2.txt", "w", encoding="utf-8") as f:
    f.write("ABCDE")
print("Готово.")
```

## Sprawa robocza

```python
with open("alfavit2.txt", "r", encoding="utf-8") as f:
    first = f.read(2)
    pos_after_first = f.tell()
    second = f.read(2)
    f.seek(0)
    reread = f.read()
print(first, pos_after_first, second, reread)
```

## Podstawowa praktyka zadań ★

Przeczytaj plik do końca, a potem przeczytaj go ponownie bez seek() — upewnij się, że kursor sam się nie powraca.

```python
with open("alfavit2.txt", "r", encoding="utf-8") as f:
    f.read()
    posle_konca = f.read()
posle_konca_pusto = posle_konca == ""
print(posle_konca_pusto)
```
