# 15-16 · Tekst, bytes i kodowanie UTF-8

Praktyka do sekcji [„Tekst, bytes i UTF-8”

## Sprawa robocza

```python
text = "Питон"
encoded = text.encode("utf-8")
chars_len = len(text)
bytes_len = len(encoded)
decoded_back = encoded.decode("utf-8")
print(chars_len, bytes_len, decoded_back)
```

## Podstawowa praktyka zadań ★

Zapisz linię cyrylicą w pliku z wyraźnym encoding=„utf-8”

```python
with open("privet3.txt", "w", encoding="utf-8") as f:
    f.write("Привет!")

with open("privet3.txt", "r", encoding="utf-8") as f:
    read_back = f.read()
print(read_back)
```
