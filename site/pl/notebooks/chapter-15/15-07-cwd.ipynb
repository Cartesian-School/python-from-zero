# 15-07 · Aktualny katalog roboczy (CWD)

Lokalna praktyka dla sekcji [„Current Working Directory (CWD)”

Ten laptop działa lokalnie, na twoim komputerze, a nie w przeglądarce: celem ćwiczenia jest odróżnienie CWD od folderu, w którym przechowywany jest oryginalny plik w rzeczywistym systemie plików.

## Krok 1 – Gdzie jestem?

```python
from pathlib import Path

print("CWD:", Path.cwd())
```

## Krok 2 – Folder ze skryptami

```python
from pathlib import Path

if "__file__" in globals():
    print("Папка файла:", Path(__file__).resolve().parent)
else:
    print("__file__ не определён в этой среде выполнения (например, в интерактивном ноутбуке) — "
          "это нормально и ожидаемо, см. раздел 15.7.")
```

## Zadanie niezależne od zadania ★★

Zapisz ten plik normalnie `.py`-script w `project/src/`, utwórz `project/data/config.json`i uruchom skrypt dwa razy: raz z `project/`, innym razem `project/src/`. Porównaj wartość `Path.cwd()` i która ścieżka `Path("data/config.json")` pozwala w każdym przypadku – i upewnij się, że niezawodną opcją jest przejście przez `Path(__file__).resolve().parent`, nie przez CWD.
