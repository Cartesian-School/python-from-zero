**Ten laptop działa lokalnie** — celem ćwiczenia jest zobaczenie prawdziwej trwałości na prawdziwym dysku, a nie w wirtualnym systemie plików przeglądarki.

## Krok 1 – Utwórz plik w prawdziwym folderze data/

```python
from pathlib import Path

Path("data").mkdir(exist_ok=True)
Path("data/moya_zametka.txt").write_text("Привет с настоящего диска!", encoding="utf-8")
print("Файл записан в data/moya_zametka.txt")
```

## Krok 2 — zatrzymaj wykonywanie, a następnie uruchom notebook ponownie

```python
from pathlib import Path

print(Path("data/moya_zametka.txt").read_text(encoding="utf-8"))
```
