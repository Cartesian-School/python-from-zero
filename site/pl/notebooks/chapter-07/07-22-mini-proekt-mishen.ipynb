## Cel

Zbierz infografikę z pierścieni, osi, losowych punktów i podpisu.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

Dwa koncentryczne pierścienie.

```python
import random
artist.reset()
random.seed(1)
for radius, color in [(100, "#DC2626"), (60, "#FAFAFC")]:
    artist.penup()
    artist.goto(0, -radius)
    artist.pendown()
    artist.fillcolor(color)
    artist.begin_fill()
    artist.circle(radius)
    artist.end_fill()
```

## Zadanie ★★ Samodzielna praktyka

Dodaj trzeci, najmniejszy pierścień (promień 25, kolor „#DC2626”

```python
artist.penup()
artist.goto(0, -25)
artist.pendown()
artist.fillcolor("#DC2626")
artist.begin_fill()
artist.circle(25)
artist.end_fill()

random.seed(2)
artist.penup()
for _ in range(4):
    x = random.randint(-90, 90)
    y = random.randint(-90, 90)
    artist.goto(x, y)
    artist.dot(10, "#059669")
```
