## Cel

Utworzenie wielu niezależnych obiektów Turtle.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

Dwa żółwie, różne kolory.

```python
artist.reset()
friend = turtle.Turtle()
friend.color("#DB2777")
friend.penup(); friend.goto(0, -80); friend.pendown()
artist.forward(100)
friend.forward(100)
```

## Podstawowa praktyka zadań ★

Stwórz trzeciego żółwia o imieniu `third`, zielone i przenieś na forward(80).

```python
third = turtle.Turtle()
third.color("green")
third.penup(); third.goto(0, 80); third.pendown()
third.forward(80)
```

## Ukończenie (wykonaj raz, na samym końcu)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
