## Cel

Rozróżniaj pencolor(), fillcolor() i color().

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

color() ustawia oba kolory jednocześnie.

```python
artist.reset()
artist.color("blue", "gold")
artist.begin_fill()
for _ in range(4):
    artist.forward(100)
    artist.right(90)
artist.end_fill()
```

## Podstawowa praktyka zadań ★

Narysuj trójkąt z niebieskim konturem (pencolor)) i zielonym wypełnieniem (fillcolor)), ustawione oddzielnie.

```python
artist.reset()
artist.pencolor("blue")
artist.fillcolor("green")
artist.begin_fill()
for _ in range(3):
    artist.forward(100)
    artist.right(120)
artist.end_fill()
```

## Ukończenie (wykonaj raz, na samym końcu)
