## Cel

Rysuj łuki o różnych rozmiarach z użyciem parametru extent.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Sprawa robocza

```python
artist.reset()
artist.circle(80, 180)
print("Позиция после полукруга:", artist.position())
```

## Eksperyment 1

Spróbuj extent=90, 270 i 360 (pełne koło) na tym samym promieniu.

```python
for extent in [90, 270, 360]:
    artist.reset()
    artist.circle(80, extent)
    print(f"extent={extent}: финальная позиция {artist.position()}")
```

## Podstawowa praktyka zadań ★

Narysuj kopułę domu: łuk 180° o promieniu 60, następnie prosto w dół po obu stronach.

```python
artist.reset()
artist.circle(60, 180)
artist.left(90)
artist.forward(60)
print("Купол с одной стеной готов.")
```
