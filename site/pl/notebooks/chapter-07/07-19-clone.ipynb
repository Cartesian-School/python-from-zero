## Cel

Stworzyć klona żółwia i sprawić, by rozstał się z oryginałem.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

Klon odwrócił się w drugą stronę.

```python
artist.reset()
artist.penup(); artist.goto(0, -100); artist.pendown()
artist.setheading(90)
copy = artist.clone()
copy.color("#DB2777")
artist.left(20); artist.forward(150)
copy.right(20); copy.forward(150)
```

## Podstawowa praktyka zadań ★

Stwórz klon i odwróć go na right(90) względem oryginału, a następnie przesuń obie żółwie do przodu o 100.

```python
artist.reset()
copy = artist.clone()
copy.color("#059669")
copy.right(90)
artist.forward(100)
copy.forward(100)
```

## Ukończenie (wykonaj raz, na samym końcu)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
