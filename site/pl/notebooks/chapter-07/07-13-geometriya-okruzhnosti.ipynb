## Cel

Zrozum kierunek rysowania circle() oraz rolę znaku promienia.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

Dodatni i ujemny promień obok siebie.

```python
artist.reset()
artist.circle(70)
artist.penup(); artist.forward(180); artist.pendown()
artist.circle(-70)
```

## Podstawowa praktyka zadań ★

Narysuj okrąg o promieniu 90 zgodnie z ruchem wskazówek zegara (użyj ujemnego promienia).

```python
artist.reset()
artist.circle(-90)
```
