## Cel

Zbuduj statyczny pokrętło z dywizji i rąk.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

Obwód i 12 dywizji.

```python
artist.reset()
artist.circle(120)
for hour in range(12):
    artist.penup()
    artist.goto(0, 0)
    artist.setheading(90 - hour * 30)
    artist.forward(105)
    artist.pendown()
    artist.forward(15)
    artist.penup()
```

## Zadanie ★★ Samodzielna praktyka

Dodaj wskazówkę godzinową (grubą, krótką) na 9 godzin i wskazówkę minutową (cieńszą, dłuższą) na 12 godzin.

```python
artist.pensize(5)
artist.pencolor("#5B24F9")
artist.goto(0, 0)
artist.setheading(90 - 9 * 30)
artist.pendown()
artist.forward(70)
artist.penup()

artist.pensize(3)
artist.pencolor("#DB2777")
artist.goto(0, 0)
artist.setheading(90 - 12 * 30)
artist.pendown()
artist.forward(95)
```

## Ukończenie (wykonaj raz, na samym końcu)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
