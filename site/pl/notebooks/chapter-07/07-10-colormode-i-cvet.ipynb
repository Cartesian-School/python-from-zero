## Cel

Opanować colormode(1.0) i colormode(255), ustawiać kolory przez RGB.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

RGB w trybie 255.

```python
artist.reset()
screen.colormode(255)
artist.pencolor(255, 0, 0)
artist.forward(100)
```

## Podstawowa praktyka zadań ★

W trybie colormode(255) narysuj linię długości 120 kolorem (0, 150, 0) — zielonym.

```python
artist.reset()
screen.colormode(255)
artist.pencolor(0, 150, 0)
artist.forward(120)
```

## Ukończenie (wykonaj raz, na samym końcu)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
