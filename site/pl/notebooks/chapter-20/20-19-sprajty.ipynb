## Cel

Zbudować niezawodną ścieżkę do zasobu przez pathlib, niezależną od bieżącego katalogu roboczego uruchomienia.

## Sprawa robocza

```python
import os
from pathlib import Path

import pygame

pygame.init()
screen = pygame.display.set_mode((600, 400))

# W laptopie nie ma zmiennej __file__, więc nie ma gdzie pobrać „katalogu samego pliku”
# Jupyter ustawia katalog notebooka jako bieżący katalog, a Path.cwd() wskazuje tam samo.
# Ale to przypadek, a nie wiarygodna technika – teraz zobaczymy dlaczego.
# W zwykłym pliku .py- pisze się tak (sekcja 20.19):
#     BASE_DIR = Path(__file__).resolve().parent
BASE_DIR = Path.cwd()
print("BASE_DIR зафиксирован один раз:", BASE_DIR)

try:
    os.chdir("/")  # как будто игрок запустил игру не из её папки
    print("Путь от Path.cwd():", Path.cwd() / "assets" / "igrok.png")
    print("Путь от BASE_DIR: ", BASE_DIR / "assets" / "igrok.png")
finally:
    os.chdir(BASE_DIR)  # возвращаемся, чтобы не сломать остальные ячейки

pygame.quit()
```

## Sprawdzenie wyniku

```python
import os
from pathlib import Path

put_ot_base = BASE_DIR / "assets" / "igrok.png"

# Główna cecha wiarygodnej ścieżki: nie zmienia się wraz ze zmianą katalogu roboczego.
try:
    os.chdir("/")
    put_ot_cwd_v_drugom_meste = Path.cwd() / "assets" / "igrok.png"
    put_ot_base_v_drugom_meste = BASE_DIR / "assets" / "igrok.png"
finally:
    os.chdir(BASE_DIR)

assert put_ot_base.is_absolute()
assert put_ot_base.name == "igrok.png"
assert put_ot_base.parent.name == "assets"

# Ścieżka z zadeklarowanego BASE_DIR przetrwała, ale ścieżka z aktualnego katalogu nie:
assert put_ot_base_v_drugom_meste == put_ot_base
assert put_ot_cwd_v_drugom_meste != put_ot_base

print("Верно: путь от BASE_DIR не изменился, а путь от Path.cwd() — изменился.")
```
