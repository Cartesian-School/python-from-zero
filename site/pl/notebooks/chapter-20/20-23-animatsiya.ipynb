## Cel

Przełączaj klatkę animacji na jej własnym timerze, zamiast każdej klatki w pętli.

## O pętli gier w tym laptopie

W normalnym życiu `.py`pliku -plik, pętla gry to `while rabotaet:`, gdzie `rabotaet` staje się `False`gdy użytkownik zamyka okno. W automatycznie uruchamianym laptopie nie ma nikogo, kto zamyka okno, więc tutaj przechodzimy przez ustaloną liczbę klatek `for _ in range(N):` jest logika każdej klatki (ruch, renderowanie, `clock.tick()`) jest dokładnie taki sam jak w prawdziwej grze.

## Sprawa robocza

```python
import pygame

pygame.init()
screen = pygame.display.set_mode((600, 400))
clock = pygame.time.Clock()

KOLICHESTVO_KADROV = 4
INTERVAL_KADRA = 0.1  # секунд на кадр анимации

tekushij_kadr = 0
vremya_animacii = 0.0
istoriya_kadrov = []

for _ in range(60):
    dt = clock.tick(60) / 1000
    vremya_animacii += dt
    while vremya_animacii >= INTERVAL_KADRA:
        vremya_animacii -= INTERVAL_KADRA
        tekushij_kadr = (tekushij_kadr + 1) % KOLICHESTVO_KADROV
    istoriya_kadrov.append(tekushij_kadr)

print("Кадр анимации сменился хотя бы раз:", len(set(istoriya_kadrov)) > 1)
print("Все значения кадра в допустимом диапазоне:", all(0 <= k < KOLICHESTVO_KADROV for k in istoriya_kadrov))
pygame.quit()
```

## Sprawdzenie wyniku

```python
assert len(set(istoriya_kadrov)) > 1
assert all(0 <= k < KOLICHESTVO_KADROV for k in istoriya_kadrov)
print("Верно: анимация переключает кадры и никогда не выходит за пределы спрайт-листа.")
```
