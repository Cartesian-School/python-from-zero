## O pętli gier w tym laptopie

W normalnym życiu `.py`pliku -plik, pętla gry to `while rabotaet:`, gdzie `rabotaet` staje się `False`gdy użytkownik zamyka okno. W automatycznie uruchamianym laptopie nie ma nikogo, kto zamyka okno, więc tutaj przechodzimy przez ustaloną liczbę klatek `for _ in range(N):` jest logika każdej klatki (ruch, renderowanie, `clock.tick()`) jest dokładnie taki sam jak w prawdziwej grze.

## Przykład roboczy – Ruch

```python
import pygame

pygame.init()
screen = pygame.display.set_mode((600, 400))
clock = pygame.time.Clock()

x_igroka = 50
skorost_x = 5

for kadr in range(10):
    x_igroka += skorost_x
    screen.fill((0, 0, 0))
    pygame.draw.rect(screen, (100, 200, 255), (x_igroka, 180, 40, 40))
    pygame.display.flip()
    clock.tick(60)

print("Позиция игрока после 10 кадров:", x_igroka)
pygame.quit()
```

## Sprawdzenie wyniku

```python
assert x_igroka == 50 + 5 * 10
print("Верно: игрок сдвинулся на", 5 * 10, "пикселей за 10 кадров.")
```

## Eksperyment 1 - Wydarzenie KEYDOWN

```python
import pygame

pygame.init()
screen = pygame.display.set_mode((600, 400))

# programowo tworzą zdarzenie naciśnięcia klawisza tak jak klawiatura
sobytie = pygame.event.Event(pygame.KEYDOWN, key=pygame.K_SPACE)
pygame.event.post(sobytie)

for event in pygame.event.get():
    if event.type == pygame.KEYDOWN and event.key == pygame.K_SPACE:
        print("Прыжок! (клавиша SPACE обработана)")

pygame.quit()
```

## Eksperyment 2 – get_pressed() dla ruchu ciągłego

```python
import pygame

pygame.init()
screen = pygame.display.set_mode((600, 400))

klavishi = pygame.key.get_pressed()
# w rzeczywistym zastosowaniu będzie to True, jeśli klucz jest naprawdę trzymany;
# tutaj po prostu sprawdzić, czy funkcja zwraca obiekt, z którym można pracować w ten sposób
print("Тип результата get_pressed():", type(klavishi))
print("Стрелка влево зажата:", klavishi[pygame.K_LEFT])
pygame.quit()
```
