# 20-18 · Rect i układ współrzędnych

Praktyka do sekcji [«Rect i układ współrzędnych»](/pl/chapters/rozdzial-20/20-18-rect-i-koordinaty.html). Tutaj Rect został odwzorowany zwykłą krotką (x, y, szerokość, wysokość) — idea jest ta sama, prawdziwe okno nie jest potrzebne.

## Cel

Obliczaj granice prostokąta ręcznie i potwierdzaj, że oś Y w Pygame rośnie w dół.

## Sprawa robocza

```python
def granicy(rect):
    x, y, w, h = rect
    return {"left": x, "top": y, "right": x + w, "bottom": y + h, "center": (x + w / 2, y + h / 2)}

igrok = (100, 200, 40, 40)
g = granicy(igrok)
print(g)

# Nasza wersja manualna traktuje środek jako ułamek. Prawdziwy pygame.Rect tylko przechowuje
# współrzędne całkowitoliczbowe i zaokrągla środek w dół: Rect(100, 200, 41, 41) będzie miało środek
# (120, 220), nie (120,5, 220,5). W przypadku rozmiarów parzystych nie ma różnicy.

# Jump Up na ekranie Pygame to ZMNIEJSZENIE y, a nie wzrost
y_do_pryzhka = 300
y_posle_pryzhka = y_do_pryzhka - 50   # выше на экране = меньшее y
print("До прыжка y =", y_do_pryzhka, " после прыжка y =", y_posle_pryzhka)
```

## Sprawdzenie wyniku

```python
assert g["right"] == 140
assert g["bottom"] == 240
assert g["center"] == (120.0, 220.0)
assert y_posle_pryzhka < y_do_pryzhka
print("Верно: границы посчитаны правильно, и прыжок вверх уменьшает y, как и на экране Pygame.")
```
