Praktyka do sekcji [„Game Loop”

## Cel

Wyraźnie podziel kod pojedynczej klatki na trzy nazwane fazy.

## O pętli gier w tym laptopie

W normalnym życiu `.py`pliku -plik, pętla gry to `while rabotaet:`, gdzie `rabotaet` staje się `False`gdy użytkownik zamyka okno. W automatycznie uruchamianym laptopie nie ma nikogo, kto zamyka okno, więc tutaj przechodzimy przez ustaloną liczbę klatek `for _ in range(N):` jest logika każdej klatki (ruch, renderowanie, `clock.tick()`) jest dokładnie taki sam jak w prawdziwej grze.

## Sprawa robocza

```python
import pygame

pygame.init()
screen = pygame.display.set_mode((600, 400))
clock = pygame.time.Clock()

x = 50
schet_sobytij = 0

def input_faza():
    global schet_sobytij
    for event in pygame.event.get():
        schet_sobytij += 1

def update_faza():
    global x
    x += 3

def render_faza():
    screen.fill((20, 20, 40))
    pygame.draw.rect(screen, (100, 200, 255), (x, 180, 40, 40))
    pygame.display.flip()

for kadr in range(20):
    input_faza()
    update_faza()
    render_faza()
    clock.tick(60)

print("Позиция после 20 кадров:", x)
```

## Sprawdzenie wyniku

```python
assert x == 50 + 3 * 20
print("Верно: три фазы выполнились в правильном порядке 20 раз подряд.")
```

## Podstawowa praktyka zadań ★

Zamiana update_faza() i render_faza(), aby upewnić się, że końcowa pozycja x się nie zmienia (kolejność tych dwóch faz nie wpływa na arytmetykę x), ale faktyczne renderowanie jednej klatki przesuwa się o klatkę względem logiki (w tym mini-przykładzie różnica nie jest widoczna bez ekranu, ale właśnie tak pojawia się błąd z sekcji 20.23 dotyczący opóźnień animacji za klatką).

```python
x = 50
for kadr in range(20):
    input_faza()
    render_faza()
    update_faza()
print("Позиция после смены порядка:", x)
assert x == 50 + 3 * 20
```
