## Pro input() w tym laptopie

Ten laptop jest automatyczny, więc `input()` tymczasowo zastąpione wcześniej przygotowanymi odpowiedziami.

```python
_answers = iter(['Python3Code'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Sprawa robocza

```python
password = input("Придумайте пароль: ")

has_digit = False
has_letter = False
has_space = False

for ch in password:
    if ch.isdigit():
        has_digit = True
    elif ch.isalpha():
        has_letter = True
    elif ch == " ":
        has_space = True

dlinnyj_dostatochno = len(password) >= 8
podhodit = dlinnyj_dostatochno and has_digit and has_letter and not has_space

print("Подходит:", podhodit)
```
