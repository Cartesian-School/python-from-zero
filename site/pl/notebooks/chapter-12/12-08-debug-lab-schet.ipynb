## Cel

Znajdź i napraw błąd: licznik jest resetowany wewnątrz pętli.

## Uszkodzony kod (Nie wykonany)

```python
for correct, given in zip(answers, user_answers):
    score = 0
    if given == correct:
        score += 1
```

`score = 0` znajduje się WEWNĄTRZ pętli — licznik nigdy się nie kumuluje.

## Zadanie ★ Podstawowe ćwiczenia — poprawiona wersja

````python
answers = ["python", "git", "sql"]
user_answers = ["python", "python", "sql"]

score = 0
for correct, given in zip(answers, user_answers):
    if given == correct:
        score += 1

print("Итоговый счёт:", score)
````
