## Cel

Zaplanuj kształt w współrzędnych, a następnie przetłumacz plan na kod.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

Wypełniony trójkąt w trzech współrzędnych.

```python
artist.reset()
artist.fillcolor("#B9A0FC")
artist.penup()
artist.goto(-60, -50)
artist.pendown()
artist.begin_fill()
artist.goto(60, -50)
artist.goto(0, 70)
artist.goto(-60, -50)
artist.end_fill()
```

## Podstawowa praktyka zadań ★

Narysuj wypełniony czworokąt na współrzędnych (-80, -40), (80, -40), (60, 60), (-60, 60).

```python
artist.reset()
artist.fillcolor("#059669")
artist.penup()
artist.goto(-80, -40)
artist.pendown()
artist.begin_fill()
artist.goto(80, -40)
artist.goto(60, 60)
artist.goto(-60, 60)
artist.goto(-80, -40)
artist.end_fill()
```

## Ukończenie (wykonaj raz, na samym końcu)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
