## Cel

Opanuj absolutny kierunek: `setheading()`, `heading()`, `home()`.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Sprawa robocza

```python
artist.reset()
artist.setheading(90)
artist.forward(100)
artist.setheading(180)
artist.forward(100)
print("Курс:", artist.heading())
```

## Eksperyment 1

Sprawdź kurs dla wszystkich czterech kierunków: 0°, 90°, 180°, 270°.

```python
artist.reset()
for ugol in [0, 90, 180, 270]:
    artist.setheading(ugol)
    artist.forward(60)
    print(f"Курс {ugol}° -> позиция {artist.position()}")
```

## Eksperyment 2

`home()` przywraca żółwia do pierwotnego punktu i początkowego kursu jedną komendą.

```python
artist.reset()
artist.setheading(45)
artist.forward(100)
print("До home():", artist.position(), artist.heading())

artist.home()
print("После home():", artist.position(), artist.heading())
```

## Typowy błąd

`left()`/`right()` zmieniają kurs *względem obecnego* oraz `setheading()` — wyznacza kurs *absolutnie*. Jeśli pomieszasz to, liczba nie będzie taka, jakiej się spodziewałem.

```python
artist.reset()
artist.setheading(90)   # смотрим вверх
artist.left(90)         # ОТНОСИТЕЛЬНЫЙ поворот от 90 -> 180, а не setheading(90)!
print("Итоговый курс:", artist.heading(), "(вероятно, не то, что ожидали)")
```

## Podstawowa praktyka zadań ★

Narysuj krzyż (plus) z czterech promieni wychodzących ze środka, używając `setheading()` dla każdej wiązki: 0°, 90°, 180°, 270°.

```python
artist.reset()
for ugol in [0, 90, 180, 270]:
    artist.setheading(ugol)
    artist.forward(80)
    artist.backward(80)
print("Готово, финальная позиция:", artist.position())
```
