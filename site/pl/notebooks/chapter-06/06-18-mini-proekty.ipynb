## Cel

Zbierz kilka technik rozdziału w jednym pełnym obrazie.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

Prosta tarcza z dwóch pierścieni.

```python
artist.reset()
for radius, color in [(60, "#DC2626"), (30, "#FAFAFC")]:
    artist.penup()
    artist.goto(0, -radius)
    artist.pendown()
    artist.fillcolor(color)
    artist.begin_fill()
    artist.circle(radius)
    artist.end_fill()
```

## Zadanie ★★ Samodzielna praktyka

Narysuj swoją tarczę z trzema pierścieniami o promieniach 80, 55, 30 z naprzemiennymi kolorami według własnego wyboru.

```python
artist.reset()
for radius, color in [(80, "#2563EB"), (55, "#FAFAFC"), (30, "#2563EB")]:
    artist.penup()
    artist.goto(0, -radius)
    artist.pendown()
    artist.fillcolor(color)
    artist.begin_fill()
    artist.circle(radius)
    artist.end_fill()
```
