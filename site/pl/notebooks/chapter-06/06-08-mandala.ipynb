## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Sprawa robocza

```python
artist.reset()
shag_ugla = 10
ugol = 0
while ugol < 360:
    artist.setheading(ugol)
    artist.forward(150)
    artist.backward(150)
    ugol += shag_ugla
print("Мандала готова, лучей:", 360 // shag_ugla)
```

## Eksperyment 1 ★ – Kolejny krok do rogu

Spróbuj `shag_ugla = 30` — wzór stanie się znacznie rzadszy.

```python
artist.reset()
shag_ugla = 30
ugol = 0
while ugol < 360:
    artist.setheading(ugol)
    artist.forward(150)
    artist.backward(150)
    ugol += shag_ugla
print("Лучей на этот раз:", 360 // shag_ugla)
```

## Eksperyment 2 ★★ — Niestandardowa długość wiązki

Spróbuj `forward(80)` zamiast 150.

```python
artist.reset()
shag_ugla = 10
dlina = 80
ugol = 0
while ugol < 360:
    artist.setheading(ugol)
    artist.forward(dlina)
    artist.backward(dlina)
    ugol += shag_ugla
print("Мандала меньшего размера готова.")
```

## Zadanie dodatkowe ★★★

Spróbuj dwóch kolorów: parzyste promienie w jednym kolorze, nieparzyste w drugim (użyj licznika i `%` z rozdziału 5).

```python
artist.reset()
shag_ugla = 10
ugol = 0
i = 0
while ugol < 360:
    artist.pencolor("purple" if i % 2 == 0 else "blue")
    artist.setheading(ugol)
    artist.forward(150)
    artist.backward(150)
    ugol += shag_ugla
    i += 1
artist.pencolor("black")
print("Двухцветная мандала готова.")
```
