## Cel

Mistrz begin_fill()/end_fill(), circle() i dot().

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Sprawa robocza

Wypełniony kwadrat.

```python
artist.reset()
artist.fillcolor("#B9A0FC")
artist.begin_fill()
for _ in range(4):
    artist.forward(100)
    artist.right(90)
artist.end_fill()
```

## Eksperyment 1

Narysuj koło o promieniu 60, a obok niego połowę koła (extent=180).

```python
artist.reset()
artist.circle(60)
artist.penup()
artist.forward(150)
artist.pendown()
artist.circle(60, 180)
```

## Podstawowa praktyka zadań ★

Narysuj wypełniony okrąg o promieniu 50 (fillcolor dowolny wybór).

```python
artist.reset()
artist.fillcolor("#DB2777")
artist.begin_fill()
artist.circle(50)
artist.end_fill()
```
