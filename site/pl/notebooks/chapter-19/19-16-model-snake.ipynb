## Cel

Wąż – lista pozycje; snake[0] zawsze głowa.

## Sprawa robocza

```python
snake = [(0, 0), (-20, 0), (-40, 0)]

def head_of(snake):
    return snake[0]

def body_of(snake):
    return snake[1:]

print(head_of(snake), body_of(snake))
```

## Podstawowa praktyka zadań ★

```python
assert head_of(snake) == (0, 0)
assert body_of(snake) == [(-20, 0), (-40, 0)]
assert len(snake) == 3
print("Модель змейки — обычный список координат.")
```
