## Cel

Zbieraj split() i liczenie w słowniku do analizy tekstu.

## Sprawa robocza

```python
text = "кот и пёс и кот"
words = text.lower().split()
print(words)
print(len(words))
```

## Podstawowa praktyka zadań ★

Dla `text = "кот и пёс и кот"` podziel na słowa (`words`), liczyć ich liczbę (`total`), oraz przez słownik `schetchik` liczyć, ile razy pojawia się każde słowo.

```python
text = "кот и пёс и кот"
words = text.lower().split()
total = len(words)

schetchik = {}
for word in words:
    schetchik[word] = schetchik.get(word, 0) + 1

print(total)
print(schetchik)
```
