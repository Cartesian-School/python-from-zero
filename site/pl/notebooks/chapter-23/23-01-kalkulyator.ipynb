## Example

````python
import sys

sys.path.insert(0, "../../projects/tkinter/calculator")
import calculator as c

print(c.vychislit_vyrazhenie("2+2"))
print(c.vychislit_vyrazhenie("10-3*2"))
print(c.vychislit_vyrazhenie("(1+2)*3"))
print(c.vychislit_vyrazhenie("7/2"))
````

## Sprawdzenie wyniku

````python
assert c.vychislit_vyrazhenie("2+2") == "4"
assert c.vychislit_vyrazhenie("10-3*2") == "4"
assert c.vychislit_vyrazhenie("(1+2)*3") == "9"
assert c.vychislit_vyrazhenie("7/2") == "3.5"
print("Верно: базовые вычисления совпадают с ожидаемыми.")
````

## Eksperyment – nieprawidłowe i niebezpieczne dane wejściowe nie prowadzą do awarii

````python
print(c.vychislit_vyrazhenie("5/0"))
print(c.vychislit_vyrazhenie("2+*3"))
print(c.vychislit_vyrazhenie("import os"))

assert c.vychislit_vyrazhenie("5/0") == "Ошибка"
assert c.vychislit_vyrazhenie("2+*3") == "Ошибка"
assert c.vychislit_vyrazhenie("import os") == "Ошибка"
print("Верно: деление на ноль, синтаксическая ошибка и посторонний код — везде «Ошибка».")
````

## Dlaczego to jest bezpieczne — dwie niezależne bariery

1. **Biały znak lista**: Najpierw sprawdzanie funkcji `set(vyrazhenie) <= DOPUSTIMYE_SIMVOLY` — łańcuch znaków musi składać się wyłącznie z cyfr, `+ - * / ( ) .` i spacje. Litery w `"import os"` nie są uwzględnione w tym lista, więc funkcja zwraca `"Ошибка"` więcej **przed** rozmową `ast.parse()` – po prostu nie podlega analizie.
2. **Przeanalizuj `ast.parse(..., mode="eval")`**: łańcuch znaków jakby `"2+*3"` składa się tylko z dozwolonych znaków, ale nie jest poprawnym wyrażeniem arytmetycznym — `ast.parse()` windy `SyntaxError`, który też jest przechwytywany i zamieniany w `"Ошибка"`. A to, co jednak zostało rozwiązane, jest dodatkowo sprawdzane węzeł po węźle w `vychislit_uzel()`: dozwolone są tylko liczby, `+ - * /` i unary minus – bez imion i wyzwań.
