## Cel

Zbudować tabliczkę mnożenia za pomocą zagnieżdżonych pętli, zachowując wynik w listach lista.

## Podstawowa praktyka zadań ★

```python
tablica = []
for a in range(1, 6):
    stroka = []
    for b in range(1, 6):
        stroka.append(a * b)
    tablica.append(stroka)
print(tablica)
```
