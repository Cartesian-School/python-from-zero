Praktyka do sekcji [„Indeks, wartość i enumerate()”](/pl/chapters/rozdzial-10/10-10-enumerate-i-nakoplenie.html).

## Cel

opanować enumerate() i wzory miernika/akumulatora.

## Sprawa robocza

```python
slovo = "Python"
for i, letter in enumerate(slovo):
    print(i, letter)
```

## Podstawowa praktyka zadań ★

Używając enumerate(), zbierz lista linii w formie `"1: хлеб"` (numerowany od 1 zamiast 0) do zmiennej `spisok_s_nomerami`.

```python
produkty = ["хлеб", "молоко", "яйца", "сыр"]
spisok_s_nomerami = []
for i, p in enumerate(produkty):
    spisok_s_nomerami.append(f"{i + 1}: {p}")
print(spisok_s_nomerami)
```
