## Cel

Pętle Connect z losowością — z ustalonym seed() dla powtarzalności.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Przykład roboczy – Random Walk

```python
import random
artist.reset()
random.seed(7)

for _ in range(100):
    artist.setheading(random.randint(0, 360))
    artist.forward(10)
print("Блуждание готово.")
```

## Zadanie ★ Podstawowa praktyka — pole gwiazd

```python
import random
artist.reset()
artist.hideturtle()
artist.penup()
random.seed(3)

for _ in range(60):
    x = random.randint(-190, 190)
    y = random.randint(-190, 190)
    artist.goto(x, y)
    artist.dot(6, "white")
print("Звёздное поле готово.")
```
