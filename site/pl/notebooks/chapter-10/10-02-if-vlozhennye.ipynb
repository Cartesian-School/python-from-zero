# 10-02 · if wewnątrz pętli, zagnieżdżone pętle

Praktyka do sekcji [„Warunki if w for cyklach”

## Cel

Połącz if z cyklami for i gniazdowania w sobie.

## Sprawa robocza

```python
for number in range(1, 11):
    if number % 2 == 0:
        print(number, "— чётное")
```

## Eksperyment 1 - Zagnieżdżone pętle

```python
for row in range(3):
    for col in range(4):
        print(f"({row}, {col})", end=" ")
    print()
```

## Podstawowa praktyka zadań ★

Narysuj trójkąt gwiazdek: 1 gwiazdka w pierwszym rzędzie, 2 w drugim, ..., 5 w piątym.

```python
for row in range(1, 6):
    print("*" * row)
```

## Zadanie dodatkowe ★★★

Wygeneruj tablicę mnożenia z 1-5 x 1-5 zagnieżdżonych cykli.

```python
for a in range(1, 6):
    for b in range(1, 6):
        print(f"{a * b:3}", end=" ")
    print()
```
