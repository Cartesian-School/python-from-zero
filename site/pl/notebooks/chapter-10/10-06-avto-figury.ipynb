## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Przykład roboczy – Square

```python
artist.reset()
for _ in range(4):
    artist.forward(100)
    artist.right(90)
print("Квадрат готов.")
```

## Eksperyment 1 – Dowolny kształt

```python
artist.reset()
storony = 8
dlina = 60
ugol = 360 / storony

for _ in range(storony):
    artist.forward(dlina)
    artist.right(ugol)
print(f"{storony}-угольник готов.")
```

## Podstawowa praktyka zadań ★

Rysuj kształty z 3, 5 i 10 bokami w rzędzie, używając tego samego schematu przypominającego cechy.

```python
for storony in [3, 5, 10]:
    artist.reset()
    dlina = 50
    ugol = 360 / storony
    for _ in range(storony):
        artist.forward(dlina)
        artist.right(ugol)
    print(f"{storony}-угольник нарисован, угол поворота {ugol}")
```
