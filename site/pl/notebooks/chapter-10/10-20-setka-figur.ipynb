## Cel

Zastosować zagnieżdżone pętle „wiersze × kolumny” do prawdziwej grafiki.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Podstawowa praktyka zadań ★

```python
artist.reset()
artist.hideturtle()
artist.penup()
shag = 60

for row in range(5):
    for col in range(5):
        x = -140 + col * shag
        y = -140 + row * shag
        artist.goto(x, y)
        artist.dot(16, "#5B24F9")
print("Сетка готова.")
```

## Ukończenie (wykonaj raz, na samym końcu)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
