## Cel

Opanowanie cyklu sentinel to specjalne znaczenie jako sygnał stop.

## Pro input() w tym laptopie

Ten laptop jest automatyczny, bez żywej osoby przy klawiaturze – dlatego tu jest `input()` tymczasowo zastąpione wcześniej przygotowanymi odpowiedziami.

```python
_answers = iter(['5', '10', 'stop'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Podstawowa praktyka zadań ★

```python
summa = 0
while True:
    chislo = input("Введите число (stop — закончить): ")
    if chislo == "stop":
        break
    summa += int(chislo)
print("Сумма:", summa)
```
