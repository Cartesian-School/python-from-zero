## Pro input() w tym laptopie

Ten laptop jest automatyczny, bez żywej osoby przy klawiaturze – dlatego tu jest `input()` tymczasowo zastąpione wcześniej przygotowanymi odpowiedziami.

```python
_answers = iter(['10', '15', '11'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Sprawa robocza

```python
import random

random.seed(7)
zagadannoe = random.randint(1, 20)
popytki = 0

while True:
    popytka = int(input("Угадайте число от 1 до 20: "))
    popytki += 1

    if popytka == zagadannoe:
        print(f"Поздравляем, вы угадали за {popytki} попыток(ки)!")
        break
    elif popytka < zagadannoe:
        print("Загаданное число больше.")
    else:
        print("Загаданное число меньше.")
```
