## Pro input() w tym laptopie

Ten laptop jest automatyczny, bez żywej osoby przy klawiaturze – dlatego tu jest `input()` tymczasowo zastąpione wcześniej przygotowanymi odpowiedziami.

```python
_answers = iter(['привет', 'list', 'stop'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Podstawowa praktyka zadania ★ – Cykl Dowodzenia

```python
zhurnal = []
while True:
    komanda = input("Команда (help/list/stop): ").strip().lower()
    if komanda == "stop":
        print("Завершаю работу.")
        break
    elif komanda == "help":
        print("Доступно: help, list, stop")
    elif komanda == "list":
        print("Журнал:", zhurnal)
    else:
        zhurnal.append(komanda)
        print(f"Добавлено в журнал: {komanda}")
```

## Zadanie ★★ — loop-else

```python
chisla = [4, 8, 15, 16, 23, 42]
for n in chisla:
    if n == 100:
        naideno_100 = True
        break
else:
    naideno_100 = False
print(naideno_100)
```
