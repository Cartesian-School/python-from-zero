## Podstawowa praktyka zadań ★

Dano `x = [10, 20]`. Stwórz `y` jako PRAWDZIWA kopia x przez `.copy()`, dodaj wartość 30 do y. Następnie stwórz `z = [10, 20, 30]` i porównaj `z` z `y` przez `==` oraz `is`.

```python
x = [10, 20]
y = x.copy()
y.append(30)

z = [10, 20, 30]
ravny = (z == y)
odin_i_tot_zhe = (z is y)

print(x, y, ravny, odin_i_tot_zhe)
```
