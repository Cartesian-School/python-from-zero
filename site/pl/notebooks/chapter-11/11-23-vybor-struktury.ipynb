## Podstawowa praktyka zadań ★

Dla każdego scenariusza zapisz nazwę odpowiedniego typu (`"list"`, `"tuple"`, `"set"` lub `"dict"`) do odpowiadającej zmiennej:

1. `otvet_1` — przechowywanie profilu użytkownika: imię → email.
2. `otvet_2` jest lista unikalny ID odwiedzających, kolejność nie jest istotna.
3. `otvet_3` jest współrzędną punktu, która nie powinna się zmieniać.
4. `otvet_4` jest lista zakupów, które uzupełniamy i wykreślimy.

```python
otvet_1 = "dict"
otvet_2 = "set"
otvet_3 = "tuple"
otvet_4 = "list"
print(otvet_1, otvet_2, otvet_3, otvet_4)
```
