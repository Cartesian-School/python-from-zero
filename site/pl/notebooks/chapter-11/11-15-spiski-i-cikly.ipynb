## Cel

Iteracja listy, enumerate() i filtrowanie według warunków.

## Sprawa robocza

```python
scores = [95, 82, 91, 58, 77]
for score in scores:
    if score >= 90:
        print(score, "— отлично!")
```

## Podstawowa praktyka zadań ★

Od `scores` zbierz lista `otlichniki` (tylko oglądalność ≥ 90) oraz lista pary `otlichniki_s_indeksami` jest (indeks, wynik) dla tych samych wartości przy użyciu enumerate().

```python
scores = [95, 82, 91, 58, 77]

otlichniki = []
for score in scores:
    if score >= 90:
        otlichniki.append(score)

otlichniki_s_indeksami = []
for i, score in enumerate(scores):
    if score >= 90:
        otlichniki_s_indeksami.append((i, score))

print(otlichniki)
print(otlichniki_s_indeksami)
```
