## Cel

Buduj list/set/dict comprehension.

## Podstawowa praktyka zadań ★

Budowę: lista kwadraty o liczbach parzystych od 1 do 10 (`kvadraty_chetnyh`), zbiór unikalne małe litery słowa `"Programming"` (`bukvy`), oraz słownik „długość słowa →” `["python", "git", "sql"]` (`slovar_dlin`) — przez cały comprehension.

```python
kvadraty_chetnyh = [n ** 2 for n in range(1, 11) if n % 2 == 0]
bukvy = {ch.lower() for ch in "Programming"}
slovar_dlin = {word: len(word) for word in ["python", "git", "sql"]}

print(kvadraty_chetnyh)
print(sorted(bukvy))
print(slovar_dlin)
```
