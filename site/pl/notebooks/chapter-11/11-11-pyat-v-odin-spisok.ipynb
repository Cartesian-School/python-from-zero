## Cel

Przekształcić pięć oddzielnych zmiennych w jedną lista.

## Sprawa robocza

```python
score_1 = 95
score_2 = 82
score_3 = 91
score_4 = 77
score_5 = 88
print(score_1, score_2, score_3, score_4, score_5)
```

## Podstawowa praktyka zadań ★

Zbierz te same pięć liczb w lista `ocenki` i wyjść jego długość.

```python
ocenki = [95, 82, 91, 77, 88]
print(ocenki)
print(len(ocenki))
```
