Praktyka do sekcji [„Mini-projekt to automatyczna, wielokolorowa gwiazda”

## Cel

Lista Kolorów + Turtle.

## Ustawienie (wykonać raz)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Sprawa robocza

```python
artist.reset()
cveta = ["red", "orange", "yellow", "green", "blue"]

for cvet in cveta:
    artist.pencolor(cvet)
    artist.forward(150)
    artist.right(144)

artist.pencolor("black")
print("Звезда готова, лучей:", len(cveta))
```

## Zadanie niezależne od zadania ★★

Dodaj więcej kolorów do lista, aby gwiazda automatycznie się dostosowała.

```python
artist.reset()
cveta = ["red", "orange", "yellow", "green", "blue", "purple", "pink"]

for cvet in cveta:
    artist.pencolor(cvet)
    artist.forward(150)
    artist.right(144)

artist.pencolor("black")
print("Звезда с", len(cveta), "лучами готова.")
```
