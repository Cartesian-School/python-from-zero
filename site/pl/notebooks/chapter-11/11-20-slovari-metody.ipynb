Mistrz setdefault(), pop() i keys().

## Podstawowa praktyka zadań ★

Policz liczbę razy, ile każde słowo pojawia się `words`używania `setdefault()`. Zapisz posortowany klucz lista PRZED usunięciem na `vse_klyuchi`, a potem usuń klucz `"b"` przez `.pop()`, zachowując zdalną wartość w `udalyonnoe`.

```python
words = ["a", "b", "a", "c", "b", "a"]
counts = {}
for word in words:
    counts.setdefault(word, 0)
    counts[word] += 1

vse_klyuchi = sorted(counts.keys())
udalyonnoe = counts.pop("b")

print(counts)
print(vse_klyuchi)
print(udalyonnoe)
```
