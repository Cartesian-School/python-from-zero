# 11-21 · Struktury zagnieżdżone

Praktyka do sekcji [„Zagnieżdżone struktury”

## Cel

Czytaj dane z listy słowników z zagnieżdżonymi słownikami.

## Podstawowa praktyka zadań ★

Wyjmij Anna python klasę, Bob z matematyki i zbierz lista wszystkie nazwiska.

```python
students = [
    {"name": "Anna", "scores": {"math": 95, "python": 100}},
    {"name": "Bob", "scores": {"math": 82, "python": 78}},
]

anna_python = students[0]["scores"]["python"]
bob_math = students[1]["scores"]["math"]
imena = [student["name"] for student in students]

print(anna_python, bob_math, imena)
```
