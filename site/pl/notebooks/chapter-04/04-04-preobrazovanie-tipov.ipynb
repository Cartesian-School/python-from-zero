# 04-04 · Konwersja typów numerów

Praktyka do sekcji [„Konwersja typów liczb”

## Cel

Mistrzu `int()`, `float()`, `str()` oraz ich pułapki.

## Co trzeba wiedzieć

Każdy typ liczbowy jest jednocześnie funkcją-konwerterem.

## Krótkie przypomnienie

`int()` **odrzuca** część ułamkową, a nie zaokrągla.

## Sprawa robocza

```python
print(int(3.99))
print(float(7))
print(str(42), type(str(42)))
```

## Eksperyment 1

Porównaj `int(3.99)` oraz `int(-3.99)` — gdzie „odrzucana” jest część ułamkowa przy liczbie ujemnej?

```python
print(int(3.99))
print(int(-3.99))
```

## Eksperyment 2

Konwertowanie tekstu `"25"` do liczby i dodaj do niej 5.

```python
age_text = "25"
age = int(age_text)
print(age + 5)
```

## Typowy błąd

Nie każdy tekst można przerobić na liczbę.

```python
int("десять")
```

## Poprawka

Tylko tekst, który faktycznie wygląda jak liczba, można przekonwertować.

```python
int("10")
```

## Podstawowa praktyka zadań ★

Napraw błąd: złóż sznurek `"Итого: 100"` z tekstu `"Итого: "` i liczby `100` — pierwszy do `str()`, a następnie przez f- linia.

```python
total = 100
print("Итого: " + str(total))
print(f"Итого: {total}")
```
