# 04-07 · Translator systemu numerów

Praktyka do sekcji [„Systemy liczb”

## Cel

Opanować bin(), oct(), hex() i int() z podaniem podstawy.

## Sprawa robocza

```python
print(bin(10))
print(oct(10))
print(hex(10))
```

## Eksperyment 1

Przelicz liczbę 255 na wszystkie trzy systemy.

```python
n = 255
print(bin(n))
print(oct(n))
print(hex(n))
```

## Eksperyment 2

Teraz w drugą stronę: przekształć tekst `"FF"` (hex) oraz `"1010"` (binary) powrót do standardowych liczb.

```python
print(int("FF", 16))
print(int("1010", 2))
```

## Podstawowa praktyka zadań ★

Output bin(), oct() i hex() dla roku urodzenia.

```python
year = 2010
print(bin(year))
print(oct(year))
print(hex(year))
```
