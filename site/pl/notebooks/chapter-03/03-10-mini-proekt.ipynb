## Cel

By połączyć wszystko, co już znane — `input()`, nazwy, `print()` z `sep`/`end` w jeden mały, integralny program.

## Co trzeba wiedzieć

Wszystkie niezbędne narzędzia zostały już omówione w tym rozdziale: `input()`, przywłaszczenie, `print()`.

## Krótkie przypomnienie

Dla pewnej weryfikacji w poniższym przykładzie imiona są ustalone z góry, jakby użytkownik je już wprowadził — w własnym projekcie zastąp to prawdziwymi `input()`.

## Sprawa robocza

```python
name = "Анна"
city = "Варшава"
goal = "игру"

print("========================")
print(" МОЯ PYTHON-КАРТОЧКА")
print("========================")
print("Имя:", name)
print("Город:", city)
print("Хочу создать:", goal)
print("========================")
```

## Eksperyment 1

Teraz wprowadź swoje dane przez `input()`.

```python
name = input("Ваше имя: ")
city = input("Ваш город: ")
goal = input("Что вы хотите создать на Python? ")

print("========================")
print(" МОЯ PYTHON-КАРТОЧКА")
print("========================")
print("Имя:", name)
print("Город:", city)
print("Хочу создать:", goal)
print("========================")
```

## Podstawowa praktyka zadań ★

Dodaj czwartą linijkę do swojej wizytówki — swojego ulubionego języka programowania (możesz zakodować na stałe `"Python"`, a można zapytać przez `input()`).

```python
name = "Анна"
city = "Варшава"
goal = "игру"
language = "Python"

print("========================")
print(" МОЯ PYTHON-КАРТОЧКА")
print("========================")
print("Имя:", name)
print("Город:", city)
print("Хочу создать:", goal)
print("Любимый язык:", language)
print("========================")
```

## Samodzielna praktyka ★★ · CHALLENGE

Spróbuj własnego projektu: inna rama (np. z `*` zamiast `=`), wyrównanie przez `sep`, czyli separator `—` zamiast `:`.

```python
name = "Анна"
print("*" * 24)
print("Имя", name, sep=" — ")
print("*" * 24)
```
