# 03-03 · Twoja skorupa potrafi liczyć

Praktyka do sekcji [„Twoja skorupa może się liczyć”

## Cel

Używaj Python jako kalkulatora – dodawanie, odejmowanie, mnożenie, dzielenie.

## Co trzeba wiedzieć

Podstawowe znaki: `+`, `-`, `*` (mnożenie), `/` (dywizja).

## Krótkie przypomnienie

`/` zawsze zwraca liczbę zmiennoprzecinkową, nawet jeśli wynik dzieli się całkowicie — więcej o tym w rozdziale 4.

## Sprawa robocza

```python
print(2 + 2)
print(10 - 4)
print(3 * 7)
print(10 / 4)
```

## Eksperyment 1

Spróbuj podzielić dwie liczby tak, aby wynik dzielił się bez reszty — na przykład 10 przez 2. Jaki będzie wynik: `5` lub `5.0`?

```python
print(10 / 2)
```

## Eksperyment 2

Kolejność działań w Python jest taka sama jak w matematyce: najpierw mnożenie i dzielenie, potem dodawanie i odejmowanie. Sprawdź.

```python
print(2 + 3 * 4)   # сначала 3 * 4, потом + 2
print((2 + 3) * 4) # скобки меняют порядок
```

## Podstawowa praktyka zadań ★

Policz ile minut jest w trzech i pół godziny, używając wyrażenia c `*`.

```python
print(3.5 * 60)
```

## Sprawdzenie wyniku

```python
assert 3.5 * 60 == 210.0
print("Проверка пройдена: 210 минут.")
```
