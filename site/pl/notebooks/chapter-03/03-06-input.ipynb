## Cel

Mistrzu `input()` jest pierwszym poleceniem, które uczyni program naprawdę interaktywnym.

## Co trzeba wiedzieć

`input("вопрос")` drukuje pytanie, czeka, aż użytkownik wpisze tekst i naciśnie Enter, i zwraca wpisany tekst.

## Krótkie przypomnienie

W praktyce przeglądarki `input()` też działa – Pyodide-runner pokaże prawdziwe pole
wejście. W roboczym przykładzie poniżej, dla poprawności niezawodności, natychmiast przekazujemy gotową odpowiedź, tak aby
przykład był sprawdzany automatycznie — ale w następnym eksperymencie wpiszesz tekst samodzielnie.

## Sprawa robocza

```python
name = "Cartesian"  # представим, что это ответ, который ввёл пользователь
print("Привет,", name)
```

## Eksperyment 1

Teraz naprawdę: uruchom tę komórkę i wpisz swoje imię w pole, które się pojawi.

```python
name = input("Как вас зовут? ")
print("Привет,", name)
```

## Eksperyment 2

`input()` zawsze zwraca **tekst** — nawet jeśli użytkownik wpisał numery. Sprawdźmy, co pokaże `type()`?

```python
age_text = input("Сколько вам лет? ")
print(type(age_text))
```

## Typowy błąd

Próba natychmiastowego użycia wpisanego tekstu jako liczby (np. dodanie do liczby) skutkuje błędem — Python nie dodaje bezpośrednio tekstu i liczby. Pełne rozwiązanie znajduje się w następujących rozdziałach; tutaj wystarczy wiedzieć, że `input()` odpowiedział wiadomości.

```python
age_text = input("Сколько вам лет? ")
print(age_text + 1)
```

## Poprawka (wstępna)

Konwertowanie tekstu na liczbę pomoże `int(...)` — szczegółowo omówimy to w następnym rozdziale; tutaj — tylko demonstracja.

```python
age_text = input("Сколько вам лет? ")
print(int(age_text) + 1)
```

## Podstawowa praktyka zadań ★

Poproś o nazwę użytkownika przez `input()` i wyświetl powitanie w formacie `Привет, <имя>! Рады знакомству.`.

```python
name = input("Как вас зовут? ")
print("Привет,", name + "!", "Рады знакомству.")
```
