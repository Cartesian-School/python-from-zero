# 09-14 · and

Praktyka do sekcji [„and — wszystkie warunki naraz”

## Cel

Mistrzu and – wszystkie warunki muszą być True.

## Sprawa robocza

```python
age = 20
has_ticket = True
print(age >= 18 and has_ticket)
```

## Podstawowa praktyka zadań ★

Build can_enter jest True tylko wtedy, gdy age > = 18 I has_ticket jednocześnie.

```python
age = 20
has_ticket = True
can_enter = age >= 18 and has_ticket
print(can_enter)
```
