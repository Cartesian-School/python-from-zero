#09-11 · Truthiness i None

Praktyka do sekcji [„Truthiness i None”

## Cel

Mistrz truthiness i właściwe testy None.

## Sprawa robocza

```python
value = None
print(value is None)
print(bool(value))
```

## Podstawowa praktyka zadań ★

Sprawdź value na None przez `is`, i sprawdź truthiness linie „False” i liczby 0.

```python
value = None
is_none = value is None
truthy_check = bool("False")
falsy_check = bool(0)
print(is_none, truthy_check, falsy_check)
```
