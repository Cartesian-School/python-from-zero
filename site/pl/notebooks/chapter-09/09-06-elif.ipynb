## Cel

Master elif łańcuchy i warunki zagnieżdżenia.

## Sprawa robocza

```python
ocenka = 87
if ocenka >= 90:
    bukva = "A"
elif ocenka >= 80:
    bukva = "B"
elif ocenka >= 70:
    bukva = "C"
else:
    bukva = "D"

print(f"Оценка: {bukva}")
```

## Eksperyment 1

Sprawdź kilka różnych ocen przez cykl (z lekkim spojrzeniem do przodu, rozdział 10).

```python
for ocenka in [95, 82, 71, 40]:
    if ocenka >= 90:
        bukva = "A"
    elif ocenka >= 80:
        bukva = "B"
    elif ocenka >= 70:
        bukva = "C"
    else:
        bukva = "D"
    print(ocenka, "->", bukva)
```

## Eksperyment 2 - Warunki zagnieżdżone

```python
age = 25
has_license = True

if age >= 18:
    if has_license:
        print("Можно водить машину.")
    else:
        print("Сначала получите права.")
else:
    print("Ещё рано водить машину.")
```

## Zadanie niezależne od zadania ★★

Dodaj trzeci poziom zagnieżdżenia: sprawdzenie dostępności paliwa (`has_fuel`).

```python
age = 25
has_license = True
has_fuel = False

if age >= 18:
    if has_license:
        if has_fuel:
            print("Можно ехать!")
        else:
            print("Права есть, но бак пуст.")
    else:
        print("Сначала получите права.")
else:
    print("Ещё рано водить машину.")
```
