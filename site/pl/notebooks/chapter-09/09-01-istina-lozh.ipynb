# 09-01 · Prawda czy fałsz

Praktyka do sekcji [„Prawda lub fałsz”

## Cel

Opanować rodzaj bool i prawdziwość wartości różnych typów.

## Sprawa robocza

```python
is_sunny = True
print(is_sunny, type(is_sunny))
```

## Eksperyment 1

```python
print(bool(0))
print(bool(42))
print(bool(""))
print(bool("нет"))
```

## Eksperyment 2 – Nieoczekiwany przypadek

```python
print(bool("False"))   # True! строка непустая, значит "истина"
print(bool([]))        # пустой список — тоже "ложь", забегаем немного вперёд (глава 11)
```

## Podstawowa praktyka zadań ★

Sprawdź prawdziwość pięciu różnych wybranych przez siebie wartości — w tym co najmniej jednej, dla której wynik nie jest oczywisty z góry.

```python
for value in [0, 1, -1, "", " ", "0"]:
    print(repr(value), "->", bool(value))
```
