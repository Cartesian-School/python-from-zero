## Cel

Polimorfizm bez wspólnego rodzica — istotna jest tylko obecność metody.

## Sprawa robocza

```python
class Chelovek:
    def __init__(self, imya):
        self.imya = imya

    def predstavitsya(self):
        return f"Привет, я {self.imya}!"


class Robot:
    def __init__(self, nomer):
        self.nomer = nomer

    def predstavitsya(self):
        return f"БИП. Я робот номер {self.nomer}."


def poznakomit(obj):
    return obj.predstavitsya()

print(poznakomit(Chelovek("Аня")))
print(poznakomit(Robot(7)))
```

## Zadanie niezależne od zadania ★★

`Chelovek` oraz `Robot` nie mają wspólnego rodzica (poza `object`). Sprawdź, czy `poznakomit()` i tak działa z obiema i zachowaj oba wyniki.

```python
rezultat_chelovek = poznakomit(Chelovek("Боря"))
rezultat_robot = poznakomit(Robot(42))
print(rezultat_chelovek)
print(rezultat_robot)
```
