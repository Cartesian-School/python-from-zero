# 14-23 · dataclasses

Praktyka do rozdziału [„dataclasses”](/pl/chapters/rozdzial-14/14-23-dataclasses.html).

## Cel

@dataclass i field(default_factory=...) - Unikanie pułapki zmiennej wartości domyślnej.

## Sprawa robocza

```python
from dataclasses import dataclass, field

@dataclass
class Korzina:
    tovary: list = field(default_factory=list)

    def dobavit(self, tovar):
        self.tovary.append(tovar)

k1 = Korzina()
k2 = Korzina()
k1.dobavit("Хлеб")
print(k1.tovary, k2.tovary)
```

## Podstawowa praktyka zadań ★

Stwórz trzeci wózek `k3` i upewnij się, że zaczyna się pusto, niezależnie od `k1` oraz `k2`.

```python
k3 = Korzina()
k3.dobavit("Молоко")
print(k1.tovary, k2.tovary, k3.tovary)
```
