# 14-08 · Powszechna pułapka zmiennych atrybutów

Praktyka do sekcji [„Pułapka wspólnej zmiennej cechy”

## Cel

Poprawnie zadeklaruj atrybut zmienny — wewnątrz __init__, a nie w ciele klasy.

## Sprawa robocza

```python
class Korzina:
    def __init__(self):
        self.tovary = []

    def dobavit(self, tovar):
        self.tovary.append(tovar)

k1 = Korzina()
k2 = Korzina()
k1.dobavit("Хлеб")
print(k1.tovary, k2.tovary)
```

## Podstawowa praktyka zadań ★

Utwórz klasę `Zametki` z atrybutem instancji `spisok = []` (utworzony w `__init__`, nie w treści zajęć!) i `dobavit_zametku(tekst)`. Stwórz `z1` oraz `z2`, dodaj do każdego różne wpisy i upewnij się, że się nie mieszają.

```python
class Zametki:
    def __init__(self):
        self.spisok = []

    def dobavit_zametku(self, tekst):
        self.spisok.append(tekst)

z1 = Zametki()
z2 = Zametki()
z1.dobavit_zametku("Купить хлеб")
z2.dobavit_zametku("Позвонить маме")
print(z1.spisok)
print(z2.spisok)
```
