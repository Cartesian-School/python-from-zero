# 14-16 · super()

Praktyka do sekcji [„super() i procedura rozwiązywania metod”

## Cel

Uzupełnić __init__ rodzica, a nie zastępować go w całości.

## Sprawa robocza

```python
class Zhivotnoe:
    def __init__(self, klichka):
        self.klichka = klichka


class Sobaka(Zhivotnoe):
    def __init__(self, klichka, poroda):
        super().__init__(klichka)
        self.poroda = poroda

rex = Sobaka("Рекс", "Дворняга")
print(rex.klichka, rex.poroda)
```

## Zadanie niezależne od zadania ★★

Utwórz klasę `Koshka(Zhivotnoe)` __init__(self, klichka, okras), prowokacyjne `super().__init__(klichka)`, a następnie ustawiająca `self.okras`.

```python
class Koshka(Zhivotnoe):
    def __init__(self, klichka, okras):
        super().__init__(klichka)
        self.okras = okras

murka = Koshka("Мурка", "рыжий")
print(murka.klichka, murka.okras)
```
