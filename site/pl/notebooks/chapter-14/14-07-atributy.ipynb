## Cel

Odróżnić atrybut klasy (wspólny) od atrybutu instancji (własnego).

## Sprawa robocza

```python
class Sobaka:
    vid = "Собака"

    def __init__(self, klichka):
        self.klichka = klichka

rex = Sobaka("Рекс")
sharik = Sobaka("Шарик")
print(rex.vid, sharik.vid)
```

## Podstawowa praktyka zadań ★

Utwórz klasę `Rabotnik` z atrybutem klasy `company = "Cartesian School"` i atrybut instancji `imya`. Stwórz `w1` oraz `w2`, a potem przypisz `w1.company` nową wartość bezpośrednio — `w2.company` nie powinno się zmieniać.

```python
class Rabotnik:
    company = "Cartesian School"

    def __init__(self, imya):
        self.imya = imya

w1 = Rabotnik("Аня")
w2 = Rabotnik("Боря")
w1.company = "Другая компания"
print(w1.company, w2.company, Rabotnik.company)
```
