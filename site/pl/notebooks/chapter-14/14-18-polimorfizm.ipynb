## Cel

Jedno wywołanie metody – różne zachowanie w zależności od klasy obiektu.

## Sprawa robocza

```python
class Sobaka:
    def zvuk(self):
        return "Гав!"


class Koshka:
    def zvuk(self):
        return "Мяу!"


class Korova:
    def zvuk(self):
        return "Му!"


def poluchit_zvuk(zhivotnoe):
    return zhivotnoe.zvuk()

zhivotnye = [Sobaka(), Koshka(), Korova()]
zvuki = [poluchit_zvuk(zh) for zh in zhivotnye]
print(zvuki)
```

## Podstawowa praktyka zadań ★

Sprawdź `poluchit_zvuk()` w każdej klasie osobno.

```python
zvuk_sobaki = poluchit_zvuk(Sobaka())
zvuk_koshki = poluchit_zvuk(Koshka())
zvuk_korovy = poluchit_zvuk(Korova())
print(zvuk_sobaki, zvuk_koshki, zvuk_korovy)
```
