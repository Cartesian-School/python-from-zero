## Zadanie niezależne od zadania ★★

Utwórz klasę `Schet` jest licznikiem o stanie, który gromadzi się między wywołaniami: `value`zaczynając od 0 i `increment()`który zwiększa go o 1. Tutaj klasa jest uzasadniona — stan musi być zachowany MIĘDZY wywołaniami, czego funkcja sama nie jest w stanie zrobić.

```python
class Schet:
    def __init__(self):
        self.value = 0

    def increment(self):
        self.value += 1

schet = Schet()
schet.increment()
schet.increment()
print(schet.value)
```
