# 17-19 · GameState z dataclass

Praktyka do rozdziału [„GameState z dataclass”](/pl/chapters/rozdzial-17/17-19-gamestate-dataclass.html).

## Cel

Zbierz GameState jako @dataclass i sprawdź, że board NIE jest współdzielone między instancjami.

## Sprawa robocza

```python
from dataclasses import dataclass, field

@dataclass
class GameState:
    board: list[str] = field(default_factory=lambda: [""] * 9)
    current_player: str = "X"
    game_over: bool = False
    winner: str | None = None
    winning_line: tuple[int, int, int] | None = None
    score_x: int = 0
    score_o: int = 0
    draws: int = 0

s1 = GameState()
s2 = GameState()
print(s1)
```

## Zadanie ★★ Zadanie samodzielne — niezależność egzemplarzy

```python
s1.board[0] = "X"

assert s1.board[0] == "X"
assert s2.board[0] == "", "изменение board у s1 НЕ должно быть видно в s2 (field(default_factory=...))"
assert s1.current_player == "X" and s2.current_player == "X"
print("s1 и s2 — полностью независимые экземпляры GameState.")
```
