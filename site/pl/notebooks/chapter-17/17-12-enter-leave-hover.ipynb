# 17-12 · Mouse Enter/Leave i hover

Praktyka do sekcji [„Mouse Enter/Leave i hover”

## Cel

Zbuduj prosty efekt hover- przez<Enter>/<Leave>— bez jednego kliknięcia.

## Pro mainloop() w tym laptopie

Zamiast `root.mainloop()` używany tutaj `root.update()` + `root.destroy()` jest takie, że aplikacja jest zbudowana dokładnie tak samo, tylko bez niekończącego się oczekiwania na końcu (zobacz wyjaśnienie w rozdziale 16).

## Sprawa robocza

```python
import tkinter as tk

root = tk.Tk()
button = tk.Button(root, text="Наведите на меня")
button.pack()

def on_enter(_event):
    button.config(bg="#B9A0FC")

def on_leave(_event):
    button.config(bg="#FAFAFC")

button.bind("<Enter>", on_enter)
button.bind("<Leave>", on_leave)
root.update()

button.event_generate("<Enter>")
root.update()
print("После Enter:", button.cget("bg"))

button.event_generate("<Leave>")
root.update()
print("После Leave:", button.cget("bg"))
root.destroy()
```

## Zadanie niezależne od zadania ★★

Ułóż trzy przyciski w rzędzie i podświetl ten najechany, zapewniając, że pozostałe dwa pozostają w zwykłym kolorze.

```python
root = tk.Tk()
OBYCHNYJ = "#FAFAFC"
PODSVETKA = "#B9A0FC"

knopki = []
for i in range(3):
    b = tk.Button(root, text=f"Кнопка {i}", bg=OBYCHNYJ)
    b.pack(side="left")
    # k=b to ta sama technika „zamrażania”
    b.bind("<Enter>", lambda _event, k=b: k.config(bg=PODSVETKA))
    b.bind("<Leave>", lambda _event, k=b: k.config(bg=OBYCHNYJ))
    knopki.append(b)
root.update()

knopki[1].event_generate("<Enter>")
root.update()
cveta = [k.cget("bg") for k in knopki]
print("Цвета после наведения на кнопку 1:", cveta)
assert cveta == [OBYCHNYJ, PODSVETKA, OBYCHNYJ], "подсвечена должна быть ровно одна кнопка"

knopki[1].event_generate("<Leave>")
root.update()
print("Цвета после ухода курсора:      ", [k.cget("bg") for k in knopki])
assert all(k.cget("bg") == OBYCHNYJ for k in knopki)
print("Подсвечивается только наведённая кнопка, остальные остаются обычными.")
root.destroy()
```
