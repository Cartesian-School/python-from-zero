Spójrz na rzeczywiste pola obiektu tkinter.Event dla różnych typów zdarzeń.

## Pro mainloop() w tym laptopie

Zamiast `root.mainloop()` używany tutaj `root.update()` + `root.destroy()` jest takie, że aplikacja jest zbudowana dokładnie tak samo, tylko bez niekończącego się oczekiwania na końcu (zobacz wyjaśnienie w rozdziale 16).

## Praktyczny przykład — zdarzenie klawiatury

```python
import tkinter as tk

root = tk.Tk()
label = tk.Label(root, text="...")
label.pack()
poslednee = {}

def on_key(event):
    poslednee["type"] = str(event.type)
    poslednee["keysym"] = event.keysym
    poslednee["char"] = event.char
    label.config(text=f"keysym={event.keysym!r} char={event.char!r}")

root.bind("<Key>", on_key)
root.update()
root.focus_force()   # без фокуса окна event_generate для клавиш никуда не доставит событие
root.update()
label.event_generate("<Key>", keysym="Return")
root.update()
print(poslednee)
```

## Eksperyment - Wydarzenie Myszy (Enter/Leave)

```python
poslednee_mouse = {}

def on_enter(event):
    poslednee_mouse["widget"] = str(event.widget)
    poslednee_mouse["type"] = str(event.type)

label.bind("<Enter>", on_enter)
root.update()
label.event_generate("<Enter>")
root.update()
print(poslednee_mouse)
# Zwróć uwagę: zdarzenia <Enter> event.keysym/event.char nie niosą użytecznej informacji —
# to są dziedziny, które mają sens dla wydarzeń klawiaturowych.
root.destroy()
```

## Zadanie niezależne od zadania ★★

Powiąż `<Button-1>` do oddzielnego widżetu i wydrukuj `event.x`/`event.y` — współrzędne kliknięcia wewnątrz widżetu.

```python
root = tk.Tk()
metka = tk.Label(root, text="Кликните по мне", width=20, height=3)
metka.pack()
koordinaty = {}

def na_klik(event):
    koordinaty["x"] = event.x
    koordinaty["y"] = event.y

metka.bind("<Button-1>", na_klik)
root.update()

# programowo kliknąć na punkt (12, 7) WEWNĄTRZ widżetu
metka.event_generate("<Button-1>", x=12, y=7)
root.update()

print("Координаты клика внутри виджета:", koordinaty)
assert koordinaty == {"x": 12, "y": 7}
print("event.x/event.y отсчитываются от левого верхнего угла самого виджета, а не экрана.")
root.destroy()
```
