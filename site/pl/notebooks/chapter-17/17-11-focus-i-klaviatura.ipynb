## Cel

Zobacz, jak fokus na wejściu decyduje, który widżet otrzymuje zdarzenia klawiaturowe.

## Pro mainloop() w tym laptopie

Zamiast `root.mainloop()` używany tutaj `root.update()` + `root.destroy()` jest takie, że aplikacja jest zbudowana dokładnie tak samo, tylko bez niekończącego się oczekiwania na końcu (zobacz wyjaśnienie w rozdziale 16).

## Sprawa robocza

```python
import tkinter as tk

root = tk.Tk()
entry = tk.Entry(root)
entry.pack()
button = tk.Button(root, text="Кнопка")
button.pack()
root.update()
root.focus_force()   # окну нужен фокус ОС, иначе фокус внутри него никому не достанется
root.update()

entry.focus_set()
root.update()
print("Фокус на Entry:", root.focus_get() is entry)

button.focus_set()
root.update()
print("Фокус на Button:", root.focus_get() is button)
root.destroy()
```

## Zadanie niezależne od zadania ★★

Powiąż `<Key>` zarówno do root, jak i do entry (u entry — z obsługą, która wyraźnie zwraca `"break"`). Ustaw fokus na entry i sprawdź: handler entry zadziała, a handler root — nie. Następnie usuń `"break"` u entry i sprawdź jeszcze raz: teraz zadziałają OBA obsługiwacze. Wniosek: zdarzenie zatrzymuje dokładnie `"break"` gdzieś po drodze bindtags, a nie fakt, że uwaga skupia się na innym widżecie.

```python
root = tk.Tk()
entry = tk.Entry(root)
entry.pack()
root.update()
root.focus_force()
entry.focus_set()
root.update()

popalo_v_root = []
root.bind("<Key>", lambda e: popalo_v_root.append(e.keysym))

# 1) entry ma własnego handlera, który zwraca „break”
entry.bind("<Key>", lambda e: "break")
entry.event_generate("<KeyPress-5>")
root.update()
print("С 'break':   root получил", popalo_v_root, "| в поле:", repr(entry.get()))
assert popalo_v_root == [], "'break' обязан оборвать цепочку до root"

# 2) usuwa „break”
popalo_v_root.clear()
entry.delete(0, "end")
entry.unbind("<Key>")
entry.event_generate("<KeyPress-5>")
root.update()
print("Без 'break': root получил", popalo_v_root, "| в поле:", repr(entry.get()))
assert popalo_v_root == ["5"], "без 'break' событие обязано дойти и до root"
assert entry.get() == "5", "и при этом символ всё равно попадает в поле ввода"

print("\nЦепочку останавливает именно 'break', а не сам факт того,")
print("что фокус находится на другом виджете.")
root.destroy()
```
