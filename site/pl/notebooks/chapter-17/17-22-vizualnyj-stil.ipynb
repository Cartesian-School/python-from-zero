Praktyka do sekcji [„Styl wizualny X/O”

## Cel

Barwnik X fioletowy, O różowy przez prosty fg= przy tk.Button.

## Pro mainloop() w tym laptopie

Zamiast `root.mainloop()` używany tutaj `root.update()` + `root.destroy()` jest takie, że aplikacja jest zbudowana dokładnie tak samo, tylko bez niekończącego się oczekiwania na końcu (zobacz wyjaśnienie w rozdziale 16).

## Sprawa robocza

```python
import tkinter as tk

MARK_COLOR = {"X": "#5B24F9", "O": "#DB2777"}

root = tk.Tk()
btn = tk.Button(root, text="X", fg=MARK_COLOR["X"], font=("Arial", 24, "bold"))
btn.pack()
root.update()
print("Цвет X:", btn.cget("fg"))

btn.config(text="O", fg=MARK_COLOR["O"])
root.update()
print("Цвет O:", btn.cget("fg"))
root.destroy()
```

## Zadanie niezależne od zadania ★★

Złóż trzy przyciski: jeden z X, jeden z O i jeden z pustym, i sprawdź, czy MARK_COLOR.get(text, „#0D0230”

```python
NEJTRALNYJ = "#0D0230"

root = tk.Tk()
kletki = []
for tekst in ("X", "O", ""):
    b = tk.Button(root, text=tekst, fg=MARK_COLOR.get(tekst, NEJTRALNYJ),
                  font=("Arial", 24, "bold"), width=3)
    b.pack(side="left")
    kletki.append(b)
root.update()

cveta = [b.cget("fg") for b in kletki]
print("Цвета X / O / пустой клетки:", cveta)
assert cveta == ["#5B24F9", "#DB2777", NEJTRALNYJ]
print("Для пустой клетки в MARK_COLOR ключа нет — .get() отдаёт нейтральный цвет.")
root.destroy()
```
