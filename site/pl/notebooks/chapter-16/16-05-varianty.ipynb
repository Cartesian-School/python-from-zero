## Pro mainloop() w tym laptopie

W normalnym życiu `.py`- pliku aplikacja jest uruchamiana przez `root.mainloop()` - To polecenie „zamraża” `root.mainloop()` zadzwonimy `root.update()` (obsługuje zdarzenia raz, bez oczekiwania) a następnie `root.destroy()` — interfejs buduje się dokładnie tak samo, po prostu bez nieskończonego oczekiwania na końcu.

## Przykładem roboczym jest Radiobutton

```python
import tkinter as tk

root = tk.Tk()
vybor = tk.StringVar(value="chay")

tk.Radiobutton(root, text="Чай", variable=vybor, value="chay").pack()
tk.Radiobutton(root, text="Кофе", variable=vybor, value="kofe").pack()

vybor.set("kofe")   # программно выбираем вариант
root.update()
print("Выбрано:", vybor.get())
root.destroy()
```

## Eksperyment 1 – Checkbutton

```python
root = tk.Tk()
saharok = tk.BooleanVar(value=False)
tk.Checkbutton(root, text="С сахаром", variable=saharok).pack()

saharok.set(True)
root.update()
print("С сахаром:", saharok.get())
root.destroy()
```
