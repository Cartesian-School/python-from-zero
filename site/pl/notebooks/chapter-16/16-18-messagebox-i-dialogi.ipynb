## Pro mainloop() w tym laptopie

W normalnym życiu `.py`- pliku aplikacja jest uruchamiana przez `root.mainloop()` - To polecenie „zamraża” `root.mainloop()` zadzwonimy `root.update()` (obsługuje zdarzenia raz, bez oczekiwania) a następnie `root.destroy()` — interfejs buduje się dokładnie tak samo, po prostu bez nieskończonego oczekiwania na końcu.

## Pro messagebox w tym laptopie

Dialogi modalne czekają na rzeczywiste naciśnięcie przycisku przez żywego użytkownika. W automatycznie wykonywanym notatniku nikt nie naciśnie przycisku, dlatego potrzebne funkcje `messagebox` tymczasowo zastępowane przez przygotowane odpowiedzi.

```python
from tkinter import messagebox

messagebox.askyesno = lambda *args, **kwargs: True
messagebox.showinfo = lambda *args, **kwargs: None
```

```python
import tkinter as tk

root = tk.Tk()
otvet = messagebox.askyesno("Выход", "Сохранить изменения перед выходом?")
if otvet:
    messagebox.showinfo("Готово", "Изменения сохранены.")
print("Ответ пользователя:", otvet)
root.destroy()
```
