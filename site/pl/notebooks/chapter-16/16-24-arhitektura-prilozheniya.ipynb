## Cel

Sprawdź czystą funkcję domenową i napraw pułapkę lambda w pętli.

```python
def calculate_tip(amount, percent, people):
    total_tip = amount * percent / 100
    return total_tip / people

result = calculate_tip(1000, 15, 2)
print(result)
```

## Zadanie ★★ zadaniowe samodzielne – późne linkowanie do lambda

```python
callbacks = []
for i in range(3):
    callbacks.append(lambda i=i: i)

values = [cb() for cb in callbacks]
print(values)
```
