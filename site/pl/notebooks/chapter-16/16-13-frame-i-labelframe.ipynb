## Pro mainloop() w tym laptopie

W normalnym życiu `.py`- pliku aplikacja jest uruchamiana przez `root.mainloop()` - To polecenie „zamraża” `root.mainloop()` zadzwonimy `root.update()` (obsługuje zdarzenia raz, bez oczekiwania) a następnie `root.destroy()` — interfejs buduje się dokładnie tak samo, po prostu bez nieskończonego oczekiwania na końcu.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()

toolbar_frame = ttk.Frame(root)
content_frame = ttk.Frame(root)
status_frame = ttk.Frame(root)

toolbar_frame.pack(side="top", fill="x")
content_frame.pack(side="top", fill="both", expand=True)
status_frame.pack(side="bottom", fill="x")

root.update()
print(len(root.winfo_children()), "региона в root")
root.destroy()
```

## Podstawowe ćwiczenie zadań ★ – LabelFrame

```python
root = tk.Tk()
nastrojki = ttk.LabelFrame(root, text="Настройки")
nastrojki.pack(padx=10, pady=10, fill="x")
ttk.Checkbutton(nastrojki, text="Тёмная тема").pack(anchor="w")
root.update()
print(nastrojki.cget("text"))
root.destroy()
```
