## Цель

Увидеть разницу между поверхностной и глубокой копией на вложенном списке.

## Задание ★ Базовая практика

Сравните .copy() и copy.deepcopy() на списке списков: после изменения вложенного элемента через копию проверьте, изменился ли оригинал в каждом случае.

```python
original = [["Anna", 10], ["Bob", 20]]
melkaya_kopiya = original.copy()
melkaya_kopiya[0][1] = 999
posle_melkoy = original[0][1]

import copy
original2 = [["Anna", 10], ["Bob", 20]]
glubokaya_kopiya = copy.deepcopy(original2)
glubokaya_kopiya[0][1] = 999
posle_glubokoy = original2[0][1]

print("После поверхностной копии:", posle_melkoy)
print("После глубокой копии:", posle_glubokoy)
```
