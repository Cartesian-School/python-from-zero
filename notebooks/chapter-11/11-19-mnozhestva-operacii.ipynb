## Цель

Применить объединение, пересечение, разность и проверку подмножества.

## Задание ★ Базовая практика

Дано `required = {"python", "git", "sql"}` и `available = {"python", "git", "docker"}`. Найдите, каких навыков не хватает (`missing`), какие уже есть (`common`), объединение всех навыков (`vse`), и проверьте, является ли `{"python", "git"}` подмножеством `available` (`podmnozhestvo`).

```python
required = {"python", "git", "sql"}
available = {"python", "git", "docker"}

missing = required - available
common = required & available
vse = required | available
podmnozhestvo = {"python", "git"} <= available

print(missing)
print(common)
print(vse)
print(podmnozhestvo)
```
