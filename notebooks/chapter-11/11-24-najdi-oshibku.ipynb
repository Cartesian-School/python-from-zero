## Цель

Найти и исправить типичные ошибки в коде с коллекциями.

## Сломанный код (не выполняем)

```python
tovary = ["хлеб"]
tovary.append(["молоко", "яйца"])   # баг: список внутри списка
```

## Задание ★ Базовая практика — исправьте append() → extend()

````python
tovary = ["хлеб"]
tovary.extend(["молоко", "яйца"])
print(tovary)
````

## Задание ★★ Самостоятельная задача — исправьте потерю списка через `x = x.sort()`

````python
chisla = [5, 3, 1, 4]
chisla.sort()
print(chisla)
````
