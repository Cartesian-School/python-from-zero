## Цель

Записная книжка (dict) + сравнение множеств навыков в одном задании.

## Задание ★ Базовая практика

```python
contacts = {"Anna": "anna@example.com", "Bob": "bob@example.com"}
contacts["Maria"] = "maria@example.com"
contacts["Anna"] = "anna.new@example.com"
del contacts["Bob"]

required = {"python", "git", "sql"}
available = {"python", "git"}
missing = required - available

print(contacts)
print(missing)
```
