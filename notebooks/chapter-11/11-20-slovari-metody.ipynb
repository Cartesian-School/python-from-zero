## Задание ★ Базовая практика

Подсчитайте, сколько раз встречается каждое слово в списке `words`, используя `setdefault()`. Сохраните отсортированный список ключей ДО удаления в `vse_klyuchi`, затем удалите ключ `"b"` через `.pop()`, сохранив удалённое значение в `udalyonnoe`.

```python
words = ["a", "b", "a", "c", "b", "a"]
counts = {}
for word in words:
    counts.setdefault(word, 0)
    counts[word] += 1

vse_klyuchi = sorted(counts.keys())
udalyonnoe = counts.pop("b")

print(counts)
print(vse_klyuchi)
print(udalyonnoe)
```
