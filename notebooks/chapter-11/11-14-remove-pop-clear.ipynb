# 11-14 · remove, pop, clear, del

Практика к разделу [«remove, pop, clear, del»](../../site/chapters/glava-11/11-14-remove-pop-clear.html).

## Цель

Правильно выбрать между remove(), pop() и del.

## Рабочий пример

```python
names = ["Anna", "Oleg", "Maria"]
names.remove("Oleg")
print(names)
```

## Задание ★ Базовая практика

Дано `names = ["Anna", "Oleg", "Maria", "Leo"]`. Удалите `"Oleg"` по значению, затем удалите и сохраните в `last` последний элемент через pop(), затем удалите первый элемент через `del`.

```python
names = ["Anna", "Oleg", "Maria", "Leo"]
names.remove("Oleg")
last = names.pop()
del names[0]
print(names, last)
```
