Освоить Radiobutton и Checkbutton.

## Про mainloop() в этом ноутбуке

В обычном `.py`-файле приложение запускают через `root.mainloop()` — эта команда «замирает» и ждёт, пока пользователь не закроет окно. В автоматически выполняемом ноутбуке нет живого пользователя за окном, поэтому здесь вместо `root.mainloop()` мы вызовем `root.update()` (обрабатывает события один раз, без ожидания) и затем `root.destroy()` — интерфейс строится точно так же, просто без бесконечного ожидания в конце.

## Рабочий пример — Radiobutton

```python
import tkinter as tk

root = tk.Tk()
vybor = tk.StringVar(value="chay")

tk.Radiobutton(root, text="Чай", variable=vybor, value="chay").pack()
tk.Radiobutton(root, text="Кофе", variable=vybor, value="kofe").pack()

vybor.set("kofe")   # программно выбираем вариант
root.update()
print("Выбрано:", vybor.get())
root.destroy()
```

## Эксперимент 1 — Checkbutton

```python
root = tk.Tk()
saharok = tk.BooleanVar(value=False)
tk.Checkbutton(root, text="С сахаром", variable=saharok).pack()

saharok.set(True)
root.update()
print("С сахаром:", saharok.get())
root.destroy()
```
