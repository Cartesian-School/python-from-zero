## Цель

Потренироваться создавать имена (переменные) и обращаться к ним снова — без ошибки `NameError`.

## Что нужно знать

`имя = значение` создаёт имя, которое указывает на значение. Дальше это имя можно использовать вместо самого значения.

## Краткое напоминание

Правильнее думать не «имя — это коробка со значением», а «имя указывает на значение» — как стрелка. Одно и то же значение может быть у нескольких имён сразу.

## Рабочий пример

```python
city = "Warsaw"
print(city)
```

## Эксперимент 1

Создайте второе имя, указывающее на то же значение, и убедитесь, что оба имени работают.

```python
city = "Warsaw"
town = city
print(city)
print(town)
```

## Эксперимент 2

Перезапишите значение одного из имён и убедитесь, что второе имя не изменилось — присваивание меняет, на что указывает ИМЕННО ЭТО имя, а не всё, что на него ссылалось.

```python
city = "Warsaw"
town = city
city = "Krakow"
print(city)
print(town)
```

## Типичная ошибка

Обращение к имени, которое ещё не создано (опечатка или неправильный порядок строк), даёт `NameError`.

```python
print(favourite_city)
favourite_city = "Warsaw"
```

## Исправление

Имя должно быть создано присваиванием ДО того, как к нему обращаются.

```python
favourite_city = "Warsaw"
print(favourite_city)
```
