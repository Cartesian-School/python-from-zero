## Цель

Собрать вместе всё, что уже знакомо — `input()`, имена, `print()` с `sep`/`end` — в одну маленькую цельную программу.

## Что нужно знать

Все нужные инструменты уже пройдены в этой главе: `input()`, присваивание, `print()`.

## Краткое напоминание

Для надёжной проверки в примере ниже имена заданы заранее, как если бы пользователь их уже ввёл — в собственном проекте замените это на настоящие `input()`.

## Рабочий пример

```python
name = "Анна"
city = "Варшава"
goal = "игру"

print("========================")
print(" МОЯ PYTHON-КАРТОЧКА")
print("========================")
print("Имя:", name)
print("Город:", city)
print("Хочу создать:", goal)
print("========================")
```

## Эксперимент 1

Теперь введите свои собственные данные через `input()`.

```python
name = input("Ваше имя: ")
city = input("Ваш город: ")
goal = input("Что вы хотите создать на Python? ")

print("========================")
print(" МОЯ PYTHON-КАРТОЧКА")
print("========================")
print("Имя:", name)
print("Город:", city)
print("Хочу создать:", goal)
print("========================")
```

## Задание ★ Базовая практика

Добавьте в визитку четвёртую строку — любимый язык программирования (можно захардкодить `"Python"`, а можно спросить через `input()`).

```python
name = "Анна"
city = "Варшава"
goal = "игру"
language = "Python"

print("========================")
print(" МОЯ PYTHON-КАРТОЧКА")
print("========================")
print("Имя:", name)
print("Город:", city)
print("Хочу создать:", goal)
print("Любимый язык:", language)
print("========================")
```
