# 03-05 · Предскажите и проверьте: REPL

Практика к разделу [«REPL как инструмент для исследования»](../../site/chapters/glava-03/03-09-repl-kak-instrument.html).

## Цель

Потренироваться предсказывать, что покажет REPL/ячейка, прежде чем запускать код — и проверить себя с помощью `type()`.

## Что нужно знать

`type(значение)` показывает тип значения: `<class 'int'>`, `<class 'str'>` и так далее.

## Краткое напоминание

Ячейка (как и REPL) сама печатает результат *последнего* выражения — но не результат `print()`-вызовов, которые печатают сами по себе, независимо от того, последние они или нет.

## Рабочий пример

```python
type(42)
```

## Эксперимент 1

Прежде чем запускать — предскажите: какой тип у `"42"` (в кавычках)? Затем проверьте.

```python
type("42")
```

## Эксперимент 2

Предскажите тип результата `10 / 2` — это будет `int` или `float`? Затем проверьте.

```python
type(10 / 2)
```

## Типичная ошибка

Начинающие иногда ждут, что `dir(42)` покажет что-то понятное с первого взгляда — на деле это длинный список служебных имён. Это нормально: пока нужно уметь ЗАПУСКАТЬ `dir()`, а не понимать каждое имя в списке.

```python
len(dir(42)) > 0  # просто убеждаемся, что список не пустой
```

## Задание ★ Базовая практика

Выведите тип каждого из трёх значений: `3.14`, `True`, `[1, 2, 3]` — тремя отдельными вызовами `print(type(...))`.

```python
print(type(3.14))
print(type(True))
print(type([1, 2, 3]))
```

## Самостоятельная практика ★★

Используйте `help(print)`, чтобы прочитать встроенную справку по `print`, а затем выведите текстом одну вещь, которую вы там узнали (например, имена параметров).

```python
help(print)
print("Узнал(а): print принимает sep и end")
```
