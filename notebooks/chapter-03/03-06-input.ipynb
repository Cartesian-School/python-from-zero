## Цель

Освоить `input()` — первую команду, которая делает программу по-настоящему интерактивной.

## Что нужно знать

`input("вопрос")` печатает вопрос, ждёт, пока пользователь наберёт текст и нажмёт Enter, и возвращает введённый текст.

## Краткое напоминание

В браузерной практике `input()` тоже работает — Pyodide-раннер покажет настоящее поле
ввода. В рабочем примере ниже мы для надёжности сразу передаём готовый ответ, чтобы
пример проверялся автоматически — но в следующем эксперименте вы наберёте текст сами.

## Рабочий пример

```python
name = "Cartesian"  # представим, что это ответ, который ввёл пользователь
print("Привет,", name)
```

## Эксперимент 1

Теперь по-настоящему: запустите эту ячейку и введите своё имя в появившемся поле.

```python
name = input("Как вас зовут? ")
print("Привет,", name)
```

## Эксперимент 2

`input()` всегда возвращает **текст** — даже если пользователь ввёл цифры. Проверим: что покажет `type()`?

```python
age_text = input("Сколько вам лет? ")
print(type(age_text))
```

## Типичная ошибка

Попытка сразу использовать введённый текст как число (например, сложить с числом) приведёт к ошибке — Python не складывает текст и число напрямую. Полное решение — в следующих главах; здесь достаточно знать, что `input()` возвращает именно текст.

```python
age_text = input("Сколько вам лет? ")
print(age_text + 1)
```

## Исправление (предварительно)

Преобразовать текст в число поможет `int(...)` — подробно разберём это в следующей главе; здесь — только демонстрация.

```python
age_text = input("Сколько вам лет? ")
print(int(age_text) + 1)
```
