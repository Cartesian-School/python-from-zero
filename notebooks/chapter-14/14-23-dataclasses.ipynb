# 14-23 · dataclasses

Практика к разделу [«dataclasses»](../../site/chapters/glava-14/14-23-dataclasses.html).

## Цель

@dataclass и field(default_factory=...) — избегаем ловушки изменяемого значения по умолчанию.

## Рабочий пример

```python
from dataclasses import dataclass, field

@dataclass
class Korzina:
    tovary: list = field(default_factory=list)

    def dobavit(self, tovar):
        self.tovary.append(tovar)

k1 = Korzina()
k2 = Korzina()
k1.dobavit("Хлеб")
print(k1.tovary, k2.tovary)
```

## Задание ★ Базовая практика

Создайте третью корзину `k3` и убедитесь, что она тоже начинается пустой, независимо от `k1` и `k2`.

```python
k3 = Korzina()
k3.dobavit("Молоко")
print(k1.tovary, k2.tovary, k3.tovary)
```
