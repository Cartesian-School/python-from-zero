# 14-05 · self и связывание методов

Практика к разделу [«self и связывание методов»](../../site/chapters/glava-14/14-05-self-i-svyazyvanie-metodov.html).

## Цель

Увидеть, что obj.method() и Class.method(obj) — один и тот же вызов.

## Рабочий пример

```python
class Sobaka:
    def __init__(self, klichka):
        self.klichka = klichka

    def layat(self):
        return f"{self.klichka}: Гав-гав!"

rex = Sobaka("Рекс")
print(rex.layat())
print(Sobaka.layat(rex))
```

## Задание ★ Базовая практика

Создайте класс `Kot` с атрибутом `klichka` и методом `myaukat()`, возвращающим строку вида «Барсик: Мяу!».

```python
class Kot:
    def __init__(self, klichka):
        self.klichka = klichka

    def myaukat(self):
        return f"{self.klichka}: Мяу!"

kot = Kot("Барсик")
print(kot.myaukat())
print(Kot.myaukat(kot))
```
