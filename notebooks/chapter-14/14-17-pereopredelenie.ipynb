# 14-17 · Переопределение методов

Практика к разделу [«Переопределение методов»](../../site/chapters/glava-14/14-17-pereopredelenie-metodov.html).

## Цель

Расширить метод родителя через super().method(), а не заменить его целиком.

## Рабочий пример

```python
class Zhivotnoe:
    def zvuk(self):
        return "Животное подаёт голос: "


class Sobaka(Zhivotnoe):
    def zvuk(self):
        return super().zvuk() + "Гав!"

print(Sobaka().zvuk())
```

## Задание ★★ Самостоятельная задача

Создайте класс `Koshka(Zhivotnoe)`, переопределяющий `zvuk()` через `super().zvuk()` и добавляющий «Мяу!».

```python
class Koshka(Zhivotnoe):
    def zvuk(self):
        return super().zvuk() + "Мяу!"

print(Koshka().zvuk())
```
