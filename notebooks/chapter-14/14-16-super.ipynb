# 14-16 · super()

Практика к разделу [«super() и порядок разрешения методов»](../../site/chapters/glava-14/14-16-super.html).

## Цель

Дополнить __init__ родителя, а не заменить его целиком.

## Рабочий пример

```python
class Zhivotnoe:
    def __init__(self, klichka):
        self.klichka = klichka


class Sobaka(Zhivotnoe):
    def __init__(self, klichka, poroda):
        super().__init__(klichka)
        self.poroda = poroda

rex = Sobaka("Рекс", "Дворняга")
print(rex.klichka, rex.poroda)
```

## Задание ★★ Самостоятельная задача

Создайте класс `Koshka(Zhivotnoe)` с __init__(self, klichka, okras), вызывающим `super().__init__(klichka)`, затем устанавливающим `self.okras`.

```python
class Koshka(Zhivotnoe):
    def __init__(self, klichka, okras):
        super().__init__(klichka)
        self.okras = okras

murka = Koshka("Мурка", "рыжий")
print(murka.klichka, murka.okras)
```
