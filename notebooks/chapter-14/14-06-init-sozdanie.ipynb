# 14-06 · __init__ и создание объекта

Практика к разделу [«__init__ и создание объекта»](../../site/chapters/glava-14/14-06-init-i-sozdanie-obekta.html).

## Цель

Значения по умолчанию в __init__ и независимость объектов.

## Рабочий пример

```python
class Sobaka:
    def __init__(self, klichka, vozrast=1):
        self.klichka = klichka
        self.vozrast = vozrast

shchenok = Sobaka("Бим")
rex = Sobaka("Рекс", vozrast=3)
print(shchenok.vozrast, rex.vozrast)
```

## Задание ★ Базовая практика

Создайте класс `Student` с параметрами `imya` и `kurs=1` (по умолчанию). Создайте `s1` без указания курса и `s2` с курсом 3.

```python
class Student:
    def __init__(self, imya, kurs=1):
        self.imya = imya
        self.kurs = kurs

s1 = Student("Аня")
s2 = Student("Боря", 3)
print(s1.kurs, s2.kurs)
```
