Практика к разделу [«Duck typing»](../../site/chapters/glava-14/14-19-duck-typing.html).

## Цель

Полиморфизм без общего родителя — важно только наличие метода.

## Рабочий пример

```python
class Chelovek:
    def __init__(self, imya):
        self.imya = imya

    def predstavitsya(self):
        return f"Привет, я {self.imya}!"


class Robot:
    def __init__(self, nomer):
        self.nomer = nomer

    def predstavitsya(self):
        return f"БИП. Я робот номер {self.nomer}."


def poznakomit(obj):
    return obj.predstavitsya()

print(poznakomit(Chelovek("Аня")))
print(poznakomit(Robot(7)))
```

## Задание ★★ Самостоятельная задача

`Chelovek` и `Robot` не имеют общего родителя (кроме `object`). Проверьте, что `poznakomit()` всё равно работает с обоими, и сохраните оба результата.

```python
rezultat_chelovek = poznakomit(Chelovek("Боря"))
rezultat_robot = poznakomit(Robot(42))
print(rezultat_chelovek)
print(rezultat_robot)
```
