Практика к разделу [«Управляем объектами. Объекты выполняют действия»](../../site/chapters/glava-14/14-03-upravlyaem-obektami.html).

## Цель

Изменять атрибуты и определять методы.

## Рабочий пример

```python
class Sobaka:
    def __init__(self, klichka, vozrast):
        self.klichka = klichka
        self.vozrast = vozrast

    def layat(self):
        print(f"{self.klichka} говорит: Гав-гав!")

    def prazdnovat_den_rozhdeniya(self):
        self.vozrast += 1
        print(f"Теперь {self.klichka} исполнилось {self.vozrast}!")

rex = Sobaka("Рекс", 3)
rex.layat()
rex.prazdnovat_den_rozhdeniya()
```

## Задание ★ Базовая практика

Добавьте классу `Kniga` метод `opisanie()`, возвращающий строку вида «Название — Автор».

```python
class Kniga:
    def __init__(self, nazvanie, avtor):
        self.nazvanie = nazvanie
        self.avtor = avtor

    def opisanie(self):
        return f"{self.nazvanie} — {self.avtor}"

k = Kniga("Война и мир", "Толстой")
print(k.opisanie())
```

## Дополнительная задача ★★★

Добавьте класс `Biblioteka`, который хранит список книг и умеет добавлять новую книгу методом `dobavit_knigu()`.

```python
class Biblioteka:
    def __init__(self):
        self.knigi = []

    def dobavit_knigu(self, kniga):
        self.knigi.append(kniga)

    def pokazat_vse(self):
        for k in self.knigi:
            print(k.opisanie())

biblioteka = Biblioteka()
biblioteka.dobavit_knigu(Kniga("Война и мир", "Толстой"))
biblioteka.dobavit_knigu(Kniga("Мастер и Маргарита", "Булгаков"))
biblioteka.pokazat_vse()
```
