Практика к разделу [«Ловушка общего изменяемого атрибута»](../../site/chapters/glava-14/14-08-lovushka-obshchego-atributa.html).

## Цель

Правильно объявить изменяемый атрибут — внутри __init__, а не в теле класса.

## Рабочий пример

```python
class Korzina:
    def __init__(self):
        self.tovary = []

    def dobavit(self, tovar):
        self.tovary.append(tovar)

k1 = Korzina()
k2 = Korzina()
k1.dobavit("Хлеб")
print(k1.tovary, k2.tovary)
```

## Задание ★ Базовая практика

Создайте класс `Zametki` с атрибутом экземпляра `spisok = []` (созданным в `__init__`, не в теле класса!) и методом `dobavit_zametku(tekst)`. Создайте `z1` и `z2`, добавьте разные записи в каждую и убедитесь, что они не смешиваются.

```python
class Zametki:
    def __init__(self):
        self.spisok = []

    def dobavit_zametku(self, tekst):
        self.spisok.append(tekst)

z1 = Zametki()
z2 = Zametki()
z1.dobavit_zametku("Купить хлеб")
z2.dobavit_zametku("Позвонить маме")
print(z1.spisok)
print(z2.spisok)
```
