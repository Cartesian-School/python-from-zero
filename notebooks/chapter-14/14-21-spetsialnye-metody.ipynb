## Цель

__str__, __eq__ и __hash__ — как объект подключается к print(), == и set().

## Рабочий пример

```python
class Dengi:
    def __init__(self, summa):
        self.summa = summa

    def __str__(self):
        return f"{self.summa} ₽"

    def __eq__(self, other):
        return self.summa == other.summa

    def __hash__(self):
        return hash(self.summa)

a = Dengi(100)
b = Dengi(100)
print(str(a), a == b)
print(len({a, b}))
```
