# 22-18 · Выбор фреймворка по задаче

Практика к разделу [«Сравнение Flask, Django и FastAPI»](../../site/chapters/glava-22/22-18-flask-django-fastapi.html).

## Цель

Связать описание задачи с наиболее подходящим стартовым выбором фреймворка — на основе таблицы сравнения с сайта, а не личных предпочтений.

## Рабочий пример

```python
vybor = {
    "sajt_s_admin_panelyu": "Django",
    "api_s_avtomaticheskoj_dokumentaciej": "FastAPI",
    "prostoj_sajt_bez_gotovoj_adminki": "Flask",
}

for zadacha, frejmvork in vybor.items():
    print(zadacha, "->", frejmvork)
```

## Проверка результата

```python
assert vybor["sajt_s_admin_panelyu"] == "Django"
assert vybor["api_s_avtomaticheskoj_dokumentaciej"] == "FastAPI"
assert vybor["prostoj_sajt_bez_gotovoj_adminki"] == "Flask"
print("Верно: выбор соответствует таблице сравнения на сайте.")
```

## Задание ★★ Самостоятельная задача
