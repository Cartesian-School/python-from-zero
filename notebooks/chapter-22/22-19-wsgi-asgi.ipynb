# 22-19 · WSGI или ASGI?

Практика к разделу [«WSGI и ASGI: связь приложения с сервером»](../../site/chapters/glava-22/22-19-wsgi-asgi.html).

## Цель

Отнести фреймворк или сервер приложений к WSGI или ASGI по таблице сравнения с сайта.

## Рабочий пример

```python
protokol = {
    "Flask": "WSGI",
    "Starlette": "ASGI",
    "FastAPI": "ASGI",
    "Gunicorn": "WSGI",
    "Uvicorn": "ASGI",
}

for imya, tip in protokol.items():
    print(imya, "->", tip)
```

## Проверка результата

```python
assert protokol["Flask"] == "WSGI"
assert protokol["FastAPI"] == "ASGI"
assert protokol["Uvicorn"] == "ASGI"
assert protokol["Gunicorn"] == "WSGI"
print("Верно: сопоставление совпадает с разделом 22.19 сайта.")
```
