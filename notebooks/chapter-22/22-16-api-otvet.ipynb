Практика к разделу [«API: когда сервер возвращает данные вместо HTML»](../../site/chapters/glava-22/22-16-api.html).

## Цель

Написать чистую функцию, которая превращает внутреннее представление задачи в словарь, готовый для API-ответа — без Flask, чистой логикой.

## Рабочий пример

```python
def zadacha_dlya_api(id_, title, done):
    return {"id": id_, "title": title, "done": bool(done)}


zapis_iz_bazy = (1, "Купить хлеб", 0)   # так вернёт sqlite3.Row/tuple
otvet = zadacha_dlya_api(*zapis_iz_bazy)
print(otvet)
```

## Проверка результата

```python
assert otvet == {"id": 1, "title": "Купить хлеб", "done": False}
assert otvet["done"] is False   # bool, а не 0 — так короче JSON-ответ
print("Верно: числовой флаг из базы превращён в настоящий bool для API-ответа.")
```

## Задание ★★ Самостоятельная задача

Напишите функцию `spisok_dlya_api(zapisi)`, которая применяет `zadacha_dlya_api()` к списку кортежей и возвращает список словарей.

```python
def spisok_dlya_api(zapisi):
    return [zadacha_dlya_api(*z) for z in zapisi]


rezultat = spisok_dlya_api([(1, "A", 1), (2, "B", 0)])
assert rezultat == [{"id": 1, "title": "A", "done": True}, {"id": 2, "title": "B", "done": False}]
print("Верно:", rezultat)
```
