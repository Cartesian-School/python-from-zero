Практика к разделу [«Столкновения и хитбоксы»](../../site/chapters/glava-20/20-21-stolknoveniya.html). AABB-проверка пересечения — чистая функция, как и в тексте раздела.

## Цель

Реализовать и проверить проверку пересечения двух прямоугольников.

## Рабочий пример

```python
def pryamougolniki_peresekayutsya(a, b):
    ax1, ay1, aw, ah = a
    bx1, by1, bw, bh = b
    ax2, ay2 = ax1 + aw, ay1 + ah
    bx2, by2 = bx1 + bw, by1 + bh
    return ax1 < bx2 and ax2 > bx1 and ay1 < by2 and ay2 > by1

igrok = (100, 100, 40, 40)
vrag_blizko = (120, 120, 40, 40)
vrag_daleko = (400, 400, 40, 40)

print("Столкновение с близким врагом:", pryamougolniki_peresekayutsya(igrok, vrag_blizko))
print("Столкновение с далёким врагом:", pryamougolniki_peresekayutsya(igrok, vrag_daleko))
```

## Проверка результата

```python
assert pryamougolniki_peresekayutsya(igrok, vrag_blizko) is True
assert pryamougolniki_peresekayutsya(igrok, vrag_daleko) is False
# соприкосновение ровно краем — НЕ считается пересечением
assert pryamougolniki_peresekayutsya((0, 0, 40, 40), (40, 0, 40, 40)) is False
print("Верно: пересечение находится правильно, включая пограничный случай соприкосновения краями.")
```

## Задание ★★ Самостоятельная задача

Уменьшите хитбокс игрока на 10 пикселей с каждой стороны (то есть на 20 по ширине и высоте суммарно — как `rect.inflate(-20, -20)` из раздела 20.21) и посмотрите, на что это на самом деле влияет.

Проверьте на двух врагах: на `vrag_blizko`, который налезает на игрока глубоко, и на «краевом» враге `vrag_kraem = (135, 135, 40, 40)`, который задевает только угол картинки. Убедитесь, что уменьшение хитбокса меняет ответ именно для краевого врага, а настоящее, глубокое столкновение остаётся столкновением.

```python
def szhat(rect, na):
    x, y, w, h = rect
    return (x + na, y + na, w - 2 * na, h - 2 * na)

igrok_szhatyj = szhat(igrok, 10)
assert igrok_szhatyj == (110, 110, 20, 20)

vrag_kraem = (135, 135, 40, 40)   # задевает только угол картинки игрока

blizko_do = pryamougolniki_peresekayutsya(igrok, vrag_blizko)
blizko_posle = pryamougolniki_peresekayutsya(igrok_szhatyj, vrag_blizko)
kraem_do = pryamougolniki_peresekayutsya(igrok, vrag_kraem)
kraem_posle = pryamougolniki_peresekayutsya(igrok_szhatyj, vrag_kraem)

print("Глубокое перекрытие: до", blizko_do, "-> после", blizko_posle)
print("Касание углом:       до", kraem_do, "-> после", kraem_posle)

# Настоящее столкновение остаётся столкновением...
assert blizko_do is True
assert blizko_posle is True
# ...а "нечестное" попадание по прозрачному углу картинки исчезает:
assert kraem_do is True
assert kraem_posle is False
print("Верно: уменьшение хитбокса убирает попадания по краю, не трогая настоящие столкновения.")
```
