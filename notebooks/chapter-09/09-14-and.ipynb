# 09-14 · and

Практика к разделу [«and — все условия сразу»](../../site/chapters/glava-09/09-14-and.html).

## Цель

Освоить and — все условия должны быть True.

## Рабочий пример

```python
age = 20
has_ticket = True
print(age >= 18 and has_ticket)
```

## Задание ★ Базовая практика

Постройте can_enter — True, только если age >= 18 И has_ticket одновременно.

```python
age = 20
has_ticket = True
can_enter = age >= 18 and has_ticket
print(can_enter)
```
