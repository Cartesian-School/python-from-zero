## Цель

Увидеть off-by-one ошибку на границах диапазона.

## Рабочий пример

```python
number = 1
print(1 <= number <= 10)
print(1 < number < 10)
```

## Задание ★ Базовая практика

Сравните правильную (включительно) и ошибочную (строгую) проверку границы для number = 1.

```python
number = 1
in_range_correct = 1 <= number <= 10
in_range_buggy = 1 < number < 10
print(in_range_correct, in_range_buggy)
```
