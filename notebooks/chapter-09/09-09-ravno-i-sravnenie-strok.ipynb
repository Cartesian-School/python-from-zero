Практика к разделу [«Чем отличаются = и ==, и как сравнивать строки»](../../site/chapters/glava-09/09-09-ravno-i-sravnenie-strok.html).

## Цель

Не путать = и ==, освоить != и сравнение строк.

## Рабочий пример

```python
name = "Anna"
print(name == "Anna")
```

## Типичная ошибка

Одиночный `=` внутри условия — синтаксическая ошибка.

```python
age = 20
if age = 20:
    print("Совпадает")
```

## Исправление

```python
age = 20
if age == 20:
    print("Совпадает")
```

## Задание ★ Базовая практика

Проверьте возраст через `==`, сравните два слова по алфавиту и сравните строки без учёта регистра через `.lower()`.

```python
age = 20
is_adult = age == 20
apple_before_banana = "apple" < "banana"
same_case_insensitive = "Python".lower() == "python".lower()
print(is_adult, apple_before_banana, same_case_insensitive)
```
