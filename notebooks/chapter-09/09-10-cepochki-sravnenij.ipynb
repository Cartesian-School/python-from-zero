## Рабочий пример

```python
age = 30
print(18 <= age <= 65)
```

## Задание ★ Базовая практика

Проверьте, входит ли age=30 в диапазон 18..65, и входит ли score=105 в диапазон 0..100.

```python
age = 30
in_range = 18 <= age <= 65

score = 105
valid_score = 0 <= score <= 100
print(in_range, valid_score)
```
