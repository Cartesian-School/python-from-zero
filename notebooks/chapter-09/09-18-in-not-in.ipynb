# 09-18 · in / not in

Практика к разделу [«in / not in как условие»](../../site/chapters/glava-09/09-18-in-not-in.html).

## Цель

Использовать in/not in прямо в условиях.

## Рабочий пример

```python
text = "python"
print("py" in text)
```

## Задание ★ Базовая практика

Проверьте вхождение подстроки и проверьте ответ пользователя на принадлежность группе допустимых значений.

```python
text = "python"
has_py = "py" in text

answer = "да"
is_yes = answer in ("да", "yes", "y")
print(has_py, is_yes)
```
