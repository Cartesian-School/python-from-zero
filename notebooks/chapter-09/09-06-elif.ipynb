## Цель

Освоить цепочки elif и вложенные условия.

## Рабочий пример

```python
ocenka = 87
if ocenka >= 90:
    bukva = "A"
elif ocenka >= 80:
    bukva = "B"
elif ocenka >= 70:
    bukva = "C"
else:
    bukva = "D"

print(f"Оценка: {bukva}")
```

## Эксперимент 1

Проверьте несколько разных оценок через цикл (забегая немного вперёд, глава 10).

```python
for ocenka in [95, 82, 71, 40]:
    if ocenka >= 90:
        bukva = "A"
    elif ocenka >= 80:
        bukva = "B"
    elif ocenka >= 70:
        bukva = "C"
    else:
        bukva = "D"
    print(ocenka, "->", bukva)
```

## Эксперимент 2 — вложенные условия

```python
age = 25
has_license = True

if age >= 18:
    if has_license:
        print("Можно водить машину.")
    else:
        print("Сначала получите права.")
else:
    print("Ещё рано водить машину.")
```

## Задание ★★ Самостоятельная задача

Добавьте третий уровень вложенности: проверку наличия топлива (`has_fuel`).

```python
age = 25
has_license = True
has_fuel = False

if age >= 18:
    if has_license:
        if has_fuel:
            print("Можно ехать!")
        else:
            print("Права есть, но бак пуст.")
    else:
        print("Сначала получите права.")
else:
    print("Ещё рано водить машину.")
```
