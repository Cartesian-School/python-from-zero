## Цель

Построить цепочку elif и убедиться, что порядок условий важен.

## Рабочий пример

```python
temperature = -5
if temperature < 0:
    result = "мороз"
elif temperature < 15:
    result = "прохладно"
elif temperature < 25:
    result = "комфортно"
else:
    result = "жарко"
print(result)
```

## Задание ★ Базовая практика

Прогоните ту же лестницу для temperature = 10 и предскажите результат заранее.

```python
temperature = 10
if temperature < 0:
    result = "мороз"
elif temperature < 15:
    result = "прохладно"
elif temperature < 25:
    result = "комфортно"
else:
    result = "жарко"
print(result)
```
