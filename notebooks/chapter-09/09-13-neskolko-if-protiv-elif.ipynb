# 09-13 · Несколько if ≠ if/elif/else

Практика к разделу [«Несколько if ≠ if/elif/else»](../../site/chapters/glava-09/09-13-neskolko-if-protiv-elif.html).

## Цель

Увидеть разницу между независимыми if и цепочкой elif.

## Рабочий пример

```python
temperature = 30
if temperature > 20:
    print("тепло")
if temperature > 25:
    print("жарко")
```

## Задание ★ Базовая практика

Проверьте, сработают ли ОБА независимых условия одновременно для temperature = 30.

```python
temperature = 30
warm_triggered = temperature > 20
hot_triggered = temperature > 25
both_triggered = warm_triggered and hot_triggered
print(warm_triggered, hot_triggered, both_triggered)
```
