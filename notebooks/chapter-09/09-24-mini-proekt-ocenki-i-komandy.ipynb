## Цель

Собрать классификатор баллов и текстовый интерпретатор команд.

## Рабочий пример

```python
score = 82
if score >= 90:
    level = "отлично"
elif score >= 75:
    level = "хорошо"
elif score >= 50:
    level = "удовлетворительно"
else:
    level = "неудовлетворительно"
print(level)
```

## Задание ★ Базовая практика

Повторите классификацию для score = 82, и постройте интерпретатор команд для command = "START" (с нормализацией регистра).

```python
score = 82
if score >= 90:
    level = "отлично"
elif score >= 75:
    level = "хорошо"
elif score >= 50:
    level = "удовлетворительно"
else:
    level = "неудовлетворительно"

command = "START".strip().lower()
if command == "start":
    action = "Запуск..."
elif command == "stop":
    action = "Остановка..."
else:
    action = "Неизвестная команда"
print(level, action)
```
