Практика к разделу [«Short-circuit: Python иногда ленится»](../../site/chapters/glava-09/09-17-short-circuit.html).

## Цель

Увидеть, как short-circuit защищает от IndexError.

## Рабочий пример

```python
name = "Anna"
print(name and name[0] == "A")
```

## Типичная ошибка

Без short-circuit порядок операндов может вызвать IndexError.

```python
name = ""
print(name[0] == "A" and name)
```

## Исправление

```python
name = ""
safe_check = name and name[0] == "A"
print(bool(safe_check))
```
