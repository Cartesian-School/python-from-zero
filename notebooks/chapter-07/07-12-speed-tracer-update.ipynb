## Цель

Освоить пакетное рисование через tracer(0) + update().

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

20 окружностей, нарисованных мгновенно.

```python
artist.reset()
screen.tracer(0)
for i in range(20):
    artist.circle(60)
    artist.right(18)
screen.update()
```

## Задание ★ Базовая практика

Используя tracer(0) и update(), нарисуйте 24 окружности радиусом 50, каждая повёрнута на 15°.

```python
artist.reset()
screen.tracer(0)
for i in range(24):
    artist.circle(50)
    artist.right(15)
screen.update()
```

## Завершение (выполнить один раз, в самом конце)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
