## Цель

Освоить align и font в write().

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

Три выравнивания у одной точки.

```python
artist.reset()
artist.hideturtle()
artist.penup()
artist.dot(6)
artist.write("center", align="center", font=("Arial", 14, "normal"))
```

## Задание ★ Базовая практика

Выведите текст "Cartesian" крупным (24) жирным шрифтом Arial, выровненным по правому краю (align="right").

```python
artist.reset()
artist.hideturtle()
artist.penup()
artist.write("Cartesian", align="right", font=("Arial", 24, "bold"))
```

## Завершение (выполнить один раз, в самом конце)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
