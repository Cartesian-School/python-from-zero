Практика к разделу [«circle(steps=...) — от круга к многоугольнику»](../../site/chapters/glava-07/07-14-circle-steps.html).

## Цель

Превращать окружность в многоугольник через steps.

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

Пятиугольник через circle(steps=5).

```python
artist.reset()
artist.circle(80, steps=5)
```

## Задание ★ Базовая практика

Нарисуйте восьмиугольник радиусом 70, используя circle() с steps.

```python
artist.reset()
artist.circle(70, steps=8)
```
