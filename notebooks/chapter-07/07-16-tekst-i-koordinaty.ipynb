## Цель

Собрать goto(), dot() и write() в подписанный график.

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

Одна подписанная точка на оси.

```python
artist.reset()
artist.hideturtle()
artist.penup()
artist.goto(50, 0)
artist.dot(6, "#5B24F9")
artist.goto(50, -20)
artist.write("50", align="center", font=("Arial", 11, "normal"))
```

## Задание ★ Базовая практика

Поставьте три подписанные точки на оси X: -50, 0 и 50 — с числовыми подписями под каждой.

```python
artist.reset()
artist.hideturtle()
artist.penup()
for value in [-50, 0, 50]:
    artist.goto(value, 0)
    artist.dot(6, "#5B24F9")
    artist.goto(value, -20)
    artist.write(str(value), align="center", font=("Arial", 11, "normal"))
```

## Завершение (выполнить один раз, в самом конце)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
