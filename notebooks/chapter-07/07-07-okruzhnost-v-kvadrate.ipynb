## Цель

Скомбинировать фигуры и попрактиковать направление рисования окружности.

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Рабочий пример

```python
artist.reset()
razmer = 150

for _ in range(4):
    artist.forward(razmer)
    artist.right(90)

artist.penup()
artist.goto(razmer / 2, -razmer / 2)
artist.setheading(0)
artist.pendown()
artist.circle(razmer / 2)
print("Квадрат с вписанной окружностью готов.")
```

## Эксперимент 1 — направление окружности

```python
artist.reset()
artist.circle(60)    # против часовой стрелки
print("Курс после обычной окружности:", artist.heading())

artist.reset()
artist.circle(-60)   # по часовой стрелке
print("Курс после окружности с отрицательным радиусом:", artist.heading())
```

## Задание ★★ Самостоятельная задача

Впишите окружность в шестиугольник со стороной 80 (подсказка: радиус вписанной окружности шестиугольника — `сторона * 0.866`).

```python
import math

artist.reset()
storona = 80
for _ in range(6):
    artist.forward(storona)
    artist.right(60)

radius = storona * math.sqrt(3) / 2
artist.penup()
artist.goto(0, -radius)
artist.setheading(0)
artist.pendown()
artist.circle(radius)
print("Окружность вписана в шестиугольник, радиус:", round(radius, 1))
```
