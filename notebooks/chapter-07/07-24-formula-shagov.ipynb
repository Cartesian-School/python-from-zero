# 07-24 · Формула для circle(steps=...)

Практика к разделу [«circle(steps=...) — от круга к многоугольнику»](../../site/chapters/glava-07/07-14-circle-steps.html) — чистая математика, без окна Turtle.

## Цель

Считать угол поворота многоугольника, приближающего окружность, для разного числа steps.

## Рабочий пример

```python
steps = 5
povorot = 360 / steps
print(povorot)
```

## Задание ★ Базовая практика

Посчитайте угол поворота для circle(80, steps=3), circle(80, steps=4) и circle(80, steps=8). Выведите все три числа через пробел одним print().

```python
povorot_3 = 360 / 3
povorot_4 = 360 / 4
povorot_8 = 360 / 8
print(povorot_3, povorot_4, povorot_8)
```
