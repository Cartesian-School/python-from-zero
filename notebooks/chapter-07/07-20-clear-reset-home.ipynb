## Цель

Различать эффект clear(), reset() и home() на практике.

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

reset() стирает рисунок и возвращает домой.

```python
artist.forward(100)
artist.left(90)
artist.forward(60)
print(artist.position())
artist.reset()
print(artist.position())
```

## Задание ★ Базовая практика

Нарисуйте квадрат, переместитесь (penup) в точку (80, 80), затем вызовите home() и выведите position() — рисунок должен остаться на месте.

```python
artist.reset()
for _ in range(4):
    artist.forward(60)
    artist.right(90)
artist.penup()
artist.goto(80, 80)
artist.home()
print(artist.position())
```

## Завершение (выполнить один раз, в самом конце)

```python
screen.bye()
print("Окно Turtle закрыто.")
```
