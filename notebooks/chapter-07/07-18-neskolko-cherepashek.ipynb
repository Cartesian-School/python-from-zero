## Цель

Создать несколько независимых объектов Turtle.

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

Две черепашки, разные цвета.

```python
artist.reset()
friend = turtle.Turtle()
friend.color("#DB2777")
friend.penup(); friend.goto(0, -80); friend.pendown()
artist.forward(100)
friend.forward(100)
```

## Задание ★ Базовая практика

Создайте третью черепашку с именем `third`, зелёного цвета, и переместите её на forward(80).

```python
third = turtle.Turtle()
third.color("green")
third.penup(); third.goto(0, 80); third.pendown()
third.forward(80)
```
