## Цель

Отличать относительный путь от абсолютного и уметь получить абсолютный вариант через .resolve().

```python
from pathlib import Path

Path("data").mkdir(exist_ok=True)
Path("data/a.txt").write_text("привет", encoding="utf-8")
```

## Рабочий пример

```python
from pathlib import Path

rel = Path("data/a.txt")
abs_path = rel.resolve()
print(rel.is_absolute())
print(abs_path.is_absolute())
```

## Задание ★ Базовая практика

Проверьте, что имя файла не потерялось при переходе к абсолютному пути.

```python
from pathlib import Path

rel = Path("data/a.txt")
abs_path = rel.resolve()
imya_sovpadaet = abs_path.name == rel.name
print(imya_sovpadaet)
```
