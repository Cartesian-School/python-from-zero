Практика к разделу [«pathlib.Path: пути как объекты»](../../site/chapters/glava-15/15-08-pochemu-pathlib.html).

## Цель

Собирать путь через / вместо конкатенации строк.

## Рабочий пример

```python
from pathlib import Path

file_path = Path("data") / "players" / "anna.json"
print(file_path)
print(file_path.name)
print(file_path.parent.as_posix())
```

## Задание ★ Базовая практика

Постройте путь на один уровень глубже и проверьте, что число частей пути увеличилось.

```python
from pathlib import Path

file_path = Path("data") / "players" / "anna.json"
deeper_path = Path("data") / "players" / "guilds" / "anna.json"
parts_increased = len(deeper_path.parts) > len(file_path.parts)
print(parts_increased)
```
