# 15-12 · Жизненный цикл файла и with

Практика к разделу [«Жизненный цикл файла и with»](../../site/chapters/glava-15/15-12-zhiznenny-cikl-i-with.html).

## Рабочий пример

```python
with open("cikl.txt", "w", encoding="utf-8") as f:
    f.write("data")

with open("cikl.txt", "r", encoding="utf-8") as file:
    closed_inside = file.closed
closed_outside = file.closed
print(closed_inside, closed_outside)
```

## Задание ★★ Самостоятельная задача

Откройте файл БЕЗ with, прочитайте его и закройте вручную — убедитесь, что file.closed становится True только после явного close().

```python
file = open("cikl.txt", "r", encoding="utf-8")
closed_before_manual_close = file.closed
file.read()
file.close()
closed_after_manual_close = file.closed
print(closed_before_manual_close, closed_after_manual_close)
```
