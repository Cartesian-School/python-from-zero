# 15-18 · pathlib: read_text, write_text, read_bytes, write_bytes

Практика к разделу [«pathlib: удобные методы»](../../site/chapters/glava-15/15-18-pathlib-udobnye-metody.html).

## Рабочий пример

```python
from pathlib import Path

p = Path("nastroyki3.txt")
p.write_text("привет", encoding="utf-8")
text_back = p.read_text(encoding="utf-8")
print(text_back)
```

## Задание ★ Базовая практика — бинарные данные

```python
from pathlib import Path

pb = Path("signal3.bin")
pb.write_bytes(bytes([1, 2, 3]))
bytes_back = pb.read_bytes()
print(bytes_back)
```
