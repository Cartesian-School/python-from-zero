# 15-17 · Бинарные файлы и переносы строк

Практика к разделу [«Бинарные файлы и переносы строк»](../../site/chapters/glava-15/15-17-binarnye-fajly-i-perevody-strok.html).

## Рабочий пример

```python
data = bytes([10, 20, 30])
with open("signal2.bin", "wb") as f:
    f.write(data)

with open("signal2.bin", "rb") as f:
    read_data = f.read()
print(read_data)
```

## Задание ★ Базовая практика — символы против байтов

```python
text2 = "Питон🐍"
chars_len2 = len(text2)
bytes_len2 = len(text2.encode("utf-8"))
print(chars_len2, bytes_len2)
```
