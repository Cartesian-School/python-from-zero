# 15-07 · Текущая рабочая директория (CWD)

Локальная практика к разделу [«Текущая рабочая директория (CWD)»](../../site/chapters/glava-15/15-07-tekushaya-rabochaya-directoriya.html).

**Этот ноутбук выполняется локально** — на вашем компьютере, а не в браузере: смысл упражнения именно в том, чтобы увидеть разницу между CWD и папкой, где лежит исходный файл, на настоящей файловой системе.

## Шаг 1 — где я нахожусь?

```python
from pathlib import Path

print("CWD:", Path.cwd())
```

## Шаг 2 — папка самого скрипта

```python
from pathlib import Path

if "__file__" in globals():
    print("Папка файла:", Path(__file__).resolve().parent)
else:
    print("__file__ не определён в этой среде выполнения (например, в интерактивном ноутбуке) — "
          "это нормально и ожидаемо, см. раздел 15.7.")
```
