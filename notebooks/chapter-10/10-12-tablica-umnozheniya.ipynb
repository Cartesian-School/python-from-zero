## Цель

Построить таблицу умножения вложенными циклами, сохранив результат в список списков.

## Задание ★ Базовая практика

```python
tablica = []
for a in range(1, 6):
    stroka = []
    for b in range(1, 6):
        stroka.append(a * b)
    tablica.append(stroka)
print(tablica)
```
