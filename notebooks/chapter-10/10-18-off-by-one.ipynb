## Цель

Потренироваться точно определять границы range() под конкретную задачу.

## Задание ★ Базовая практика

Соберите числа от 5 до 15 включительно в `diapazon`.

```python
diapazon = []
for n in range(5, 16):
    diapazon.append(n)
print(diapazon)
```

## Задание ★★ — убывающий диапазон

Соберите чётные числа от 20 до 2 включительно, по убыванию, в `ubyvanie`.

```python
ubyvanie = []
for n in range(20, 1, -2):
    ubyvanie.append(n)
print(ubyvanie)
```
