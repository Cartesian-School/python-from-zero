## Цель

Переспрашивать пользователя в цикле, пока ввод не станет корректным.

## Про input() в этом ноутбуке

Этот ноутбук выполняется автоматически, без живого человека за клавиатурой — поэтому здесь `input()` временно подменён на заранее заготовленные ответы.

```python
_answers = iter(['abc', '', '15'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Задание ★ Базовая практика

```python
vozrast = input("Введите возраст: ")
while not vozrast.isdigit():
    print("Нужно ввести число!")
    vozrast = input("Введите возраст: ")

vozrast = int(vozrast)
print("Спасибо! Вам", vozrast, "лет")
```
