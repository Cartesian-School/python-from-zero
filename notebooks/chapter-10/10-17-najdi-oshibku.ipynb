## Сломанный код (НЕ запускать как есть)

Вот код с ошибкой №4 из урока «Отладка циклов» — счётчик сбрасывается прямо внутри цикла:

```python
chisla = [3, 8, 12, 5, 16, 9, 20]
for n in chisla:
    chetnye_count = 0   # ошибка: обнуляется на каждой итерации!
    if n % 2 == 0:
        chetnye_count += 1
```

В ячейке ниже — задание: напишите **исправленную** версию, которая действительно считает общее количество чётных чисел.

## Задание ★ Базовая практика — исправьте ошибку №4

````python
chisla = [3, 8, 12, 5, 16, 9, 20]
chetnye_count = 0
for n in chisla:
    if n % 2 == 0:
        chetnye_count += 1
print(chetnye_count)
````

## Задание ★★ — исправьте ошибку №2 (off-by-one)

Нужно собрать числа от 1 до 10 включительно в список `chisla_1_do_10`.

````python
chisla_1_do_10 = []
for n in range(1, 11):
    chisla_1_do_10.append(n)
print(chisla_1_do_10)
````
