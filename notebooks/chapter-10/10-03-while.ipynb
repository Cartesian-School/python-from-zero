## Цель

Перебирать строки циклом for и освоить цикл while.

## Рабочий пример

```python
for letter in "Python":
    print(letter)
```

## Эксперимент 1 — while

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

## Типичная ошибка — забыли увеличить счётчик

Здесь используем ограничитель предохранителем (`safety`), чтобы демонстрационная ячейка не зависла по-настоящему в автоматическом выполнении.

```python
count = 0
safety = 0
while count < 5:
    print(count)
    safety += 1
    if safety > 1000:
        print("Похоже на бесконечный цикл — останавливаем принудительно.")
        break
```

## Задание ★ Базовая практика

Посчитайте гласные буквы в слове с помощью for по строке.

```python
word = "программирование"
glasnye = "аеёиоуыэюя"
count = 0
for letter in word:
    if letter in glasnye:
        count += 1
print(f"Гласных букв: {count}")
```
