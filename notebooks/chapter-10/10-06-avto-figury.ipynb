## Цель

Переписать фигуры из главы 6 с циклами.

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Рабочий пример — квадрат

```python
artist.reset()
for _ in range(4):
    artist.forward(100)
    artist.right(90)
print("Квадрат готов.")
```

## Эксперимент 1 — любая фигура

```python
artist.reset()
storony = 8
dlina = 60
ugol = 360 / storony

for _ in range(storony):
    artist.forward(dlina)
    artist.right(ugol)
print(f"{storony}-угольник готов.")
```

## Задание ★ Базовая практика

Нарисуйте фигуры с 3, 5 и 10 сторонами подряд, используя одну и ту же функцию-подобную схему.

```python
for storony in [3, 5, 10]:
    artist.reset()
    dlina = 50
    ugol = 360 / storony
    for _ in range(storony):
        artist.forward(dlina)
        artist.right(ugol)
    print(f"{storony}-угольник нарисован, угол поворота {ugol}")
```
