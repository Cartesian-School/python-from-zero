## Цель

Ограничить число попыток циклом for и использовать loop-else, чтобы сообщить правильный ответ, если игрок не угадал.

## Про input() в этом ноутбуке

Этот ноутбук выполняется автоматически, без живого человека за клавиатурой — поэтому здесь `input()` временно подменён на заранее заготовленные ответы.

```python
_answers = iter(['1', '2', '3', '4', '5'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Задание ★★ Самостоятельная задача

Здесь задуманное число фиксировано (14), а все 5 попыток заведомо неверны — чтобы проверить именно ветку `else`.

```python
zagadannoe = 14
ugadal = None

for popytka_nomer in range(5):
    popytka = int(input("Угадайте число от 1 до 20: "))
    if popytka == zagadannoe:
        print("Угадали!")
        ugadal = True
        break
    elif popytka < zagadannoe:
        print("Больше.")
    else:
        print("Меньше.")
else:
    ugadal = False

print("Угадано:", ugadal)
```
