## Цель

Собрать счётчик и накопитель вместе в небольшом инструменте анализа текста.

## Задание ★ Базовая практика

```python
tekst = "python это просто и понятно"
slova = tekst.split()

glasnye_count = 0
for letter in tekst:
    if letter in "аеёиоуыэюя":
        glasnye_count += 1

samoe_dlinnoe = ""
for slovo in slova:
    if len(slovo) > len(samoe_dlinnoe):
        samoe_dlinnoe = slovo

print(f"Слов: {len(slova)}, гласных: {glasnye_count}, самое длинное: {samoe_dlinnoe}")
```
