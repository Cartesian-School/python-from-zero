## Цель

Освоить все три сигнатуры range() и отрицательный шаг.

## Рабочий пример

```python
for n in range(0, 10, 2):
    print(n)
```

## Эксперимент 1 — отрицательный шаг

```python
for n in range(10, 0, -2):
    print(n)
```

## Задание ★ Базовая практика

Соберите в список `otschet` числа от 100 до 50 включительно, с шагом -5 (обратный отсчёт).

```python
otschet = []
for n in range(100, 45, -5):
    otschet.append(n)
print(otschet)
```
