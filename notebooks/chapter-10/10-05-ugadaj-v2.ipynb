## Цель

Переписать игру из главы 9 с неограниченным числом попыток.

## Про input() в этом ноутбуке

Этот ноутбук выполняется автоматически, без живого человека за клавиатурой — поэтому здесь `input()` временно подменён на заранее заготовленные ответы.

```python
_answers = iter(['10', '15', '11'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Рабочий пример

```python
import random

random.seed(7)
zagadannoe = random.randint(1, 20)
popytki = 0

while True:
    popytka = int(input("Угадайте число от 1 до 20: "))
    popytki += 1

    if popytka == zagadannoe:
        print(f"Поздравляем, вы угадали за {popytki} попыток(ки)!")
        break
    elif popytka < zagadannoe:
        print("Загаданное число больше.")
    else:
        print("Загаданное число меньше.")
```
