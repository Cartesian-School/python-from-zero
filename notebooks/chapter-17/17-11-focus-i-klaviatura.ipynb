# 17-11 · Focus и клавиатура

Практика к разделу [«Focus и клавиатура»](../../site/chapters/glava-17/17-11-focus-i-klaviatura.html).

## Цель

Увидеть, как фокус ввода определяет, какой виджет получает клавиатурные события.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
entry = tk.Entry(root)
entry.pack()
button = tk.Button(root, text="Кнопка")
button.pack()
root.update()

entry.focus_set()
root.update()
print("Фокус на Entry:", root.focus_get() is entry)

button.focus_set()
root.update()
print("Фокус на Button:", root.focus_get() is button)
root.destroy()
```
