Практика к разделу [«Focus и клавиатура»](../../site/chapters/glava-17/17-11-focus-i-klaviatura.html).

## Цель

Увидеть, как фокус ввода определяет, какой виджет получает клавиатурные события.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
entry = tk.Entry(root)
entry.pack()
button = tk.Button(root, text="Кнопка")
button.pack()
root.update()
root.focus_force()   # окну нужен фокус ОС, иначе фокус внутри него никому не достанется
root.update()

entry.focus_set()
root.update()
print("Фокус на Entry:", root.focus_get() is entry)

button.focus_set()
root.update()
print("Фокус на Button:", root.focus_get() is button)
root.destroy()
```

## Задание ★★ Самостоятельная задача

Привяжите `<Key>` и к root, и к entry (у entry — с обработчиком, который явно возвращает `"break"`). Установите фокус на entry и проверьте: сработает обработчик entry, а обработчик root — нет. Затем уберите `"break"` у entry и проверьте ещё раз: теперь сработают ОБА обработчика. Вывод: событие останавливает именно `"break"` где-то на пути по bindtags, а не сам факт того, что фокус находится на другом виджете.

```python
root = tk.Tk()
entry = tk.Entry(root)
entry.pack()
root.update()
root.focus_force()
entry.focus_set()
root.update()

popalo_v_root = []
root.bind("<Key>", lambda e: popalo_v_root.append(e.keysym))

# 1) у entry есть свой обработчик, который возвращает "break" — цепочка обрывается
entry.bind("<Key>", lambda e: "break")
entry.event_generate("<KeyPress-5>")
root.update()
print("С 'break':   root получил", popalo_v_root, "| в поле:", repr(entry.get()))
assert popalo_v_root == [], "'break' обязан оборвать цепочку до root"

# 2) убираем "break" — событие идёт дальше по bindtags и доходит до root
popalo_v_root.clear()
entry.delete(0, "end")
entry.unbind("<Key>")
entry.event_generate("<KeyPress-5>")
root.update()
print("Без 'break': root получил", popalo_v_root, "| в поле:", repr(entry.get()))
assert popalo_v_root == ["5"], "без 'break' событие обязано дойти и до root"
assert entry.get() == "5", "и при этом символ всё равно попадает в поле ввода"

print("\nЦепочку останавливает именно 'break', а не сам факт того,")
print("что фокус находится на другом виджете.")
root.destroy()
```
