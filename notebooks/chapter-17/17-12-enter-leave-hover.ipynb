# 17-12 · Mouse Enter/Leave и hover

Практика к разделу [«Mouse Enter/Leave и hover»](../../site/chapters/glava-17/17-12-enter-leave-hover.html).

## Цель

Построить простой hover-эффект через <Enter>/<Leave> — без единого клика.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
button = tk.Button(root, text="Наведите на меня")
button.pack()

def on_enter(_event):
    button.config(bg="#B9A0FC")

def on_leave(_event):
    button.config(bg="#FAFAFC")

button.bind("<Enter>", on_enter)
button.bind("<Leave>", on_leave)
root.update()

button.event_generate("<Enter>")
root.update()
print("После Enter:", button.cget("bg"))

button.event_generate("<Leave>")
root.update()
print("После Leave:", button.cget("bg"))
root.destroy()
```

## Задание ★★ Самостоятельная задача

Соберите три кнопки в ряд и подсветите наведённую, гарантируя, что остальные две остаются в обычном цвете.

```python
root = tk.Tk()
OBYCHNYJ = "#FAFAFC"
PODSVETKA = "#B9A0FC"

knopki = []
for i in range(3):
    b = tk.Button(root, text=f"Кнопка {i}", bg=OBYCHNYJ)
    b.pack(side="left")
    # k=b — тот же приём "заморозки" значения, что и с lambda i=indeks в разделе 17.4
    b.bind("<Enter>", lambda _event, k=b: k.config(bg=PODSVETKA))
    b.bind("<Leave>", lambda _event, k=b: k.config(bg=OBYCHNYJ))
    knopki.append(b)
root.update()

knopki[1].event_generate("<Enter>")
root.update()
cveta = [k.cget("bg") for k in knopki]
print("Цвета после наведения на кнопку 1:", cveta)
assert cveta == [OBYCHNYJ, PODSVETKA, OBYCHNYJ], "подсвечена должна быть ровно одна кнопка"

knopki[1].event_generate("<Leave>")
root.update()
print("Цвета после ухода курсора:      ", [k.cget("bg") for k in knopki])
assert all(k.cget("bg") == OBYCHNYJ for k in knopki)
print("Подсвечивается только наведённая кнопка, остальные остаются обычными.")
root.destroy()
```
