# 17-22 · Визуальный стиль X/O

Практика к разделу [«Визуальный стиль X/O»](../../site/chapters/glava-17/17-22-vizualnyj-stil.html).

## Цель

Покрасить X в фиолетовый, O — в розовый через прямой fg= у tk.Button.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

MARK_COLOR = {"X": "#5B24F9", "O": "#DB2777"}

root = tk.Tk()
btn = tk.Button(root, text="X", fg=MARK_COLOR["X"], font=("Arial", 24, "bold"))
btn.pack()
root.update()
print("Цвет X:", btn.cget("fg"))

btn.config(text="O", fg=MARK_COLOR["O"])
root.update()
print("Цвет O:", btn.cget("fg"))
root.destroy()
```

## Задание ★★ Самостоятельная задача

Соберите три кнопки: с X, с O и пустую — и проверьте, что MARK_COLOR.get(text, "#0D0230") даёт нейтральный цвет для пустой клетки.

```python
NEJTRALNYJ = "#0D0230"

root = tk.Tk()
kletki = []
for tekst in ("X", "O", ""):
    b = tk.Button(root, text=tekst, fg=MARK_COLOR.get(tekst, NEJTRALNYJ),
                  font=("Arial", 24, "bold"), width=3)
    b.pack(side="left")
    kletki.append(b)
root.update()

cveta = [b.cget("fg") for b in kletki]
print("Цвета X / O / пустой клетки:", cveta)
assert cveta == ["#5B24F9", "#DB2777", NEJTRALNYJ]
print("Для пустой клетки в MARK_COLOR ключа нет — .get() отдаёт нейтральный цвет.")
root.destroy()
```
