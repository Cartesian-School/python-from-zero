# 17-05 · Проверка победы

Практика к разделу [«Проверяем после каждого хода, победил ли игрок»](../../site/chapters/glava-17/17-05-proverka-pobedy.html).

## Цель

Проверить все восемь выигрышных линий.

## Рабочий пример

```python
def proverit_pobedu(polya_teksty):
    linii_pobedy = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6),
    ]
    for a, b, c in linii_pobedy:
        znacheniya = (polya_teksty[a], polya_teksty[b], polya_teksty[c])
        if znacheniya[0] != "" and znacheniya[0] == znacheniya[1] == znacheniya[2]:
            return znacheniya[0]
    return None

# верхняя строка выиграна X
doska = ["X", "X", "X", "", "O", "O", "", "", ""]
print(proverit_pobedu(doska))
```

## Эксперимент 1 — все восемь линий

```python
VSE_LINII = [
    ((0, 1, 2), "строка 1"), ((3, 4, 5), "строка 2"), ((6, 7, 8), "строка 3"),
    ((0, 3, 6), "столбец 1"), ((1, 4, 7), "столбец 2"), ((2, 5, 8), "столбец 3"),
    ((0, 4, 8), "диагональ 0-4-8"), ((2, 4, 6), "диагональ 2-4-6"),
]

def doska_dlya_linii(liniya, znak):
    """Собирает поле, на котором znak занял ровно эту линию."""
    doska = [""] * 9
    for i in liniya:
        doska[i] = znak
    return doska

vse_linii_ok = True
for liniya, opisanie in VSE_LINII:
    for znak in ("X", "O"):
        result = proverit_pobedu(doska_dlya_linii(liniya, znak))
        if result != znak:
            vse_linii_ok = False
        print(f"{opisanie:16} {znak}: получили {result}, ожидали {znak} -> "
              f"{'OK' if result == znak else 'ОШИБКА'}")

# контрольный случай: поле заполнено, но выигрышной линии нет
bez_pobeditelya = proverit_pobedu(["X", "O", "X", "X", "O", "O", "O", "X", "X"])
print(f"\n{'нет победителя':16}  : получили {bez_pobeditelya}, ожидали None -> "
      f"{'OK' if bez_pobeditelya is None else 'ОШИБКА'}")
print("Все восемь линий верны для обоих игроков:",
      vse_linii_ok and bez_pobeditelya is None)
```

## Проверка результата

```python
assert proverit_pobedu(["X","X","X","","","","","",""]) == "X"
assert proverit_pobedu(["O","X","O","X","O","X","X","O","X"]) is None
print("Обе проверки пройдены.")
```

## Задание ★ Базовая практика — ничья

```python
def pole_zapolneno(polya_teksty):
    return all(t != "" for t in polya_teksty)

polnaya_bez_pobeditelya = ["X","O","X","X","O","O","O","X","X"]
print("Заполнено:", pole_zapolneno(polnaya_bez_pobeditelya))
print("Победитель:", proverit_pobedu(polnaya_bez_pobeditelya))
```
