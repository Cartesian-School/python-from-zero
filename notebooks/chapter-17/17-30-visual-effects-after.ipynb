## Цель

Собрать один спокойный неблокирующий переход победной линии через after().

## Про mainloop() в этом ноутбуке

В остальных ноутбуках главы вместо `root.mainloop()` используется `root.update()` + `root.destroy()`. Здесь — исключение, и оно осмысленное: `after(400, ...)` планирует вызов через 400 мс, и чтобы это время реально прошло, событийный цикл должен поработать. Поэтому ниже запускается настоящий `mainloop()`, а выйти из него помогает `root.after(500, root.quit)`.

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
buttons = [tk.Button(root, text="X") for _ in range(3)]
for b in buttons:
    b.pack(side="left")

ticks_seen = []

def pulse(tick=0):
    ticks_seen.append(tick)
    color = "#6EE7B7" if tick == 0 else "#D1FAE5"
    for b in buttons:
        b.config(bg=color)
    if tick == 0:
        root.after(400, pulse, 1)

pulse()
root.update()
root.after(500, root.quit)
root.mainloop()

print("Тиков анимации:", ticks_seen)
root.destroy()
```

## Задание ★★ Самостоятельная задача

Добавьте `cancel_pulse()`, который вызывает `root.after_cancel(job_id)`, и проверьте, что после отмены анимация не продолжается.

```python
root = tk.Tk()
buttons = [tk.Button(root, text="X") for _ in range(3)]
for b in buttons:
    b.pack(side="left")

ticks = []
job = {"id": None}

def pulse(tick=0):
    ticks.append(tick)
    color = "#6EE7B7" if tick == 0 else "#D1FAE5"
    for b in buttons:
        b.config(bg=color)
    if tick == 0:
        job["id"] = root.after(400, pulse, 1)   # запомнили id запланированного вызова
    else:
        job["id"] = None

def cancel_pulse():
    if job["id"] is not None:
        root.after_cancel(job["id"])
        job["id"] = None

pulse()             # первый тик выполнен, второй — запланирован
root.update()
cancel_pulse()      # ...и тут же отменён

root.after(600, root.quit)   # ждём дольше 400 мс: если бы тик не отменили, он бы сработал
root.mainloop()

print("Тиков после отмены:", ticks)
assert ticks == [0], "второй тик был отменён — его быть не должно"
assert job["id"] is None
print("after_cancel() снял запланированный вызов: на новом раунде он уже не сработает.")
root.destroy()
```
