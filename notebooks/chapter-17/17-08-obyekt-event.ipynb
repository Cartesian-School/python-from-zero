## Цель

Посмотреть на реальные поля объекта tkinter.Event для разных типов событий.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример — событие клавиатуры

```python
import tkinter as tk

root = tk.Tk()
label = tk.Label(root, text="...")
label.pack()
poslednee = {}

def on_key(event):
    poslednee["type"] = str(event.type)
    poslednee["keysym"] = event.keysym
    poslednee["char"] = event.char
    label.config(text=f"keysym={event.keysym!r} char={event.char!r}")

root.bind("<Key>", on_key)
root.update()
root.focus_force()   # без фокуса окна event_generate для клавиш никуда не доставит событие
root.update()
label.event_generate("<Key>", keysym="Return")
root.update()
print(poslednee)
```

## Эксперимент — событие мыши (Enter/Leave)

```python
poslednee_mouse = {}

def on_enter(event):
    poslednee_mouse["widget"] = str(event.widget)
    poslednee_mouse["type"] = str(event.type)

label.bind("<Enter>", on_enter)
root.update()
label.event_generate("<Enter>")
root.update()
print(poslednee_mouse)
# Обратите внимание: у события <Enter> event.keysym/event.char не несут полезной информации —
# это поля, осмысленные для КЛАВИАТУРНЫХ событий.
root.destroy()
```

## Задание ★★ Самостоятельная задача

Привяжите `<Button-1>` к отдельному виджету и распечатайте `event.x`/`event.y` — координаты клика внутри виджета.

```python
root = tk.Tk()
metka = tk.Label(root, text="Кликните по мне", width=20, height=3)
metka.pack()
koordinaty = {}

def na_klik(event):
    koordinaty["x"] = event.x
    koordinaty["y"] = event.y

metka.bind("<Button-1>", na_klik)
root.update()

# программно кликаем в точку (12, 7) ВНУТРИ виджета
metka.event_generate("<Button-1>", x=12, y=7)
root.update()

print("Координаты клика внутри виджета:", koordinaty)
assert koordinaty == {"x": 12, "y": 7}
print("event.x/event.y отсчитываются от левого верхнего угла самого виджета, а не экрана.")
root.destroy()
```
