# 17-23 · Hover preview через bind()

Практика к разделу [«Hover preview через bind()»](../../site/chapters/glava-17/17-23-hover-preview.html).

## Цель

Собрать hover-превью на реальной кнопке и доказать, что модель не меняется.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
board = [""] * 9
btn = tk.Button(root, text="")
btn.pack()

def on_enter(_event, index=4):
    if board[index]:
        return
    btn.config(text="X", fg="#B9A0FC")

def on_leave(_event, index=4):
    btn.config(text=board[index], fg="#0D0230")

btn.bind("<Enter>", on_enter)
btn.bind("<Leave>", on_leave)
root.update()

btn.event_generate("<Enter>")
root.update()
print("Во время наведения — текст кнопки:", btn.cget("text"), " модель:", board[4])

btn.event_generate("<Leave>")
root.update()
print("После ухода курсора — текст кнопки:", repr(btn.cget("text")), " модель:", repr(board[4]))
root.destroy()
```

## Задание ★★ Самостоятельная задача

Проверьте, что наведение на клетку, где `board[index]` уже не пустое, НЕ переписывает существующую отметку.

```python
root = tk.Tk()
board = ["", "", "", "", "X", "", "", "", ""]   # клетка 4 уже занята ходом X
btn = tk.Button(root, text=board[4])
btn.pack()

def on_enter(_event, index=4):
    if board[index]:
        return                      # клетка занята — превью не показываем
    btn.config(text="O", fg="#B9A0FC")

btn.bind("<Enter>", on_enter)
root.update()

btn.event_generate("<Enter>")
root.update()

print("Текст занятой клетки после наведения:", repr(btn.cget("text")))
print("Модель:", repr(board[4]))
assert btn.cget("text") == "X", "наведение не должно перезаписывать уже сделанный ход"
assert board[4] == "X"
print("Превью не рисуется поверх занятой клетки — чужой ход остался на месте.")
root.destroy()
```
