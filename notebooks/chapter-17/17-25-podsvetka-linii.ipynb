## Цель

Подсветить ИМЕННО победную линию, а не любые совпавшие символы на поле.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

WIN_BG = "#D1FAE5"
NEUTRAL_BG = "#FAFAFC"

root = tk.Tk()
board = ["X", "X", "X", "X", "", "", "", "", ""]  # верхняя строка ПОБЕДНАЯ, четвёртый X — нет
winning_line = (0, 1, 2)

buttons = [tk.Button(root, text=board[i]) for i in range(9)]
for i, btn in enumerate(buttons):
    btn.config(bg=WIN_BG if i in winning_line else NEUTRAL_BG)
root.update()

for i in (0, 1, 2, 3):
    print(i, buttons[i].cget("bg") == WIN_BG)
root.destroy()
```

## Задание ★★ Самостоятельная задача

Убедитесь, что клетка 3 (тоже 'X', но не часть `winning_line`) НЕ подсвечена — используйте `bg` кнопки для проверки.

```python
root = tk.Tk()
board = ["X", "X", "X", "X", "", "", "", "", ""]
winning_line = (0, 1, 2)

buttons = [tk.Button(root, text=board[i]) for i in range(9)]
for i, btn in enumerate(buttons):
    btn.config(bg=WIN_BG if i in winning_line else NEUTRAL_BG)
root.update()

print("Клетка 3: текст", repr(buttons[3].cget("text")),
      "| фон", buttons[3].cget("bg"))
assert buttons[3].cget("text") == "X", "клетка 3 тоже занята символом X"
assert buttons[3].cget("bg") == NEUTRAL_BG, \
    "но она НЕ входит в winning_line, поэтому подсвечивать её нельзя"
assert all(buttons[i].cget("bg") == WIN_BG for i in winning_line)
print("Подсвечена ровно победная линия, а не все клетки с «X».")
root.destroy()
```
