## Цель

Создать переменные состояния и поле из девяти кнопок.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()

tekuschij_igrok = "X"
polya = []
igra_okonchena = False

pole_frame = tk.Frame(root)
pole_frame.pack()

for indeks in range(9):
    knopka = tk.Button(pole_frame, text="", font=("Arial", 24, "bold"), width=3, height=1)
    knopka.grid(row=indeks // 3, column=indeks % 3)
    polya.append(knopka)

print("Создано кнопок:", len(polya))
print("Позиция кнопки 4 (центр):", polya[4].grid_info()["row"], polya[4].grid_info()["column"])
root.update()
root.destroy()
```
