# 17-24 · Управление клавиатурой

Практика к разделу [«Управление клавиатурой»](../../site/chapters/glava-17/17-24-upravlenie-klaviaturoj.html).

## Цель

Показать, что клавиша '1'..'9' ведёт в ТОТ ЖЕ attempt_move(), что и клик мышью.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
board = [""] * 9
current_player = ["X"]
moves_log = []

def attempt_move(index):
    if board[index]:
        return
    board[index] = current_player[0]
    moves_log.append(index)
    current_player[0] = "O" if current_player[0] == "X" else "X"

def on_key(event):
    if event.char and event.char.isdigit():
        n = int(event.char)
        if 1 <= n <= 9:
            attempt_move(n - 1)

root.bind("<Key>", on_key)
root.update()
root.focus_force()   # без фокуса окна event_generate для клавиш никуда не доставит событие
root.update()

root.event_generate("<Key>", keysym="3")
root.update()
print("board после клавиши '3':", board)
print("Ход записан в тот же moves_log:", moves_log)
root.destroy()
```

## Задание ★ Базовая практика

Проверьте, что клавиша `0` (вне диапазона 1-9) не вызывает `attempt_move()`.

```python
root = tk.Tk()
board = [""] * 9
moves_log = []

def attempt_move(index):
    if board[index]:
        return
    board[index] = "X"
    moves_log.append(index)

def on_key(event):
    if event.char and event.char.isdigit():
        n = int(event.char)
        if 1 <= n <= 9:
            attempt_move(n - 1)

root.bind("<Key>", on_key)
root.update()
root.focus_force()
root.update()

root.event_generate("<Key>", keysym="0")
root.update()
print("После клавиши '0':", board, "| ходов:", moves_log)
assert moves_log == [], "клавиша 0 вне диапазона 1-9 — ход делаться не должен"

root.event_generate("<Key>", keysym="9")
root.update()
print("После клавиши '9':", board, "| ходов:", moves_log)
assert moves_log == [8], "клавиша 9 — это последняя клетка, индекс 8"
print("В attempt_move() попадают только клавиши 1-9.")
root.destroy()
```
