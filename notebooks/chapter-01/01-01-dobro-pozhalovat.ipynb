## Краткое напоминание

Команда `print(...)` выводит на экран переданные ей значения. Строковый литерал записывают
в одинарных `'...'` или двойных `"..."` кавычках; выбранный вид кавычек не меняет тип строки.

## Рабочий пример

```python
print("Привет, мир!")
```

## Эксперимент 1

Замените текст внутри кавычек на своё собственное приветствие и запустите ячейку снова.

```python
print("Привет, Cartesian School!")
```

## Эксперимент 2

`print()` умеет выводить не только текст. Попробуйте передать ему число или даже пример
арифметики — Python сам всё посчитает.

```python
print(2024 - 1991)
print(7 * 6)
```

## Типичная ошибка

Если убрать кавычки, Python решит, что вы обращаетесь к чему-то по имени — переменной или
функции, — а не к обычному тексту. Такого имени не существует, и Python сообщит об ошибке
`NameError`.

```python
# ошибка: текст без кавычек
print(Привет)
```

## Исправление

Достаточно вернуть кавычки — тогда Python понимает, что это обычный текст, а не имя
переменной.

```python
print("Привет")
```

## Заглянем внутрь: type()

У каждого значения в Python есть тип. Функция `type()` позволяет спросить реализацию Python,
к какому типу относится конкретное значение.

```python
print(type("Привет"))
print(type(42))
print(type(3.14))
```

## Спросим сам Python: help()

Функция `help()` показывает встроенную справку, доступную в установленной реализации. Она
полезна рядом с кодом, но не обязана дословно совпадать с более полной и версионированной
страницей docs.python.org.

```python
help(print)
```
