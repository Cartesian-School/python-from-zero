## Цель

Различать выбор с повторами (choices) и без повторов (sample).

## Рабочий пример

```python
import random
varianty = ["камень", "ножницы", "бумага"]
print(random.choice(varianty))
```

## Задание ★ Базовая практика

Зафиксируйте `random.seed(3)`, затем выведите `random.sample(range(1, 6), k=3)` — три разных числа без повторов (ожидается `[2, 5, 4]`).

```python
import random
random.seed(3)
print(random.sample(range(1, 6), k=3))
```
