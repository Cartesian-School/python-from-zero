Найти и исправить ошибку приоритета операций в готовой формуле.

## Типичная ошибка

Эта формула должна была посчитать среднее трёх оценок, но забыла скобки — результат неверен.

```python
ocenka1, ocenka2, ocenka3 = 4, 5, 3
srednyaya_s_oshibkoj = ocenka1 + ocenka2 + ocenka3 / 3
print(srednyaya_s_oshibkoj)   # 10.0 — неверно! это НЕ среднее трёх оценок
```

## Задание ★ Базовая практика

Исправьте формулу так, чтобы она правильно считала среднее трёх оценок (добавьте скобки вокруг суммы). Ожидается 4.0.

```python
ocenka1, ocenka2, ocenka3 = 4, 5, 3
srednyaya = (ocenka1 + ocenka2 + ocenka3) / 3
print(srednyaya)
```
