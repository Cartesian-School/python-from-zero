Практика к разделу [«Ассоциативность операторов»](../../site/chapters/glava-05/05-10-associativnost.html).

## Цель

Увидеть на практике, что ** — правоассоциативно, а большинство операторов — левоассоциативны.

## Рабочий пример

```python
print(2 ** 3 ** 2)     # 512, а не 64
```

## Эксперимент 1

Сравните левую ассоциативность вычитания.

```python
print(20 - 5 - 3)      # (20 - 5) - 3 = 12, а не 20 - (5 - 3) = 18
```

## Задание ★ Базовая практика

Выведите `2 ** 2 ** 3` — из-за правой ассоциативности это 2 ** (2 ** 3), а не (2 ** 2) ** 3.

```python
print(2 ** 2 ** 3)
```
