## Цель

Освоить разницу между randint() (обе границы включены) и randrange() (верхняя не включена).

## Рабочий пример

```python
import random
print(random.randint(1, 6))
```

## Эксперимент 1

randrange(1, 7) — аналог randint(1, 6), потому что верхняя граница не включена.

```python
import random
print(random.randrange(1, 7))
```

## Задание ★ Базовая практика

Зафиксируйте `random.seed(1)`, затем выведите `random.randint(1, 10)` — с этим seed результат всегда одинаковый (ожидается 3), что позволяет проверить задание автоматически.

```python
import random
random.seed(1)
print(random.randint(1, 10))
```
