# 05-08 · Отрицательное floor-деление

Практика к разделу [«Отрицательное floor-деление»](../../site/chapters/glava-05/05-08-otricatelnoe-delenie.html).

## Цель

Предсказывать результат // и % для отрицательных чисел.

## Рабочий пример

```python
print(-7 // 2)
print(-7 % 2)
```

## Эксперимент 1 · Предскажите перед запуском

Предскажите результат, затем проверьте.

```python
print(7 // -2)
print(7 % -2)
```

## Задание ★ Базовая практика

Выведите `-17 // 5` и `-17 % 5` через пробел одним print().

```python
print(-17 // 5, -17 % 5)
```
