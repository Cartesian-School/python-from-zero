## Задание ★ Базовая практика

Посчитайте sin(90°), округлив до 2 знаков (не забудьте перевести градусы в радианы!). Ожидается 1.0.

```python
import math
ugol_radiany = math.radians(90)
print(round(math.sin(ugol_radiany), 2))
```
