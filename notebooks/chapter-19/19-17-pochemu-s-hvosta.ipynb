# 19-17 · Почему тело движется с хвоста

Практика к разделу [«Почему тело движется с хвоста»](../../site/chapters/glava-19/19-17-pochemu-s-hvosta.html).

## Цель

Показать, что обновление от хвоста к голове не теряет позиции.

## Рабочий пример

```python
def move_from_tail(segments, head_pos):
    segments = list(segments)
    for i in range(len(segments) - 1, 0, -1):
        segments[i] = segments[i - 1]
    if segments:
        segments[0] = head_pos
    return segments

print(move_from_tail([(0, 0), (-20, 0), (-40, 0)], (20, 0)))
```

## Задание ★★ Найдите баг в обратном порядке

```python
def move_from_head_BROKEN(segments, head_pos):
    segments = list(segments)
    segments[0] = head_pos
    for i in range(1, len(segments)):
        segments[i] = segments[i - 1]   # уже перезаписанное значение!
    return segments

correct = move_from_tail([(0, 0), (-20, 0), (-40, 0)], (20, 0))
broken = move_from_head_BROKEN([(0, 0), (-20, 0), (-40, 0)], (20, 0))
print("Правильно:", correct)
print("Сломано:", broken)
assert correct == [(20, 0), (0, 0), (-20, 0)]
assert broken == [(20, 0), (20, 0), (20, 0)]   # все сегменты схлопнулись
print("Подтверждено: обратный порядок схлопывает сегменты в одну точку.")
```
