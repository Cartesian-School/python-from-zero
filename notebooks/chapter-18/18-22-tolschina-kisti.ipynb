## Цель

Scale для толщины линии, связанный с состоянием рисования.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
width = {"value": 1}   # НЕ 12 — иначе не будет видно, сработал ли command

def set_width(value):
    width["value"] = int(value)

scale = tk.Scale(root, from_=1, to=20, orient="horizontal", command=set_width)
scale.pack()
scale.set(12)
root.update()
print("Текущая толщина:", width["value"])   # 12, а не 1 — значит command сработал
root.destroy()
```

## Проверка результата

```python
assert width["value"] == 12, "command у Scale должен был записать 12 в width"
print("Верно: ползунок изменил состояние, а не только своё внутреннее значение.")
```
