## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import sys
sys.path.insert(0, "../../projects/tkinter/paint-app")

import tkinter as tk
import paint_app as p

root = tk.Tk()
app = p.PaintApp(root)
root.update()

class FakeEvent:
    def __init__(self, x, y):
        self.x, self.y = x, y

app.set_tool(p.Tool.RECTANGLE)
app.on_press(FakeEvent(10, 10))
app.on_drag(FakeEvent(100, 80))
app.on_release(FakeEvent(100, 80))
print("Фигур в документе:", len(app.document))
root.destroy()
```
