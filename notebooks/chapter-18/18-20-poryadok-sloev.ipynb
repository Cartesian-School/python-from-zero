Практика к разделу [«Порядок наложения»](../../site/chapters/glava-18/18-20-poryadok-sloev.html).

## Цель

Смоделировать tag_raise() и tag_lower() как список, где последний элемент рисуется сверху.

## Рабочий пример

```python
def podnyat(poryadok, item_id):
    """Аналог canvas.tag_raise(item_id) — переносит элемент в КОНЕЦ списка (наверх)."""
    poryadok.remove(item_id)
    poryadok.append(item_id)
    return poryadok


def opustit(poryadok, item_id):
    """Аналог canvas.tag_lower(item_id) — переносит элемент в НАЧАЛО списка (вниз)."""
    poryadok.remove(item_id)
    poryadok.insert(0, item_id)
    return poryadok


poryadok = [12, 13, 14]
podnyat(poryadok, 12)
print("после podnyat(12):", poryadok)

poryadok_vniz = [12, 13, 14]
opustit(poryadok_vniz, 14)
print("после opustit(14):", poryadok_vniz)
```

## Задание ★ Базовая практика

```python
assert poryadok == [13, 14, 12]
assert poryadok_vniz == [14, 12, 13]

poryadok2 = [1, 2, 3]
podnyat(poryadok2, 2)
assert poryadok2 == [1, 3, 2]
opustit(poryadok2, 2)
assert poryadok2 == [2, 1, 3], "tag_lower опускает элемент под все остальные"
print("Последний элемент списка — верхний в порядке наложения, первый — нижний.")
```
