# 18-20 · Порядок наложения

Практика к разделу [«Порядок наложения»](../../site/chapters/glava-18/18-20-poryadok-sloev.html).

## Цель

Смоделировать tag_raise() как список, где последний элемент рисуется сверху.

## Рабочий пример

```python
def podnyat(poryadok, item_id):
    """Аналог canvas.tag_raise(item_id) — переносит элемент в конец списка (наверх)."""
    poryadok.remove(item_id)
    poryadok.append(item_id)
    return poryadok


poryadok = [12, 13, 14]
podnyat(poryadok, 12)
print(poryadok)
```

## Задание ★ Базовая практика

```python
assert poryadok == [13, 14, 12]
poryadok2 = [1, 2, 3]
podnyat(poryadok2, 2)
assert poryadok2 == [1, 3, 2]
print("Последний элемент списка — верхний в порядке наложения.")
```
