## Цель

Воспроизвести и починить два бага истории: очистка холста обязана обнулять историю (Debug Lab 11), а новое действие — обнулять redo (Debug Lab 10).

## Рабочий пример — Debug Lab 11: очистка и история

```python
document = ["A"]
undo_stack = [["A"]]
redo_stack = []


def clear_canvas(document, undo_stack, redo_stack):
    """История описывает документ. Нет документа — нет и истории."""
    document.clear()
    undo_stack.clear()
    redo_stack.clear()


clear_canvas(document, undo_stack, redo_stack)
print(document, undo_stack, redo_stack)
```

## Задание ★ Базовая практика

```python
assert document == []
assert undo_stack == [], "история отмены должна очищаться вместе с документом"
assert redo_stack == []
print("Очистка холста обнулила и документ, и всю его историю.")
```

## Рабочий пример — Debug Lab 10: новое действие обнуляет redo

Второй баг из справочника: если `_commit_action()` забывает очистить `redo_stack`, то Ctrl+Y после нового рисунка возвращает на холст фигуру из отменённой ветки, никак не связанную с текущим документом.

```python
doc2, undo2, redo2 = [], [], []


def commit_action(shapes):
    doc2.extend(shapes)
    undo2.append(shapes)
    redo2.clear()   # ← строка, без которой возникает Debug Lab 10


def undo():
    if not undo2:
        return
    shapes = undo2.pop()
    del doc2[len(doc2) - len(shapes):]
    redo2.append(shapes)


commit_action(["A"])
commit_action(["B"])
undo()                 # отменили B — оно ушло в redo2
commit_action(["C"])   # новое действие ПОСЛЕ отмены
print(doc2, redo2)
```

## Задание ★ Базовая практика

```python
assert doc2 == ["A", "C"]
assert redo2 == [], "новое действие обязано сделать старую ветку redo недействительной"
print("Debug Lab 10 воспроизведён и починен: redo не воскресит чужую фигуру.")
```
