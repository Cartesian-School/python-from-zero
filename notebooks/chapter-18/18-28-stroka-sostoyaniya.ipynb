## Цель

Строка состояния, обновляемая по координатам мыши и параметрам инструмента.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
canvas = tk.Canvas(root, width=300, height=200, bg="white")
canvas.pack()
status_var = tk.StringVar()
tk.Label(root, textvariable=status_var).pack()

def update_status(x, y):
    status_var.set(f"Инструмент: Карандаш | x={x} y={y} | Цвет: #111827 | Толщина: 4")

canvas.bind("<Motion>", lambda event: update_status(event.x, event.y))
root.update()
update_status(0, 0)
print(status_var.get())
root.destroy()
```
