## Цель

Привязать Ctrl+Z/Ctrl+Y к отмене и повтору на уровне окна.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
log = []

def undo(_event=None):
    log.append("undo")

def redo(_event=None):
    log.append("redo")

root.bind("<Control-z>", undo)
root.bind("<Control-y>", redo)
root.update()

# ОСТОРОЖНО: root.bind("<Control-z>") возвращает СТРОКУ-привязку, а для
# несуществующей привязки — пустую строку "". Поэтому проверять надо через
# bool(...), а не через `is not None`: пустая строка тоже "не None"!
print("Ctrl+Z привязан:", bool(root.bind("<Control-z>")))
print("Ctrl+Q не привязан:", bool(root.bind("<Control-q>")))

# Сами обработчики вызываем напрямую — как если бы клавиши уже нажали
undo()
redo()
undo()
print("Журнал вызовов:", log)
root.destroy()
```

## Проверка результата

```python
assert log == ["undo", "redo", "undo"], "порядок вызовов должен совпадать с порядком нажатий"
print("Верно: обработчики отработали в том же порядке, в каком их вызвали.")
```
