## Цель

Изменить уже существующий элемент Canvas: coords, itemconfig, move, delete.

## Про mainloop() в этом ноутбуке

Вместо `root.mainloop()` здесь используется `root.update()` + `root.destroy()` — приложение строится точно так же, просто без бесконечного ожидания в конце (см. объяснение в ноутбуках главы 16).

## Рабочий пример

```python
import tkinter as tk

root = tk.Tk()
canvas = tk.Canvas(root, width=300, height=200, bg="white")
canvas.pack()

item_id = canvas.create_rectangle(10, 10, 80, 60, outline="black", width=2)
root.update()

canvas.itemconfig(item_id, outline="blue", width=4)
canvas.move(item_id, 30, 0)
root.update()
print("Текущие координаты:", canvas.coords(item_id))
root.destroy()
```
