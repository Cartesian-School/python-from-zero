# 08-13 · Длина строки: len()

Практика к разделу [«Длина строки: len()»](../../site/chapters/glava-08/08-13-dlina-stroki.html).

## Цель

Считать символы строки функцией len() и понимать связь длины с индексами.

## Рабочий пример

```python
word = "Python"
print(len(word))

phrase = "Python с нуля"
print(len(phrase))
```

## Задание ★ Базовая практика

Для слова «Cartesian» найдите его длину `dlina` и индекс последнего символа `posledniy_index` (через len()) — затем убедитесь, что `word[posledniy_index]` совпадает с `word[-1]`.

```python
word = "Cartesian"
dlina = len(word)
posledniy_index = dlina - 1
print(dlina, posledniy_index)
print(word[posledniy_index] == word[-1])
```
