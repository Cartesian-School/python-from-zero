## Цель

Научиться находить невидимые пробелы и другие частые ошибки строк.

## Рабочий пример

```python
password = "секрет "
print(password == "секрет")
print(repr(password))
```

## Задание ★ Базовая практика

Проверьте `password = "секрет "` (с пробелом в конце): сравните его с `"секрет"` в `is_equal_before`, затем очистите через strip() в `cleaned` и сравните снова в `is_equal_after`.

```python
password = "секрет "
is_equal_before = password == "секрет"
cleaned = password.strip()
is_equal_after = cleaned == "секрет"
print(is_equal_before, is_equal_after)
```
