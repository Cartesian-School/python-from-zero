## Цель

Освоить replace, split, join, count, find, index, startswith, endswith.

## Рабочий пример

```python
text = "я люблю Java"
print(text.replace("Java", "Python"))

sentence = "кот и пёс"
words = sentence.split()
print(words)
print("-".join(words))
```

## Эксперимент 1 — find() и index()

```python
text = "Python"
print(text.find("th"))
print(text.find("zz"))
```

## Задание ★ Базовая практика

Для имени файла «report_final.pdf» проверьте, заканчивается ли оно на «.pdf» (`is_pdf`), найдите позицию символа «_» (`position`), и разбейте имя по «_» на список частей (`parts`).

```python
filename = "report_final.pdf"
is_pdf = filename.endswith(".pdf")
position = filename.find("_")
parts = filename.split("_")
print(is_pdf, position, parts)
```
