# 08-18 · Перебор строки в цикле

Практика к разделу [«Перебираем строку в цикле»](../../site/chapters/glava-08/08-18-cikl-po-stroke.html).

## Цель

Пройтись по строке циклом for и посчитать вхождения символа вручную.

## Рабочий пример

```python
word = "Python"
for ch in word:
    print(ch)
```

## Задание ★ Базовая практика

В строке «миссисипи» посчитайте в цикле for, сколько раз встречается буква «и» — сохраните результат в `count`.

```python
text = "миссисипи"
bukva = "и"
count = 0
for ch in text:
    if ch == bukva:
        count += 1
print(count)
```
