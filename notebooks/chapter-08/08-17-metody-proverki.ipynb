# 08-17 · Методы проверки строки

Практика к разделу [«Методы проверки: isalpha и другие»](../../site/chapters/glava-08/08-17-metody-proverki.html).

## Цель

Освоить isalpha, isdigit, isalnum, isspace для проверки состава строки.

## Рабочий пример

```python
print("Python".isalpha())
print("2026".isdigit())
print("Python3".isalnum())
print("   ".isspace())
```

## Задание ★ Базовая практика

Проверьте три строки: `s1 = "Python"` методом isalpha() → `r1`, `s2 = "2026"` методом isdigit() → `r2`, `s3 = "   "` методом isspace() → `r3`.

```python
s1 = "Python"
s2 = "2026"
s3 = "   "
r1 = s1.isalpha()
r2 = s2.isdigit()
r3 = s3.isspace()
print(r1, r2, r3)
```
