# 04-09 · Почините порядок операций

Практика к разделу [«Порядок выполнения операций»](../../site/chapters/glava-04/04-09-poryadok-operacij.html).

## Цель

Предсказывать результат выражений с учётом приоритета операторов.

## Рабочий пример

```python
print(2 + 3 * 4)
```

## Эксперимент 1

Предскажите, затем проверьте — как скобки меняют результат.

```python
print((2 + 3) * 4)
```

## Задание ★ Базовая практика

Выражение `10 - 2 ** 2` должно дать 6 без скобок — проверьте, что приоритет ** выше, чем -.

```python
print(10 - 2 ** 2)
```

## Самостоятельная практика ★★

Расставьте скобки так, чтобы `2 + 3 * 4 - 1` дало 19 (а не 13).

```python
print((2 + 3) * (4 - 1) + 4)
```
