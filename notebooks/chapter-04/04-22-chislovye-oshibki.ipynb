## Цель

Найти и исправить типичные числовые ошибки.

## Рабочий пример

Эта ячейка работает нормально — разминка перед поиском ошибок.

```python
price = 100
quantity = 2
print(price * quantity)
```

## Лаборатория 1 · ValueError

Эта ячейка упадёт — прочитайте ошибку, затем исправьте её ниже.

```python
print(int("FF"))
```

## Исправление 1

```python
print(int("FF", 16))
```

## Лаборатория 2 · ZeroDivisionError

```python
price = 100
quantity = 0
print(price / quantity)
```

## Исправление 2

Проверьте quantity перед делением (используйте любое ненулевое значение здесь).

```python
price = 100
quantity = 4
print(price / quantity)
```
