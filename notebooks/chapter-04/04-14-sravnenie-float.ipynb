# 04-14 · Почините сравнение float

Практика к разделу [«Сравниваем float правильно»](../../site/chapters/glava-04/04-14-sravnenie-float.html).

## Цель

Заменить ненадёжный == на math.isclose().

## Рабочий пример

Это сравнение ненадёжно — проверьте сами.

```python
print(0.1 + 0.2 == 0.3)
```

## Исправление

```python
from math import isclose
print(isclose(0.1 + 0.2, 0.3))
```

## Задание ★ Базовая практика

Исправьте сравнение `1.1 + 2.2 == 3.3`, используя isclose().

```python
from math import isclose
print(isclose(1.1 + 2.2, 3.3))
```
