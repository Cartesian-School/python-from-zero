Практика к разделу [«Почему 0.1 + 0.2 не равно 0.3»](../../site/chapters/glava-04/04-13-pochemu-01-02.html).

## Цель

Увидеть погрешность float своими глазами и понять, что она объяснима.

## Рабочий пример

```python
print(0.1 + 0.2)
```

## Эксперимент 1

Посмотрите, что реально хранится за 0.1.

```python
print((0.1).as_integer_ratio())
```

## Эксперимент 2

Предскажите: будет ли `0.5 + 0.25` равно ровно `0.75`? Проверьте — иногда сумма получается точной.

```python
print(0.5 + 0.25 == 0.75)
```

## Задание ★ Базовая практика

Проверьте `1.1 + 2.2` — равно ли оно ровно `3.3`?

```python
print(1.1 + 2.2)
print(1.1 + 2.2 == 3.3)
```
