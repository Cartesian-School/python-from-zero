## Цель

Понять разницу между Decimal(str) и Decimal(float).

## Рабочий пример

```python
from decimal import Decimal
print(Decimal("0.1") + Decimal("0.2"))
```

## Эксперимент 1

Сравните Decimal из строки и Decimal из float.

```python
from decimal import Decimal
print(Decimal("19.99"))
print(Decimal(19.99))
```

## Задание ★ Базовая практика

Посчитайте сумму покупки: цена 24.99, количество 3 — используя Decimal, созданный из строки.

```python
from decimal import Decimal
price = Decimal("24.99")
quantity = 3
print(price * quantity)
```
