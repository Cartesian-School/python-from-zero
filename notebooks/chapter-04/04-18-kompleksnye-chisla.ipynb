## Цель

Освоить real, imag и abs() для комплексных чисел.

## Рабочий пример

```python
z = 3 + 4j
print(z.real)
print(z.imag)
```

## Эксперимент 1

Проверьте abs() — длину вектора от начала координат.

```python
z = 3 + 4j
print(abs(z))
```
