# 12-15 · Корзина покупок

Практика к разделу [«Проекты: журнал оценок и корзина покупок»](../../site/chapters/glava-12/12-14-zhurnal-i-korzina.html#korzina-pokupok).

## Цель

Список словарей + накопитель: чек и итоговая сумма.

## Рабочий пример

```python
cart = [
    {"name": "Молоко", "price": 4.50, "qty": 2},
    {"name": "Хлеб", "price": 2.20, "qty": 1},
    {"name": "Сыр", "price": 9.90, "qty": 1},
]

total = 0
for item in cart:
    line_total = item["price"] * item["qty"]
    total += line_total
    print(f"{item['name']:<10} {item['qty']} x {item['price']:.2f} = {line_total:.2f}")

total = round(total, 2)
print(f"Итого: {total:.2f}")
```
