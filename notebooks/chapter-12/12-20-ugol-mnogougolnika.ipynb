Чистая геометрия без Turtle: одна формула для любого числа сторон.

## Рабочий пример

```python
storony_1 = 5
ugol_1 = 360 / storony_1

storony_2 = 8
ugol_2 = 360 / storony_2

print(f"{storony_1} сторон → угол {ugol_1}°")
print(f"{storony_2} сторон → угол {ugol_2}°")
```
