## Цель

while True + if/elif/else + break — паттерн, который встречается в меню, консолях и ботах.

## Про input() в этом ноутбуке

Этот ноутбук выполняется автоматически, поэтому `input()` временно подменён на заранее заготовленные ответы.

```python
_answers = iter(['help', 'hello', 'status', 'exit'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Рабочий пример

```python
history = []

while True:
    command = input("> ").strip().lower()
    history.append(command)

    if command == "exit":
        print("До встречи!")
        break
    elif command == "help":
        print("Команды: help, hello, status, exit")
    elif command == "hello":
        print("Привет!")
    elif command == "status":
        print(f"Команд выполнено: {len(history)}")
    else:
        print(f"Неизвестная команда: {command}")

print(history)
```
