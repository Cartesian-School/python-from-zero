## Цель

Подсчёт частоты слов и поиск самого частого слова через max(key=...).

## Про input() в этом ноутбуке

Этот ноутбук выполняется автоматически, поэтому `input()` временно подменён на заранее заготовленные ответы.

```python
_answers = iter(['Python is great and python is fun'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Рабочий пример

```python
text = input("Введите текст: ")
normalized = text.lower().split()

counts = {}
for word in normalized:
    counts[word] = counts.get(word, 0) + 1

samoe_chastoe = max(counts, key=counts.get)

print(counts)
print("Самое частое слово:", samoe_chastoe, "—", counts[samoe_chastoe], "раз(а)")
```
