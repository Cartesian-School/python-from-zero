## Цель

От задачи до кода на маленькой программе: список, min/max/sum/среднее.

## Про input() в этом ноутбуке

Этот ноутбук выполняется автоматически, поэтому `input()` временно подменён на заранее заготовленные ответы.

```python
_answers = iter(['10', '20', '30'])

def input(prompt=""):
    answer = next(_answers)
    print(prompt + answer)
    return answer
```

## Рабочий пример

```python
a = float(input("Первое число: "))
b = float(input("Второе число: "))
c = float(input("Третье число: "))

chisla = [a, b, c]
minimum = min(chisla)
maksimum = max(chisla)
summa = sum(chisla)
average = summa / len(chisla)

print(f"Минимум: {minimum}")
print(f"Максимум: {maksimum}")
print(f"Сумма: {summa}")
print(f"Среднее: {average:.2f}")
```
