# 13-16 · nonlocal

Практика к разделу [«Вложенные функции и nonlocal»](../../site/chapters/glava-13/13-16-vlozhennye-funkcii-nonlocal.html).

## Цель

Использовать nonlocal, чтобы вложенная функция изменяла счётчик объемлющей функции.

## Задание ★ Базовая практика

```python
def outer():
    count = 0

    def inner():
        nonlocal count
        count += 1

    inner()
    inner()
    inner()
    return count

result = outer()
print(result)
```
