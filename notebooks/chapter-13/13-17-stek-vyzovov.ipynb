Практика к разделу [«Стек вызовов и traceback»](../../site/chapters/glava-13/13-17-stek-vyzovov.html).

## Цель

Проследить порядок вызовов и возвратов во вложенных функциях.

## Задание ★ Базовая практика

```python
def c():
    print("C")
    return 3

def b():
    print("B")
    return c() + 1

def a():
    print("A")
    return b() + 1

result = a()
print("Итог:", result)
```
