## Цель

Предсказать, что видно снаружи после rebinding и после mutation.

## Задание ★ Базовая практика

```python
def add_one(number):
    number += 1
    return number

x = 10
y = add_one(x)

def add_item(items):
    items.append("Python")

skills = ["Git"]
add_item(skills)

print(x, y, skills)
```
