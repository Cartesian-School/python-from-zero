## Задание ★ Базовая практика

```python
def build_profile(name, age, city="Unknown"):
    return f"{name}, {age}, {city}"

p1 = build_profile("Anna", 25)
p2 = build_profile(age=30, name="Bob", city="Warsaw")
print(p1)
print(p2)
```
