## Цель

*args, **kwargs и распаковка аргументов на месте вызова.

## Задание ★ Базовая практика

```python
def total(*numbers):
    return sum(numbers)

def show_profile(**fields):
    return fields

def move(x, y):
    return x + y

t1 = total(1, 2, 3, 4)
profile = show_profile(name="Anna", city="Warsaw")

point = (10, 20)
moved = move(*point)

print(t1, profile, moved)
```
