# 13-09 · Цикл или функция?

Практика к разделу [«Зачем программе функции»](../../site/chapters/glava-13/13-09-zachem-programme-funkcii.html).

## Цель

Определить и переиспользовать функцию для вычисления.

## Рабочий пример

```python
def summa_kvadratov(a, b):
    return a ** 2 + b ** 2

def kvadrat_summy(a, b):
    return (a + b) ** 2

print(summa_kvadratov(3, 4))
print(kvadrat_summy(3, 4))
```

## Задание ★ Базовая практика

Используйте обе функции, чтобы найти разницу между квадратом суммы и суммой квадратов для a=3, b=4.

```python
raznost = kvadrat_summy(3, 4) - summa_kvadratov(3, 4)
print(raznost)
```
