## Задание ★ Базовая практика — исправленные версии

```python
def sign(number):
    if number > 0:
        return "positive"
    elif number < 0:
        return "negative"
    else:
        return "zero"

def contains_even(numbers):
    for n in numbers:
        if n % 2 == 0:
            return True
    return False

s1 = sign(-5)
s2 = sign(0)
c1 = contains_even([1, 3, 5, 6])
c2 = contains_even([1, 3, 5])
print(s1, s2, c1, c2)
```
