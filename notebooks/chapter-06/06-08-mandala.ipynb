## Цель

Собрать все приёмы главы в одном узоре.

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
print("Окно Turtle готово.")
```

## Рабочий пример

```python
artist.reset()
shag_ugla = 10
ugol = 0
while ugol < 360:
    artist.setheading(ugol)
    artist.forward(150)
    artist.backward(150)
    ugol += shag_ugla
print("Мандала готова, лучей:", 360 // shag_ugla)
```

## Эксперимент 1 ★ — другой шаг угла

Попробуйте `shag_ugla = 30` — узор станет гораздо более разреженным.

```python
artist.reset()
shag_ugla = 30
ugol = 0
while ugol < 360:
    artist.setheading(ugol)
    artist.forward(150)
    artist.backward(150)
    ugol += shag_ugla
print("Лучей на этот раз:", 360 // shag_ugla)
```

## Эксперимент 2 ★★ — своя длина луча

Попробуйте `forward(80)` вместо 150.

```python
artist.reset()
shag_ugla = 10
dlina = 80
ugol = 0
while ugol < 360:
    artist.setheading(ugol)
    artist.forward(dlina)
    artist.backward(dlina)
    ugol += shag_ugla
print("Мандала меньшего размера готова.")
```

## Дополнительная задача ★★★

Попробуйте два цвета: чётные лучи — одним цветом, нечётные — другим (используем счётчик и `%` из главы 5).

```python
artist.reset()
shag_ugla = 10
ugol = 0
i = 0
while ugol < 360:
    artist.pencolor("purple" if i % 2 == 0 else "blue")
    artist.setheading(ugol)
    artist.forward(150)
    artist.backward(150)
    ugol += shag_ugla
    i += 1
artist.pencolor("black")
print("Двухцветная мандала готова.")
```
