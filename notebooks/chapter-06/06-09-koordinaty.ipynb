## Цель

Освоить систему координат окна Turtle и команду goto().

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

Переместимся в четыре разные точки, отмечая каждую точкой.

```python
artist.reset()
artist.penup()
for x, y in [(100, 0), (-100, 0), (0, 100), (0, -100)]:
    artist.goto(x, y)
    artist.dot(12, "#5B24F9")
artist.goto(0, 0)
```

## Эксперимент 1

Предскажите, в какой четверти экрана окажется точка (-150, -100), прежде чем запускать.

```python
artist.reset()
artist.penup()
artist.goto(-150, -100)
artist.dot(14, "#DB2777")
```

## Задание ★ Базовая практика

Поставьте точки в трёх вершинах треугольника: (0, 100), (-100, -100), (100, -100).

```python
artist.reset()
artist.penup()
for x, y in [(0, 100), (-100, -100), (100, -100)]:
    artist.goto(x, y)
    artist.dot(12, "#5B24F9")
```
