## Цель

Научиться предсказывать курс и координату черепашки на бумаге, не запуская окно.

## Рабочий пример

Черепашка стартует в (0, 0), курс 0° (восток). Команда forward(100) переместит её в (100, 0), курс останется 0°.

```python
start = (0, 0)
kurs = 0
# forward(100) при курсе 0° (восток) прибавляет 100 к x
finish = (start[0] + 100, start[1])
print(kurs, finish)
```
