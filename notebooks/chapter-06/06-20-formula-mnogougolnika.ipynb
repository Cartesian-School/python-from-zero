## Цель

Считать угол поворота для правильного многоугольника с произвольным числом сторон.

## Рабочий пример

```python
n = 6
povorot = 360 / n
print(povorot)
```

## Задание ★ Базовая практика

Посчитайте угол поворота для девятиугольника (n=9) и двенадцатиугольника (n=12). Выведите оба числа через пробел одним print().

```python
povorot_9 = 360 / 9
povorot_12 = 360 / 12
print(povorot_9, povorot_12)
```
