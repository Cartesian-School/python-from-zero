## Цель

Различать позицию и курс черепашки; связать градусы с направлениями сторон света.

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

Четыре курса из одной точки.

```python
artist.reset()
for angle in [0, 90, 180, 270]:
    artist.setheading(angle)
    artist.forward(80)
    artist.backward(80)
```

## Эксперимент 1

Предскажите курс после этих команд, затем проверьте heading().

```python
artist.reset()
artist.left(45)
artist.right(90)
print(artist.heading())
```

## Задание ★ Базовая практика

Поверните черепашку так, чтобы её курс стал строго 225°, используя любые left()/right() от начального курса 0°, и выведите artist.heading().

```python
artist.reset()
artist.left(225)
print(artist.heading())
```
