## Цель

Применить формулу поворота 360/n к разным многоугольникам.

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

Треугольник — поворот 360/3 = 120°.

```python
artist.reset()
for _ in range(3):
    artist.forward(100)
    artist.right(120)
```

## Эксперимент 1

Предскажите угол поворота для восьмиугольника (n=8), затем проверьте.

```python
artist.reset()
povorot = 360 / 8
for _ in range(8):
    artist.forward(60)
    artist.right(povorot)
```

## Задание ★ Базовая практика

Нарисуйте правильный десятиугольник (n=10), вычислив угол поворота по формуле 360/n (не вписывайте число вручную).

```python
artist.reset()
n = 10
povorot = 360 / n
for _ in range(n):
    artist.forward(50)
    artist.right(povorot)
```
