## Цель

Смоделировать случайное блуждание (random walk) и увидеть роль random.seed().

## Настройка (выполнить один раз)

```python
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.pensize(3)
artist.pencolor("#5B24F9")
print("Окно Turtle готово.")
```

## Рабочий пример

Случайное блуждание с фиксированным seed.

```python
import random

artist.reset()
random.seed(1)
for _ in range(30):
    artist.forward(15)
    artist.right(random.randint(-60, 60))
```
